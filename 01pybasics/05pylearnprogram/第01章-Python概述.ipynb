## 第1章 Python概述

- 计算机编程的本质是将现实生活中自动执行的动作解构为计算机能理解的逻辑，简单来说就是指示计算机用一种它能理解的语言完成某个任务。
- **Python中所有数据都由对象或对象间的关系表示**。任何对象都有属性和方法。属性是对象的特征，方法是对象的行为。每个对象包含四个部分：**ID(标识符)、类型、值、引用计数**。
    - ID：唯一标识符(可用`id()`查看)，用于标识对象，创建时自动生成，并且不会改变。
    - 类型：标识对象是什么类型(可用`type()`查看)，用于确定对象的属性和方法。
    - 值：对象的值，用于存储对象的数据。有些是可变对象，有些是不可变对象。
    - 引用计数：对象的引用计数(可用`sys.getrefcount()`查看)，用于确定对象是否被使用。
- Python代码的组织形式：
    - 语句：是最基本的代码单位。
    - 函数：用于封装某个任务，方便重复使用。
    - 类：用于创建自定义对象，方便组织代码。
    - 模块：是一个.py文件，用于组织代码。
    - 包：是一个文件夹，用于组织模块。文件夹中可以包含一个`__init__.py`文件(Python3.3及以上不是必须)，用于初始化模块。
- Python中的名字空间(作用域)：**使用缩进来定义代码所属的作用域，搜索名称时的顺序为：局部、外层、全局、内置(LEGB)**。
    - 局部作用域(local scope)：最内层，函数内部，包含局部名称。
    - 外层作用域(enclosing scope)：外部函数内部，包含外部函数的局部名称。
    - 全局作用域(global scope)：模块级别，包含了全局名称。
    - 内置作用域(builtin scope)：包含了Python内置的函数和异常。
