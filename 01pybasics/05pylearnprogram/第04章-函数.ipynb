## 第4章 函数

### 1.函数的定义

- 语法：
    ```python
    def 函数名(参数1, 参数2, ...):
        函数体
        return 返回值
    ```
- 说明：
    - 函数名：函数的标识符，用于调用函数，命名规则与变量相同。
    - 参数1, 参数2, ...：函数的输入参数，可以没有也可以有多个，用于接收外部数据。
    - 函数体：函数的执行代码，用于实现函数的功能。如果没有具体的实现代码，用`pass`语句占位。
    - return：用于返回函数的执行结果。可以没有也可以有多个。
    - 返回值：函数的输出结果。如果函数没有return语句，默认返回None。函数可以返回多个值，用逗号隔开(自动打包成元组)。

### 2.函数的作用域

- 函数作用域遵循`LEGB`规则：**L(局部作用域) -> E(外部作用域) -> G(全局作用域) -> B(内置作用域)**。
- `global`语句：在局部作用域定义全局变量。
- `nonlocal`语句：在内部函数中定义外部变量。

````python
def outter():
    test = 1

    def inner():
        nonlocal test               # 定义外部变量
        test = 2                    # 为外部变量赋值
        print(f'inner: {test}')     # 2

    inner()
    print(f'outter: {test}')        # 2，外部变量test被修改为2

test = 0                            # 定义全局变量
outter()
print(f'global: {test}')            # 0，全局变量test未被修改
````

### 3.函数的参数

- 参数规则：
    - 参数传递本质，就是将对象赋值给函数的局部变量名。
    - 在函数内给参数名重新赋值，不会影响调用方的原变量。
    - 在函数内修改可变对象的元素，会影响调用方的原对象。
- 参数名称：
    - 形参：函数定义中声明的名称。
    - 实参：函数调用时传入的实际值。
- 形参类型：
    - 位置形参：按参数定义的顺序传递，不能跳过或重复。
    - 关键字形参：按参数名传递，顺序可以任意，但是名称必须正确。
    - 默认形参(可选形参)：在函数定义时指定默认值，调用时可以不传递。
        - **特别注意：默认值在函数定义时创建一次，后续所有调用共享同一个对象。若默认值是可变对象(如列表、字典)，多次调用会累积修改。**
    - 可变位置形参(\*args)：可以传递0个或多个位置参数，所有位置参数都被打包成一个**元组**。
    - 可变关键字形参(\*\*kwargs)：可以传递0个或多个关键字参数，所有关键字参数都被打包成一个**字典**。
    - 仅位置形参(Python3.8+)：放在`/`**左侧**的参数只能通过位置传递，不能通过关键字传递。
    - 仅关键字形参：放在`*`或`*args`**右侧**的参数只能通过关键字传递，不能通过位置传递。
- 形参组合：
    - 顺序：**仅位置形参 → 普通位置/关键字形参 → 可变位置形参(\*args) → 仅关键字参数 → 可变关键字参数(\*\*kwargs)**。
    - 仅位置和普通参数中，必选参数必须在可选参数左侧。

````python
def func(a, b, c=7, *args, **kwargs):
    print('a, b, c:', a, b, c)              # a, b, c: 1 2 3 → a、b是位置形参，c是可选形参
    print('args:', args)                    # args: (5, 7, 9) → 可变数量的位置形参
    print('kwargs:', kwargs)                # kwargs: {'A': 'a', 'B': 'b'} → 可变数量的关键字形参
func(1, 2, 3, 5, 7, 9, A='a', B='b')


def allparams(a, /, b, c=42, *args, d=256, e, **kwargs):
    print('a, b, c:', a, b, c)              # a, b, c: 1 2 3 → a是仅位置形参，b是位置形参，c是可选形参
    print('d, e:', d, e)                    # d, e: 256 7 → d是仅关键字形参(没有传递值，使用默认值)，e是仅关键字形参
    print('args:', args)                    # args: (4, 5, 6) → 可变数量的位置形参
    print('kwargs:', kwargs)                # kwargs: {'f': 9, 'g': 10} → 可变数量的关键字形参
allparams(1, 2, 3, 4, 5, 6, e=7, f=9, g=10)
````

````python
def f1(p1, p2, /, p_or_kw, *, kw): pass
f1(1, 2, 3, kw=4)
f1(1, 2, p_or_kw=3, kw=4)

def f2(p1, p2=None, /, p_or_kw=None, *, kw): pass
f2(1, 2, 3, kw=4)

def f3(p1, p2=None, /, *, kw): pass
f3(1, 2, kw=4)

def f4(p1, p2=None, /): pass
f4(1)

def f5(p1, p2, /, p_or_kw): pass
f5(1, 2, p_or_kw=3)

def f6(p1, p2, /): pass
f6(1, 2)

# def f7(p1, p2=None, /, p_or_kw, *, kw): pass          # 错误, p_or_kw不能为位置形参
# def f8(p1=None, p2, /, p_or_kw=None, *, kw): pass     # 错误，p1可选形参不能在位置形参p2之前
# def f9(p1=None, p2, /): pass                          # 错误，p1可选形参不能在位置形参p2之前
````

### 5.导入模块

- 绝对导入：
    - import 模块名
    - import 模块名 as 别名
    - from 模块名 import 函数名
    - from 模块名 import 函数名 as 别名
    - from 模块名 import *
- 相对导入：
    - from . import 模块名
    - from .. import 模块名
    - from .模块名 import 函数名
