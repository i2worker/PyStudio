## 第3章 迭代和决策

### 1.条件

- 语法：`if 条件: 代码块 elif 条件: 代码块 else: 代码块`。
- 三元运算符：`值1 if 条件 else 值2`。当条件为真时，返回值1，否则返回值2。

````python
# 优惠券折扣
# 使用字典作为分派器，防止使用if-else语句的重复代码
customers = [
    dict(id=1, total=200, coupon_code='F20'),  # F20：固定，20
    dict(id=2, total=150, coupon_code='P30'),  # P30：百分比，30%
    dict(id=3, total=100, coupon_code='P50'),  # P50：百分比，50%
    dict(id=4, total=110, coupon_code='F15'),  # F15：固定，15
]

discounts = {
    'F20': (0.0, 20.0),  # 每个值是(percent, fixed)
    'P30': (0.3, 0.0),
    'P50': (0.5, 0.0),
    'F15': (0.0, 15.0),
}

for customer in customers:
    code = customer['coupon_code']
    percent, fixed = discounts.get(code, (0.0, 0.0))  # type: ignore
    customer['discount'] = percent * customer['total'] + fixed

for customer in customers:
    print(customer['id'], customer['total'], customer['discount'])
````

### 2.循环

- for循环：`for 变量 in 可迭代对象: 代码块 else: 代码块`。else语句是可选的，用于在循环正常结束时(没有被break语句中断)执行代码。
    - range()函数：用于生成一个整数序列，从0开始，到指定的结束值(不包含结束值)。
    - enumerate()函数：用于将可迭代对象(如列表、元组等)转换为一个索引序列，同时包含索引值和元素值。
    - zip()函数：用于将多个可迭代对象打包成一个元组序列，每个元组包含来自每个可迭代对象的对应元素。以最短的可迭代对象为准，其他可迭代对象的元素数量会截断。
- while循环：`while 条件: 代码块 else: 代码块`。else语句是可选的，用于在循环正常结束时(没有被break语句中断)执行代码。
- break语句：用于跳出整个循环。
- continue语句：用于跳过当前循环迭代，继续下一次迭代。


````python
# 将十进制数转换为二进制数
n = 79
result = []
while n > 0:
    n, remainder = divmod(n, 2)
    result.append(remainder)
print(result[::-1])
````

````python
# 找出指定范围内的所有素数，使用for循环和else子句
from math import sqrt
primes = []
upto = 100
for n in range(2, upto + 1):
    for d in range(2, int(sqrt(n)) + 1):
        if n % d == 0: break
    else:
        primes.append(n)

    # 使用any()函数
    # if any(n % d == 0 for d in range(2, int(sqrt(n)) + 1)): continue
    # primes.append(n)

print(primes)
````

### 3.海象运算符

- 语法：`变量 := 表达式`。将表达式的值赋值给变量，同时返回表达式的值。

````python
# 海象运算符
flavors = ["pistachio", "malaga", "vanilla", "chocolate", "strawberry"]
prompt = "Choose your flavor: "
print(flavors)
while (choice := input(prompt)) not in flavors:
    print(f"Sorry, '{choice}' is not a valid option.")
print(f"You chose '{choice}'.")
````

### 4.itertools模块简介

- `itertools.count(start, step)`：用于生成一个从指定起始值start开始，每次增加step值的无限序列。
- `itertools.compress(data, selectors)`：生成来自data在selectors中对应元素为真值的元素，当data或selectors可迭代对象被耗尽时将停止。
- `itertools.permutations(iterable, r=None)`：生成iterable中所有可能的r个元素的排列，r为None时生成所有排列。
