## 第2章 内置的数据类型

- 数值：
    - 整数：`int`，Python中的整数拥有无限大的范围，仅仅受限于可用的虚拟内存。
    - 浮点数：`float`，Python中的浮点数是64位双精度。如果要表示更精确的浮点数，可以使用`decimal`模块。
        - **注意：由于精度问题，`0.1 + 0.2`可能不是`0.3`，而是`0.30000000000000004`**。
        - `Decimal`从float创建会继承浮点误差，必须从整数或字符串创建才能保证精度。
    - 复数：`complex`，由实部和虚部组成，分别用`real`和`imag`属性表示。
    - 分数：`Fraction`，由分子和分母组成，分别用`numerator`和`denominator`属性表示。
    - 数值运算：`+`， `-`, `*`, `/`(浮点除法), `//`(向下整数除法), `%`(取余), `**`(指数)。
- 小值缓存：为提高性能，Python会缓存一些小值，如整数-5到256之间的整数。

````python
print(0.3 - 0.1 * 3)                    # -5.551115123125783e-17，因为精度问题，存在误差

from decimal import Decimal as D
print(D(0.1) * D(3) - D(0.3))           # 2.775557561565156540423631668E-17，仍存在误差
print(D('0.1') * D(3) - D('0.3'))       # 0.0，精确表示

from fractions import Fraction as F
print(F(1, 2) + F(1, 3))                # 5/6
f = F(10, 6)
print(f)                                # 5/3
print(f.numerator)                      # 5
print(f.denominator)                    # 3

a = 1000000; b = 1000000
id(a) == id(b)                          # False（大数字，不同对象）
a = 5; b = 5
id(a) == id(b)                          # True（小数字，缓存了同一个对象）
````

- 布尔值：
    - 真：`True`，非空值。
    - 假：`False`，`False`、`None`、`0`、`0.0`、`''`(空串)、`[]`(空列表)、`()`(空元组)、`{}`(空字典)、`set()`(空集)。
    - 布尔运算：`and`、`or`、`not`。


- 不可变序列：
    - 字符串：`str`，用引号`"`、单引号`'`、三引号`"""`、三单号`'''`括号括起来的不可变的字符序列。
        - 编码：`encode()`方法将字符串编码为字节序列(`bytes`)。`decode()`方法将字节序列解码为字符串。
        - 格式化：`%`格式化运算符、`format()`方法、`f字符串`。
    - 元组：`tuple`，元素可重复、不可变的有序序列，由括号`()`表示，用逗号`,`分隔元素。
        - 单元素元组必须加**逗号**：`(1,)` 才是元组，`(1)` 只是带括号的整数。

- 可变序列：
    - 列表：`list`，元素可重复、可变的有序序列，由方括号`[]`表示，用逗号`,`分隔元素。
    - 字节数组：`bytearray`，可变的字节序列。

- 集合类型
    - 集合：`set`，可变、无序、不重复的元素集合。用花括号`{}`表示，用逗号`,`分隔元素。
    - 冻结集：`frozenset`，不可变、无序、不重复的元素集合。

- 映射类型：
    - 字典：`dict`，键值对集合，由花括号`{}`表示，用冒号`:`分隔键值对，用逗号`,`分隔键值对。
        - Python3.6以上版本，保留键的插入顺序。


- 其他数据类型：
    - 日期和时间：在标准库中，用于处理日期和时间的主要模块是`datetime`、`calendar`、`zoneinfo`、`time`。
    - 其他集合类型：`collections`模块提供了其他集合类型，如`namedtuple`(命名元组)、`defaultdict`(默认字典)、`ChainMap`(链式映射)等。
    - 枚举：`enum`模块提供了枚举类型，Python3.4+引入，用符号名绑定唯一常量值。

````python
from collections import namedtuple, defaultdict, ChainMap

# 命名元组
Vision = namedtuple('Vision', ['left', 'right'])
vision = Vision(1, 2)
print(vision.left)          # 按字段名访问
print(vision.right)

# 默认字典
dd = defaultdict(int)   # 初始化默认字典，值为0
dd['age'] += 18                   # 解决键不存在的问题
print(dd['age'])
print(dd)

# 链式映射：快速合并多个映射，查找按顺序搜索，写入只操作第一个映射
dc = {'host': 'localhost', 'port': 4567}
cp = {'port': 5678}
conn = ChainMap(cp, dc)
print(conn['port'])                 # 5678，从第一个映射找到
print(conn['host'])                 # localhost，从第二个映射中找到
conn['host'] = 'packtpub.com'       # 写入第一个映射
del conn['port']                    # 删除第一个映射的port
print(conn['port'])                 # 4567，现在从第二个映射取

# 枚举
from enum import Enum
class Color(Enum):
    RED = 1
    GREEN = 2
    BLUE = 3
print(Color.RED)
print(Color.RED.value)
print(Color.RED.name)
````

**全部内置类型速查表**

| 类型                     | 可变？   | 有序？     | 可哈希？        | 典型用途         |
| ------------------------ | -------- | ---------- | --------------- | ---------------- |
| `int` / `float` / `bool` | ❌ 不可变 | —          | ✅ 是            | 数值计算         |
| `str`                    | ❌ 不可变 | ✅ 是       | ✅ 是            | 文本处理         |
| `tuple`                  | ❌ 不可变 | ✅ 是       | 元素都可哈希则✅ | 异构数据、字典键 |
| `bytes`                  | ❌ 不可变 | ✅ 是       | ✅ 是            | 二进制文本       |
| `frozenset`              | ❌ 不可变 | ❌ 否       | ✅ 是            | 不可变集合       |
| `list`                   | ✅ 可变   | ✅ 是       | ❌ 否            | 同构数据集合     |
| `dict`                   | ✅ 可变   | ✅ 是(3.7+) | ❌ 否            | 键值对映射       |
| `set`                    | ✅ 可变   | ❌ 否       | ❌ 否            | 去重、集合运算   |
| `bytearray`              | ✅ 可变   | ✅ 是       | ❌ 否            | 可变二进制数据   |
