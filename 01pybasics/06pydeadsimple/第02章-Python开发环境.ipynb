## 第2章 Python开发环境

- 一个优秀的Python开发环境通常包含：语言解释器、pip包管理器、虚拟环境、面向Python的代码编辑器，以及一个或多个用于检查代码错误与问题的静态分析器。

- Python语言解释器：
    - Linux安装：`sudo apt install python3 python3-pip python3-venv python3-dev`。
    - 指定为执行文件：文件开头添加 `#!/usr/bin/env python3`，自动在虚拟环境或系统路径中寻找Python3解释器。
    - 指定文件编码：Python文件默认是UTF-8编码，如果需要指定其他编码，可以在文件第一行或第二行添加 `# -*- coding: 编码名 -*-`。

- 虚拟环境：
    - 创建：`python3 -m venv venv`。
        - 如果想在虚拟环境中使用系统中的包：`python3 -m venv --system-site-packages venv`。
    - 激活：
        - Windows：`venv\Scripts\activate`。
        - Linux：`source venv/bin/activate`。
    - 退出：
        - Windows：`venv\Scripts\deactivate`。
        - Linux：`deactivate`。

- pip包管理器：
    - 安装包：`pip install 包名`。
        - 指定版本：`pip install 包名==版本号`。
        - 通过清单文件安装：`pip install -r requirements.txt`。
    - 更新包：`pip install --upgrade 包名`。
        - 通过清单文件更新：`pip install --upgrade -r requirements.txt`。
    - 卸载包：`pip uninstall 包名`。

- 静态分析工具：
    - pylint：用于检查Python代码的风格、错误和潜在问题。
        - 可以在项目根目录创建一个pylintrc文件来控制Pylint的行为：`pylint --generate-rcfile > pylintrc`。
    - flake8：PyFlakes(语法检查)、pycodestyle(风格检查)、mccabe(复杂度检查)等多个工具的组合。
        - 可以使用配置文件.flake8，或在tox.ini / setup.cfg中配置\[flake8\]区块，来控制flake8的行为。
    - mypy：专注于类型注解的静态类型检查器。

- 自动格式化工具：
    - autopep8：基于pycodestyle，自动修复代码使其符合PEP8。可传递`--aggressive`参数进行更深入的优化。
    - black：被誉为“不妥协的格式化工具”，配置极简，追求代码风格的绝对统一。

- 测试工具：
    - unittest：Python的标准库，用于编写和运行测试用例。
    - pytest：一个基于unittest的测试框架，提供了更简洁的语法和更多的功能。

- 代码编辑器：
    - PyCharm：最受欢迎的Python IDE，社区版免费，专业版付费。功能全面集成，包括自动补全、调试、重构、VCS 集成、自带静态分析等功能。
    - VSCode：第二受欢迎的编辑器，免费开源，跨平台，配合官方Python扩展功能强大，可集成Pylint、Flake8、Mypy等。
