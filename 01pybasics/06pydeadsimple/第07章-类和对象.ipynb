## 第7章 类和对象

- **类的概念：**
    - 类：创建对象的蓝图，定义属性和方法的模板。
    - 对象/实例：类的具体实现，每个实例都有自己的属性和方法。
    - 属性：对象的特征，用于存储数据。
    - 方法：对象的行为，用于执行操作。
    - 继承：一个类继承并扩展另一个现有类(is-a关系)。
    - 组合：一个对象包含其他对象(has-a 关系)。


- **声明类：**
    - 语法：`class 类名(父类名1, 父类名2, ...):`。在Python中，一切皆对象，一切都继承自`object`类。
    - 初始化函数：`__init__(self, 参数1, 参数2, ...)`。当创建一个新实例时，初始化函数会自动调用，**一般用于初始化对象的实例属性**，禁止返回任何值。
    - 构造函数：`__new__(cls, *args, **kwargs)`。当创建一个新实例时，首先调用构造函数，然后调用初始化函数。通常不需要定义构造函数，Python会自动提供，返回一个新的对象实例。
    - 析构函数：`__del__(self)`。当对象被销毁时调用，用于释放资源。只有在对象本身由垃圾回收器清理时，析构函数才会被调用。

- **属性**：所有属于类或实例的变量都称为属性​。
    - 类属性：属于类本身的属性称为类属性，有时也称为类变量。类属性声明通常放在类的顶部，而不是在方法中，使用`类名.属性名`来访问。类属性是所有实例共享的属性，而不是每个实例的独立属性。
        - ⚠️ 通过实例对类属性赋值会**创建实例属性**，遮蔽同名类属性，不影响其他实例。
    - 实例属性：属于实例本身的属性称为实例属性，有时也称为成员变量。实例属性通常在初始化函数中声明，使用`self.属性名`来访问。实例属性是每个实例独有的属性，对于其他实例不可用。

- **方法：**
    - 实例方法：属于实例本身的方法称为实例方法。实例方法的第一个参数(通常名为`self`)，提供对实例属性或方法的访问。
    - 类方法：属于类本身的方法称为类方法。使用`@classmethod`装饰器来装饰。类方法的第一个参数(通常名为`cls`)，提供对类属性或方法的访问。
    - 静态方法：定义在类中的普通函数，无特殊第一个参数，逻辑上与类相关但不访问类属性或实例属性。使用`@staticmethod`装饰器来装饰。

- **作用域命名约定**：Python没有正式的**数据隐藏**概念，而是通过命名约定实现。
    - 公共：常规命名方式，可安全外部修改/访问。
    - 私有：以下划线`_`开头，是社交契约，约定仅在类内部使用，外部不建议直接访问。
    - 名称改写：以双下划线`__`开头，自动改写为`_类名__属性名`，防止被子类遮蔽，提供弱数据隐藏，名称改写并不是真正的数据隐藏，仍可以通过`_类名__属性名`来访问。

- **特性**：一种特殊的实例方法，可以让实例方法像属性一样调用。
    - 定义方法一：`特性名 = property(fget=方法名, fset=方法名, fdel=方法名)`
    - 定义方法二：`@property`、`@特性名.setter`、`@特性名.deleter`。

````python
class SecretAgent:

    _codeword = None                        # 类属性(所有实例共享，约定为私有属性)

    def __init__(self, codename):           # 初始化方法
        self.codename = codename            # 实例属性(约定为公共属性)
        self._secrets = []                  # 实例属性(约定为私有属性)

    def __del__(self):                      # 析构方法
        print(f"{self.codename} deleted.")

    def remember(self, secret):             # 实例方法
        self._secrets.append(secret)

    @staticmethod
    def inquire():                          # 静态方法
        print("I know nothing.")

    @classmethod
    def inform(cls, codeword):              # 类方法
        cls._codeword = codeword

    @classmethod
    def _encrypt(cls, message, *, decrypt=False):
        if cls._codeword is None:
            raise ValueError("No codeword has been informed.")
        code = sum(ord(c) for c in cls._codeword)
        if decrypt:
            code = -code
        return ''.join(chr(ord(m) + code) for m in message)

    @property                               # 特性(getter)
    def secret(self):
        return self._secrets[0] if self._secrets else None

    @secret.setter                          # 特性(setter)
    def secret(self, value):
        self._secrets.append(self._encrypt(value))

    @secret.deleter                         # 特性(deleter)
    def secret(self):
        self._secrets.clear()

mouse = SecretAgent("Mouse")
mouse.inform("Parmesano")

print(mouse.secret)                         # 调用特性(getter)
mouse.secret = "12345 Main Street"          # 设置特性(setter)
print(mouse.secret)

print(mouse._secrets)                       # 直接访问特性与之关联的实例有属性
del mouse.secret                            # 删除特性(deleter)
print(mouse._secrets)
````

- **特殊方法：**
    - 转换方法：
        - `__repr__`：将对象转换为规范的字符串表示。
        - `__str__`：将对象转换为更友好的字符串表示。
        - `__hash__`：返回对象的哈希值，通常用于字典键。
    - 比较方法：
        - `__eq__`：定义对象的相等性比较。
        - `__ne__`：定义对象的不相等性比较。
        - `__gt__`：定义对象的大于比较。
        - `__ge__`：定义对象的大于等于比较。
        - `__lt__`：定义对象的小于比较。
        - `__le__`：定义对象的小于等于比较。
    - 运算符方法：
        - `__add__`：定义对象的加法运算。
        - `__sub__`：定义对象的减法运算。
        - `__mul__`：定义对象的乘法运算。
        - `__div__`：定义对象的除法运算。
        - `__mod__`：定义对象的取余运算。
        - `__pow__`：定义对象的指数运算。
    - 让对象可调用：
        - `__call__`：定义对象像函数一样调用的行为。

- **类装饰器**：与函数装饰器相同，接受一个类，返回一个增强后的类或替代品。

````python
# 不带参数的类装饰器模板代码: 装饰器 → 包装器
from functools import wraps
def decorator(cls):
    @wraps(cls)
    def wrapper(*args, **kwargs):
        print("在实例化前执行的代码")
        instance = cls(*args, **kwargs)
        print("在实例化后执行的代码")
        return instance
    return wrapper

@decorator
class MyClass:
    pass
````

````python
# 带参数的类装饰器模板代码: 装饰器工厂 → 装饰器 → 包装器
from functools import wraps
def factory(arg):
    def decorator(cls):
        @wraps(cls)
        def wrapper(*args, **kwargs):
            print(arg)
            instance = cls(*args, **kwargs)
            return instance
        return wrapper
    return decorator

@factory("hello")
class MyClass:
    pass
````

- **对象的结构模式匹配**：Python 3.10+ 支持在模式匹配中使用对象属性。

````python
class Pizza:
    def __init__(self, topping, second_topping=None):
        self.first = topping
        self.second = second_topping

order = Pizza("pepperoni", "mushroom")

match order:
    case Pizza(first='pepperoni', second='mushroom'):       # 匹配具体属性
        print("ANSI standard pizza")
    case Pizza(first='pineapple'):                          # 匹配部分属性
        print("Is this even pizza?")
    case Pizza(first=first, second='cheese'):
        print(f"Very cheesy pizza with {first}.")
    case Pizza(first=first, second=second):                 # 兜底匹配
        print(f"Pizza with {first} and {second}.")
````

````python
class Point:
    __match_args__ = ('x_pos', 'y_pos', 'z_pos')            # 定义匹配参数

    def __init__(self, x, y, z):
        self.x_pos = x
        self.y_pos = y
        self.z_pos = z

point = Point(0, 123, 0)

match point:
    case Point(0, 0, 0):                            # 全部匹配
        print("You are here.")
    case Point(0, _, 0):                            # 部分匹配
        print("Look up!")
````

- **何时使用类：**
    - 数据是否有凝聚力：属性是否共同描述一个事物？
    - 方法是否直接相关：方法是否都是该对象能做的事？
    - 避免上帝类：一个类只做一件事(单一职责原则)。
    - 类属性和类方法提供静态类模式，是跨模块共享状态的推荐方式。


- **核心知识脉络**

```text
对象与类
│
├── 1. 类与对象基础⭐
│   ├── 类 = 蓝图，对象 = 实例
│   ├── 封装、组合（has-a）、继承（is-a）
│   └── Python 3 隐式继承 object
│
├── 2. 初始化函数⭐
│   ├── __init__(self, ...) 定义实例属性初始值
│   ├── self 引用当前实例
│   ├── 所有实例属性应在 __init__() 中声明
│   ├── 构造函数 __new__(cls) 几乎永远不需要写
│   └── __init__() 禁止返回值
│
├── 3. 析构函数⭐
│   ├── __del__(self) 垃圾回收时调用
│   ├── del 只删名字绑定，不删值本身
│   └── ⚠️ 垃圾回收行为是实现特定的
│
├── 4. 属性⭐
│   ├── 实例属性 vs 类属性
│   │   ├── 实例属性：__init__ 内 self.x，每个实例独立
│   │   └── 类属性：类体内方法外，所有实例共享
│   ├── 实例赋值类属性 → 创建实例属性，遮蔽类属性
│   └── 命名约定
│       ├── 公有：attribute
│       ├── 非公有：_attribute（社交契约）
│       └── 名称改写：__attribute → _ClassName__attribute
│           └── ⚠️ 不是真正的数据隐藏
│
├── 5. 方法⭐
│   ├── 实例方法：首参 self，访问实例 + 类属性
│   ├── 类方法：@classmethod，首参 cls，访问类属性
│   │   └── 💡 常用于替代构造器（工厂方法）
│   └── 静态方法：@staticmethod，无特殊参数，不访问实例/类属性
│
├── 6. 特性⭐⭐
│   ├── 避免裸 getter/setter（反模式）
│   ├── property() 函数方式
│   │   └── prop = property(fget=..., fset=..., fdel=...)
│   ├── @property 装饰器方式（推荐）⭐
│   │   ├── @property → getter
│   │   ├── @prop.setter → setter
│   │   └── @prop.deleter → deleter
│   └── 省略 fget → 不可读；省略 fset → 不可写；省略 fdel → 不可删
│
├── 7. 特殊方法⭐⭐
│   ├── __repr__()：规范字符串（应能重建对象）
│   ├── __str__()：用户可读字符串（未定义时回退 __repr__）
│   ├── 比较方法：__eq__ / __lt__ / __le__ / __gt__ / __ge__
│   │   └── ⚠️ 不支持时返回 NotImplemented（不是 NotImplementedError）
│   ├── __hash__()：必须与 __eq__() 同时定义
│   ├── 二元运算符：__add__ / __sub__ / __mul__ / __truediv__ 等
│   ├── 反射运算符：__radd__ / __rsub__ 等
│   ├── 原地运算符：__iadd__ / __isub__ 等（+= / -=）
│   ├── 一元运算符：__neg__ / __pos__ / __abs__ / __invert__
│   └── __call__()：使实例可调用（obj()）
│
├── 8. 类装饰器⭐
│   ├── 装饰器包装类实例化
│   ├── 不带参数：decorator(cls) → wrapper(*args, **kwargs)
│   └── 带参数（三层嵌套）：factory(args) → decorator(cls) → wrapper(*args, **kwargs)
│
├── 9. 结构模式匹配⭐
│   ├── 按属性值匹配：case Pizza(first='pepperoni')
│   ├── 捕获模式：case Pizza(first=first, second=second)
│   └── __match_args__：启用位置匹配
│       └── 💡 dataclasses 中自动定义
│
└── 10. 何时使用类⭐
    ├── 数据有凝聚力？方法直接相关？
    ├── 单一职责原则（避免上帝类）
    └── 静态类共享状态 > 全局变量
```

- **警告与提示表**

| 类型 | 内容                                                         |
| ---- | ------------------------------------------------------------ |
| ⚠️    | 实例属性**应在 `__init__()` 中声明**，不要在其他方法中随意添加，避免运行时意外错误 |
| ⚠️    | `__init__()` **禁止返回任何值**，否则抛出 `TypeError`        |
| ⚠️    | `del obj` 只删除**名字绑定**，不删除值本身；实例仍被其他引用时不会触发 `__del__()` |
| ⚠️    | 对类属性的实例赋值会**创建实例属性**，遮蔽类属性，不影响其他实例——容易引发难以追踪的bug |
| ⚠️    | **名称改写不是真正的数据隐藏**——仍可通过 `_ClassName__attr` 访问 |
| ⚠️    | 定义 `__hash__()` 时**必须同时定义 `__eq__()`**，相等对象必须有相同哈希值 |
| ⚠️    | 比较方法中返回 `NotImplemented`（不是 `NotImplementedError`），让 Python 处理反射运算 |
| ⚠️    | 类装饰器若需参数，是**三层嵌套**：`factory(args)` → `decorator(cls)` → `wrapper(*args, **kwargs)` |
| ⚠️    | 结构模式匹配中**属性名必须匹配**，否则需定义 `__match_args__` 启用位置匹配 |
| ⚠️    | `__del__()` 的调用时机是实现特定的，不要依赖其精确行为       |
| 💡    | 所有实例属性在 `__init__()` 中声明，让类的数据结构一目了然   |
| 💡    | 优先用 `@property` 装饰器方式定义特性，可读性远高于 `property()` 函数方式 |
| 💡    | 类方法 `@classmethod` 常用于**替代构造器（工厂方法）**——通过不同方式创建实例 |
| 💡    | 静态方法适合与类逻辑相关但不访问实例/类数据的工具函数        |
| 💡    | `__repr__()` 应包含所有重建实例所需数据；未定义 `__str__()` 时 Python 回退到 `__repr__()` |
| 💡    | `__call__()` 让对象拥有函数行为，同时保留状态——本质是带状态的闭包 |
| 💡    | 跨模块共享状态用静态类（类属性+类方法），优于全局变量        |
| 💡    | dataclasses 中 `__match_args__` 自动定义，可直接用位置模式匹配 |
