## 第3章 语法速成课程

- **语句与表达式：**
    - 语句：每一行以换行结束的代码(不需要特殊字符)，也称为简单语句。
        - 复合语句由一个称为标题的代码行和一个或多个子句组成，并用**缩进**而非花括号`{}`来表示代码块的嵌套关系。
            - 可以省略换行符和缩进，让标题代码行和子句代码行在同一行。如：`if x > 0: print(True)`。
        - 若需将多个语句放在同一行，可用分号`;`分隔。
    - 表达式：求值为单个值的代码片段。表达式计算为一个值，然后这个值可以用在语句中。

- **注释和文档字符串：**
    - 注释：用`#`开头，直到行末。Python中没有多行注释，只能逐行注释。
    - 文档字符串：用`"""`或`'''`括起来的字符串，允许包含多行文本。通常放在函数、类、模块等的开头，用于描述其用途和参数。

- **变量和常量：**
    - 变量：
        - Python没有用于声明新变量的关键字，用变量名直接赋值即可。
        - **在使用变量之前必须先定义变量**，否则将会报错。
        - **不要更改变量中存储的数据的类型**，即使是替换值也如此。
    - 常量：
        - Python没有任何正式定义的常量。
        - 通常使用全大写的名称和下画线`_`来指示变量作为常量使用。

- **数学操作：**
    - 数据类型：
        - 整数(int)：在Python中，整数始终有符号，实际上没有最大值。整数默认使用十进制基数，但也可以指定为二进制(0b101010)、八进制(0o52)、十六进制(0x2A)。
        - 浮点数(float)：存储带有小数部分的数字，也可以使用科学记数法(如2.49e4)​。可以使用`float("nan")`指定无效数字，`float("inf")`表示正无穷大​，`float("-inf")`表示负无穷大。
        - 复数类型(complex)：在值的后面附加`j`来存储虚数，再使用加号`+`将实部与虚部结合起来，如`24+42j`。
        - Decimal类型：存储小数点固定的十进制数字，如`Decimal("0.333")`。
        - Fraction类型：存储分数，如`Fraction(1, 3)`。
    - 数据运算：
        - 简单运算符：+、-、*、/(浮点数除法)、//(向下整除法)、%(求余)、**(指数运算)。
        - 赋值运算符：+=、-=、*=、/=、//=、%=。

- **逻辑操作：**
    - 条件语句：`if 条件: 代码块 elif 条件: 代码块 else: 代码块`。
        - 三元运算符：`name = value_if_true if condition else value_if_false`。
    - 比较运算符：==、!=、>、<、>=、<=。
    - 逻辑运算符：and、or、not。注意短路求值。
    - 布尔值：True、False。常量None、表示零的值，以及空的集合都是False，而大多数其他值都是True。
        - **is运算符**只用于与`None`直接进行比较，其他情况下使用常规比较运算符。在实践中，通常使用if spam或if not spam，而不是直接将其与True或False进行比较。
    - 海象运算符(`:=`)：在Python3.8中引入，用于在赋值时同时计算表达式的值。如：`if (a := 10) > 5: print(a)`。注意必须加入括号，否则会出现逻辑错误。


- **字符串：**
    - 字符串字面量：可以使用单引号`'`、双引号`"`或三重引号`"""`或`'''`定义字符串。注意：三重引号中的所有内容，包括换行符和前导空格，都是字面量。
    - 原始字符串：可以使用`r`或`R`前缀定义原始字符串，原始字符串中的转义字符不会被解释为特殊字符。
    - 格式化字符串：可以是`f`字符串、`%`格式化字符串、`.format`方法来格式化字符串。
    - 字符串转换：`str()`函数返回字符串的可读表示形式，`repr()`函数返回字符串的原始表示形式(即Python看到的形式)。
    - 字符串拼接：首选`f-字符串`，不适用时再考虑使用`join()`方法。在实践中，f-字符串是最快的，join()是次佳选择。


- **函数：**
    - 在Python中，函数是“一等公民”​，可以像对待其他任何对象一样对待它们。
    - 定义：`def 函数名(参数列表): 函数体`。

- **类和对象：**
    - 在Python中，“一切皆对象”​，包括函数、类、模块等。
    - 定义：`class 类名: 类体`。

- **异常处理：**
    - 语法：`try: 代码块1 except 异常类型: 代码块2 else: 代码块3 finally: 代码块4`。    


- **元组和列表：**
    - 元组：不可变的有序集合，用括号`()`表示。
    - 列表：可变的有序集合，用方括号`[]`表示。
    - 指导原则：对于不同类型的元素集合(异构集合)使用**元组**，对于相同类型的元素集合(同构集合)使用**列表**。

- **循环：**
    - while循环：`while 条件: 循环体`。
    - for循环：`for 变量 in 可迭代对象: 循环体`。
    - 循环控制：`break`(终止循环)、`continue`(跳过当前循环，继续下一次循环)。


- **结构模式匹配：**
    - 语法：`match 表达式: case 模式1: 代码块1 case 模式2: 代码块2 ... case 默认: 代码块n`。
    - 文本模式和通配符：`case "字符串": 代码块`，匹配指定字符串后执行代码块。`case _: 代码块`，匹配任意值后执行代码块，必须放在最后面作为默认模式。
    - 或模式：`case 模式1 | 模式2: 代码块`，匹配指定模式中的任意一个后执行代码块。
    - 捕获模式：`case 变量: 代码块`，类似于通配符模式，只是将未匹配到的值赋值给指定变量。
    - 约束模式：`case 变量 if 条件: 代码块`，在匹配到指定变量后，判断是否满足指定条件，若满足则执行代码块。

````python
# 文本模式和通配符模式
launch = 'pizza'
match launch:
    case 'pizza':               # 文本模式，依次比较，找到匹配后执行代码块，不再检查后续模式
        print("Pizza time!")
    case _:                     # 通配符模式，匹配任何值，必须放在最后
        print("Yummy.")

# 或模式
launch = 'soup'
match launch:
    case 'salad' | 'soup':      # 或模式，匹配指定模式中的任意一个后执行代码块
        print('Eating healthy, eh?')
    case _:
        print("Yummy.")


# 捕获模式
launch = 'sandwich'
match launch:
    case 'salad' | 'soup':
        print('Eating healthy, eh?')
    case order:                 # 完整捕获模式，将未匹配到的值赋给指定变量
        print(f"Enjoy your {order}.")

# 捕获模式
lunch = ['vanilla', 'ice cream']
match lunch:
    case (flavor, 'ice cream'): # 部分捕获模式，将序列中未匹配到的值赋给指定变量
        print(f"Here's your very grown-up {flavor}...lunch.")
    case _:
        print("Yummy.")

# 约束模式
lunch = 'vanilla ice cream'
match lunch:
    case ice_cream if 'ice cream' in ice_cream:  # 约束模式，添加额外条件，只有条件满足时才匹配
        flavor = ice_cream.replace('ice cream', '').strip()
        print(f"Here's your very grown-up {flavor}...lunch.")
    case order:
        print(f"Enjoy your {order}.")
````

- **警告与提示表**

| 类型 | 内容                                                         |
| ---- | ------------------------------------------------------------ |
| ⚠️    | **缩进绝不能混用Tab和空格**——会导致 `TabError`               |
| ⚠️    | `#` 在字符串内部是普通字符，不会变成注释                     |
| ⚠️    | 不要用三引号字符串当"多行注释"——解释器会把它当有效字符串     |
| ⚠️    | docstring必须放在模块/函数/类/方法的**最顶部**才被识别       |
| ⚠️    | `is` 比较对象ID而非值，不要用于值比较（仅与None比）          |
| ⚠️    | 字典的 `in` 判断的是**键**而非值                             |
| ⚠️    | f-string表达式中**不能用反斜杠**，需提前计算                 |
| ⚠️    | `range()` 左闭右开——stop不包含                               |
| ⚠️    | `zip()` 在最短序列处停止，多余元素**静默丢弃**               |
| ⚠️    | 循环 `else` 仅在**未被break**时执行（语义反直觉）            |
| ⚠️    | 海象运算符 `:=` 通常需要括号                                 |
| ⚠️    | 捕获模式中裸变量名会被当作**捕获**，不是使用其值             |
| ⚠️    | 模式匹配在复杂场景下迅速达到极限，不确定时用if/elif          |
| 💡    | 不确定运算符优先级时，**加括号**永远最安全                   |
| 💡    | 交换变量无需临时变量：`a, b = b, a`                          |
| 💡    | 同时获取商和余数 → `divmod(a, b)`                            |
| 💡    | 优先用 `enumerate()` 而非 `range(len(...))`                  |
| 💡    | 判断布尔值优先用隐式写法：`if spam:` 而非 `if spam is True:` |
| 💡    | 值比较用 `!=` 而非 `not ==`（可读性更好）                    |
| 💡    | 类型提示不影响运行时，仅用于静态检查                         |
| 💡    | 三元运算符仅用于简单二值赋值，复杂逻辑用完整if/else          |
