## 第6章 函数和匿名函数

- **函数式编程**：函数式编程围绕**纯函数**进行组织，纯函数没有副作用或状态，并且每个函数只执行一个任务。函数式编程规则：
    - 每个函数都应该只做一件特定的事情；
    - 一个函数的实现方式不应该影响程序其他部分的行为；
    - 尽量避免副作用；
    - 函数不应该有状态或受外部状态的影响。

- **递归函数**：在定义中直接或间接调用自身的函数。递归还会影响程序性能。必须小心使用递归！
    - 条件：有基线条件(终止递归的出口，否则会无限递归)；每次递归调用必须将问题规模缩小，逐步逼近基线条件。
    - 递归调用：从最深层到最外层依次返回。
    - 递归深度限制：Python默认递归深度为1000，超过后会报错。可通过`sys.setrecursionlimit(n)`调高，但需谨慎——设置过高可能导致C层面的栈溢出。

- **参数**：函数定义时声明的变量叫**形参**，函数调用时传递的给形参的数据叫**实参**。
    - 位置参数：根据函数定义时的参数顺序传递。位置参数必须在所有关键字参数之前。
    - 关键字参数：根据函数定义时的参数名传递。
    - 默认参数值：在函数定义时给参数赋值指定默认值。
        - <strong style="color:red;">默认参数值只在函数定义时计算一次，之后所有调用共享同一个对象。永远不要使用可变类型作为默认参数值，而要使用None作为哨兵值。</strong>
    - 可变位置参数(\*args)：参数名前加`*`，多余的位置参数被打包成**元组**，必须出现在所有普通位置参数之后，之后的参数只能通过关键字传递。
    - 可变关键字参数(\*\*kwargs)：参数名前加`**`，多余的关键字参数被打包成**字典**。
    - 仅位置参数：函数定义中`/`之前的参数只能通过位置传递。
    - 仅关键字参数：函数定义中`*`之后的参数只能通过关键字传递。
    - 函数定义顺序：`def func(pos_only=None, /, pos_kw=None, *, kw_only=None)`。


| 顺序 | 参数类型                     | 分隔符              | 调用方式                 |
| ---- | ---------------------------- | ------------------- | ------------------------ |
| 1    | 仅限位置参数                 | `/` 之前            | 只能位置                 |
| 2    | 普通位置/关键字参数          | `/` 之后、`*` 之前  | 位置或关键字             |
| 3    | `*args`（可变位置参数）      | —                   | 收集多余位置参数为元组   |
| 4    | 仅限关键字参数               | `*` 或 `*args` 之后 | 只能关键字               |
| 5    | `**kwargs`（可变关键字参数） | 最后                | 收集多余关键字参数为字典 |

- **闭包**：是一个嵌套函数，它"包裹"了一个或多个非局部名字。当外层函数返回内层函数时，内层函数就变成了闭包。这种模式充当一种"函数工厂"。
    - 闭包中的递归问题：利用"默认参数在函数定义时求值"的特性，将外层变量的值快照到内层参数的默认值中。注意：这只适用于不可变类型。
    - 闭包的状态：通过`nonlocal`语句来修改闭包中的变量，保存闭包的状态。通常除非没有其他方案，否则应避免使用。

- **lambda函数**：是一种匿名函数，由单个表达式组成，语法：`lambda [参数列表]: 表达式`。最常见的用法之一是作为键函数使用。
    - 只能包含单个返回表达式——不能包含 if/for/while/try 等语句。
    - 复杂逻辑必须用 `def` 写正式函数。
    - lambda函数适合简短、清晰的逻辑片段。


- **装饰器**：是一种在函数定义前添加额外功能的语法糖。语法模板：

    ```python
    def decorator(func):
        @functools.wraps(func)                  # 保留原函数元信息
        def wrapper(*args, **kwargs):           # 装饰器函数，接受任意参数
            print("函数调用前执行的代码")
            result = func(*args, **kwargs)      # 调用原函数
            print("函数调用后执行的代码")
            return result                       # 返回原函数结果
        return wrapper                          # 返回装饰器函数

    @decorator                                  # 通过@语法糖应用装饰器函数
    def test():
        pass
    ```

装饰器关键要点：

| 要点                           | 说明                                                         |
| ------------------------------ | ------------------------------------------------------------ |
| **装饰器本质**                 | 一个闭包，包裹被修改的可调用对象                             |
| **`func` 参数**                | 被修改的可调用对象                                           |
| **`wrapper`**                  | 装饰器逻辑所在，可以是任何可调用对象                         |
| **`*args, **kwargs`**         | 不知道会有多少参数，所以用可变参数                           |
| **`@functools.wraps(func)`**  | 保留原函数元信息                                        |
| **`return result`**          | **必须返回原函数结果**！否则返回值被吞掉                     |
| **`@` 语法糖**                 | `@decorator` ≡ `test = decorator(test)` |
| **可堆叠**                     | 多个装饰器按顺序逐层包裹，从内到外依次执行                     |

- **类型提示**：在变量、函数参数或返回值前添加类型信息，用于类型检查和文档生成。
    - 提示：Python本身不会对这些类型提示采取任何行动，而只会将这种符号系统识别为有效，并存储在函数的`__annotations__`属性中。
    - 变量类型提示：`name: str`。
    - 函数类型提示：`def func(a: int = 1, b: str = 'hello') -> bool: pass`。

- **核心知识脉络**

```text
函数与匿名函数
│
├── 一、纯函数
│   ├── 定义：无副作用，基于输入返回新输出
│   ├── 函数是一等对象：可传递、返回、赋值
│   └── 函数式编程两条规则：1)只做一件事 2)不影响外部
│
├── 二、递归
│   ├── 函数调用自身，需基础条件终止
│   ├── Python递归深度上限约1000（实际≈997）
│   ├── sys.setrecursionlimit(n) 可调高，但需谨慎
│   └── ⚠️ 许多Pythonista认为递归是反模式
│
├── 三、默认参数值 ⚠️⚠️
│   ├── 可选参数，无默认值的必须在有默认值之前
│   ├── 可变默认参数陷阱：只计算一次，所有调用共享
│   └── 解决方案：用 None 做哨兵值，函数体内创建新对象
│
├── 四、位置参数与关键字参数
│   ├── 位置参数：按顺序映射
│   ├── 关键字参数：按名称映射，消除歧义
│   └── 规则：位置参数必须在关键字参数之前
│
├── 五、可变参数
│   ├── *args：收集多余位置参数为元组
│   ├── **kwargs：收集多余关键字参数为字典
│   └── 常见用法：参数转发 func(*args, **kwargs)
│
├── 六、仅限位置/关键字参数 ⭐
│   ├── / 之前 → 仅限位置参数（Python 3.8+）
│   ├── * 之后 → 仅限关键字参数
│   └── 完整顺序：pos_only / pos_kw *args kw_only **kwargs
│
├── 七、嵌套函数与 nonlocal
│   ├── 嵌套函数可读取外层变量
│   ├── nonlocal：修改外层函数变量（跳过Local）
│   └── global：修改全局变量（跳过Local+Enclosing）
│
├── 八、闭包 ⭐
│   ├── 函数工厂：返回"记住"外层变量的内层函数
│   ├── 递归闭包：用默认参数做快照，不用nonlocal
│   ├── 有状态闭包：nonlocal修改外层变量 ⚠️ 难调试
│   └── 规则：闭包中看到 nonlocal → 格外小心
│
├── 九、匿名函数
│   ├── 语法：lambda 参数: 表达式（单表达式）
│   ├── 适用场景：排序键函数、回调、短逻辑
│   ├── 限制：不支持语句、不支持类型提示
│   └── 复杂逻辑 → 必须用 def
│
├── 十、装饰器 ⭐⭐
│   ├── 本质：闭包包裹函数，增强行为
│   ├── @functools.wraps(func)：保留原函数元信息（必须！）
│   ├── wrapper 必须返回 result（必须！）
│   ├── @ 语法糖 ≡ func = decorator(func)
│   └── 可堆叠多个装饰器
│
└── 十一、类型提示
    ├── 可选注解，不影响运行时
    ├── 参数: 类型, 返回 -> 类型
    ├── Mypy 静态类型检查
    └── 鸭子类型与类型提示兼容
```

- **警告与提示**

| 类型       | 内容                                                         |
| ---------- | ------------------------------------------------------------ |
| ⚠️ 警告     | **默认参数值在函数定义时只计算一次**——可变类型作为默认值会导致所有调用共享同一个对象，用 `None` + 函数内检查替代 |
| ⚠️ 警告     | **递归深度默认约1000**，超过抛 `RecursionError`；调高限制需谨慎，可能导致C层级栈溢出 |
| ⚠️ 警告     | **闭包中看到 `nonlocal` → 格外小心**——它意味着有状态，可能导致调用间行为不一致 |
| ⚠️ 警告     | 有状态闭包调试困难——**外部无法查看捕获变量的当前值**，仅在无更好方案时使用 |
| ⚠️ 警告     | 装饰器内**必须** `@functools.wraps(func)` 保留原函数 `__name__`/`__doc__` 等元信息 |
| ⚠️ 警告     | 装饰器 wrapper **必须** `return result` 返回原函数结果       |
| ⚠️ 警告     | **lambda不支持类型提示**，且只能包含单个表达式               |
| ⚠️ 警告     | 参数默认值为 `None` 时，类型提示用 `typing.Optional[type]` 而非直接写 `type` |
| 💡 技巧     | 纯函数无副作用，是编写可测试、可维护代码的基础               |
| 💡 技巧     | 使用关键字参数消除歧义——牢记Python之禅"面对歧义，拒绝猜测"       |
| 💡 技巧     | `*args` 和 `**kwargs` 让函数接受任意数量参数，常用于参数转发 |
| 💡 技巧     | `/` 和 `*` 精确控制API调用方式，避免误用                     |
| 💡 技巧     | 递归闭包用**默认参数做快照**，而非 `nonlocal` 修改外层变量   |
| 💡 技巧     | lambda适合短逻辑（排序key、回调），复杂逻辑必须用 `def`      |
| 💡 技巧     | 装饰器是"函数工厂"，用 `@` 语法糖消除重复代码                |
| 💡 技巧     | 类型提示不影响运行时，配合Mypy可在编码阶段发现类型错误       |
