## 第5章 变量和数据类型

- **变量**：Python中，用名称(name)和值(value)来代替变量。​“变量”指代这两者的组合：一个name指向一个value。
    - 变量名：变量的名称，用来标识变量，指向一个值或对象。
    - 变量值：变量存储的值，内存中一个特定的数据实例。

- **赋值**：将一个名称绑定到一个值或对象。赋值不复制，而是指向同一个原始值。
    - is运算符：比较两个名称所绑定的值在内存中的位置是否相同。
    - ==运算符：比较两个名称所绑定的值是否相等。

````python
answer = 42
insight = answer
# 说明1：insight不是指向值42的副本，而是指向同一个原始值。 answer ──→ 42 ←── insight
# 说明2：insight不是绑定到answer，而是绑定到answer被赋值时已经绑定到的同一个值。
````

- **数据类型**：Python是一种动态、强类型语言。
    - `type()`函数：查看值的数据类型，也就是返回值属于哪个类的实例。常用is操作符配合使用：`if type(value) is 类名`。
    - `isinstance()`函数：查看值是否属于某个类的实例。比`type()`更好，因为它可以检查继承关系。

- **作用域**
    - <strong style="color:red;">名称有作用域，但是没有类型；值有类型，但是没有作用域。</strong>
        - 在Python中，只有**函数和推导式**创建了作用域，而**类和模块**创建了命名空间(每个直接声明在类中的名称都是这个类的属性​，通过类名或对象名使用点`.`运算符访问)。
    - 作用域解析顺序：按照LEGB原则解析。L(局部) -> E(嵌套) -> G(全局) -> B(内置)。
        - `global`语句：用于在函数内部声明全局变量。
            - 规则：如果想从局部作用域重新绑定全局名字时，必须先使用`global`关键字。如果只是**访问**全局名字的当前值，则不需要。
        - `nonlocal`语句：用于在嵌套函数中声明非局部变量。
            - 从最内层的嵌套作用域开始查找指定名字，如果没找到，向外层继续找，直到找到或确认该名字不存在于非全局的外层作用域中。

- **垃圾回收和引用计数器：**
    - 引用计数器：每个值都有一个引用计数，表示当前有多少个引用指向该值，每次值被绑定到一个名字时引用计数增加，当引用计数为0时，值将会被垃圾回收器回收。

- **可变与不可变数据类型：**
    - 不可变数据类型：包括`int`, `float`, `bool`, `str`, `tuple`, `frozenset`, `bytes`等，一旦创建就不能被修改。
        - 不可变对象："修改" = 创建新对象 + 重绑定。
    - 可变数据类型：包括`list`, `dict`, `set`, `bytearray`, 自定义类实例，创建后可以被修改。
        - 可变对象："修改" = 原地修改，ID不变。

- **函数传参：**
    - Python既不是值传递，也不是引用传递，**本质是"按赋值传递"，赋值永远不会复制数据**。
    - 对于不可变数据类型，函数参数会重新绑定到新值，修改不会影响外部变量。
    - 对于可变数据类型，函数参数会绑定到同一个对象，修改会影响外部变量。

- **集合与引用**：
    - <strong style="color: red;">集合中的元素也是引用，集合中的元素与变量的名称没有任何区别。</strong>

````python
board = [["-"] * 3] * 3
board[0][0] = "X"
print(board)
# 所有行都变了！
# ['X', '-', '-']
# ['X', '-', '-']
# ['X', '-', '-']
# 原因：外层的列表由3个列表组成，每个列表都是指向同一个对象，所以当修改其中一个列表时，其他列表也会被修改。

# 正确做法：
board = [["-"] * 3 for _ in range(3)]
board[0][0] = "X"
print(board)
````

- **复制：**
    - 浅复制：`copy.copy()`。浅复制**只复制对象本身**，不复制对象内部引用的其他可变值。
    - 深复制：`copy.deepcopy()`。递归复制所有嵌套的可变对象。

- **警告与提示表**

| 类型 | 内容                                                         |
| ---- | ------------------------------------------------------------ |
| ⚠️    | **赋值永远不复制数据**——`a = b` 只是别名，不是复制           |
| ⚠️    | 函数内**赋值**会自动创建局部变量，遮蔽同名全局变量——必须用 `global` 才能修改全局变量 |
| ⚠️    | 可变类型的别名修改互相可见——`highs = temps` 后修改一个另一个也变 |
| ⚠️    | 函数修改传入的可变参数会产生**副作用**——如 `temperatures.sort()` 修改原列表 |
| ⚠️    | `[row] * 3` 创建三个**别名**而非三个独立列表——井字棋经典Bug  |
| ⚠️    | 元组内的可变对象仍可被修改——元组的不可变仅限于自身结构       |
| ⚠️    | 浅复制不复制嵌套可变对象——`copy.copy()` 只复制顶层           |
| ⚠️    | **可变默认参数**被所有调用共享——函数/类的默认值只创建一次    |
| 💡    | 核心原则：**名字有作用域无类型，值有类型无作用域** |
| 💡    | 传参等价于赋值——Python是"按赋值传递"                         |
| 💡    | 可变默认参数用 `None` 做哨兵：`def func(a=None)`             |
| 💡    | 类型检查优先用 `isinstance()` 而非 `type()`                  |
| 💡    | 避免引用循环可用 `weakref` 弱引用                            |
