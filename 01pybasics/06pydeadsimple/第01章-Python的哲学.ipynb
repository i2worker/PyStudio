## 第1章 Python的哲学

- Python是一门由荷兰程序员Guido van Rossum于1991年开发的编程语言，被归类为**高级通用语言**，完整支持**过程式、面向对象和函数式**等多种编程范式，其设计哲学强调代码可读性。

- 外界对Python存在五个最普遍的错误认知：
    - Python仅仅是一种脚本语言。Python是图灵完备的，其他语言能做的事，Python也能做。
    - Python很慢。Python是一种解释型语言，它的运行速度确实比编译型语言慢。但是在大多数实际项目中，Python的速度完全足够。
    - Python无法编译。虽然 Python是解释型语言，但完全可以将Python编译成机器码。
    - Python在幕后被编译了。Python解释器会将代码转换为字节码，然后由解释器执行。绝大多数Python项目以源码或字节码形式分发，用户需安装解释器。
    - Python不适合大型项目。实际上，Python的项目结构比C++和Java简单得多。一旦理解了包、模块和导入系统，处理多文件代码变得轻而易举。

- Python之禅：
    - 美丽胜于丑陋。
    - 明了胜于晦涩。
    - 简单胜于复杂。
    - 复杂胜于凌乱。
    - 扁平胜于嵌套。
    - 稀疏胜于稠密。
    - 可读性至关重要。
    - 特殊情况也不足以违反规则。
    - 尽管实用胜于纯粹。
    - 错误不应被无声忽略。
    - 除非你刻意沉默。
    - 面对歧义，拒绝猜测的诱惑。
    - 应该有一种——而且最好只有一种——明显的做法。
    - 尽管那种做法一开始可能并不明显，除非你是荷兰人。
    - 现在做胜于永远不做。
    - 尽管永远不做往往胜于现在仓促地做。
    - 如果实现难以解释，那就是坏主意。
    - 如果实现容易解释，那可能是好主意。
    - 命名空间是一个绝妙的主意——让我们多使用它们吧！

- Python任何新功能最初都以Python增强提案(PEP)的形式出现，每个提案有唯一编号，发布在[https://peps.python.org](https://peps.python.org)目录下。
