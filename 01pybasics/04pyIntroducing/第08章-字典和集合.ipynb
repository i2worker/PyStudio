## 第8章 字典和集合

### 1.字典

- 字典是一种**可变**的、**无序**的**键值对**集合，字典项的顺序是任意的，字典项的键是唯一的，可以是**任意不可变类型**，而值可以重复，可以是任意类型。
    - 创建方式1：将以逗号`,`分隔的一系列`键:值对`放入花括号`{}`内。
    - 创建方式3：使用`dict()`函数将一个双值序列转换为字典，或者传入具名参数和值创建一个字典。

````python
dict1 = {'a': 1, 'b': 2, 'c': 3}
dict2 = dict(a=1, b=2, c=3, d=4)
dict3 = dict([('a', 1), ('b', 2), ('c', 3)])
dict4 = dict(['a1', 'b2', 'c3'])
````

- 字典操作：
    - 获取值：`dict[key]`，如果键不存在，则会抛出KeyError异常。
    - 获取值：`dict.get(key, default)`，如果键不存在，则返回默认值。
    - 修改并获取值：`dict.setdefault(key, default)`，如果键存在，则返回值；如果键不存在，则创建新键，返回默认值。
    - 修改值：`dict[key] = value`，如果键存在，则更新值，否则创建新键。
    - 删除键值对：`del dict[key]`，如果键存在，则删除键值对，否则会抛出KeyError异常。
    - 清空字典：`dict.clear()`。
    - 遍历键：`for key in dict.keys()`，返回字典所有键的视图，不是列表。
    - 遍历值：`for value in dict.values()`，返回字典所有值的视图，不是列表。
    - 遍历键值对：`for key, value in dict.items()`，返回字典所有键值对的视图，不是列表。
    - 获取键值对数量：`len(dict)`。
    - 合并字典：`dict.update(dict2)`，如果键存在，则更新值，否则创建新键。
    - 合并字典：`{**dict, **dict2}`，返回的是浅拷贝。
    - 按键删除并返回值：`dict.pop(key, default)`，如果键存在，则删除键值对，返回值；如果键不存在，则返回默认值。
    - 复制字典：`dict.copy()`，返回的是浅拷贝。如果需要深拷贝，需要使用`copy.deepcopy()`函数。
    - 比较字典：支持`==`和`!=`运算符，比较字典的键值对是否相等。
    - 字典推导式：`{键: 值 for 表达式 in 可迭代对象}`。

### 2.集合

- 集合是一种**可变**的、**无序**的、**不重复**的元素集合，并且元素必须是**不可变类型**。
    - 创建方式1：`set()`，将可迭代对象转换为集合，重复元素会被自动过滤，字典只使用键创建集合。
    - 创建方式2：`{元素1, 元素2, ...}`，直接创建集合，其中元素不能重复。
        - **注意：空集合的创建方式为`set()`，而不是`{}`。**
    - 创建不可变集合：`frozenset()`，将集合转换为不可变集合，即创建后不能修改集合中的元素。

- 集合操作：
    - 添加元素：`set.add(元素)`。
    - 删除元素：`set.remove(元素)`。
    - 清空集合：`set.clear()`。
    - 获取集合大小：`len(set)`。
    - 检查元素是否存在：`元素 in set`。

- 集合运算：
    - 交集：`set1 & set2`，或者`set1.intersection(set2)`，返回两个集合的交集，即同时存在于两个集合中的元素。
    - 并集：`set1 | set2`，或者`set1.union(set2)`，返回两个集合的并集，即存在于任意一个集合中的元素。
    - 差集：`set1 - set2`，或者`set1.difference(set2)`，返回两个集合的差集，即存在于第一个集合中而不在第二个集合中的元素。
    - 对称差集：`set1 ^ set2`，或者`set1.symmetric_difference(set2)`，返回两个集合的对称差集，即存在于任意一个集合中的元素，但不在两个集合都存在的元素。
    - 子集：`set1 <= set2`，或者`set1.issubset(set2)`，判断第一个集合是否为第二个集合的子集。真子集用`<`表示。
    - 超集：`set1 >= set2`，或者`set1.issuperset(set2)`，判断第一个集合是否为第二个集合的超集。真超集用`>`表示。

````python
s1 = set([1, 2, 3, 4])
s2 = {3, 4, 5, 6}

s1 & s2    # 交集　{3, 4}
s1 | s2    # 并集　{1, 2, 3, 4, 5, 6}
s1 - s2    # 差集　{1, 2}
s1 ^ s2    # 对称差集　{1, 2, 5, 6}
````
