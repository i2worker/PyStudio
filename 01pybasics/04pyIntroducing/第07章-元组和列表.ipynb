## 第7章 元组和列表


### 1.元组

- 元组是一种**不可变序列类型**，赋值后元素固定无法修改。元组可以存储多个值，**每个值可以重复、可以是任意类型**。
    - 创建方式1：使用圆括号`()`和逗号`,`，例如`(1, 2, 3)`。也可以省略圆括号，例如`1, 2, 3`。
        - 注意：只有一个元素的元组，必须在元素后面添加逗号，例如`(1,)`或`1,`。
    - 创建方式2：使用`tuple()`函数将其他序列类型转换为元组，例如`tuple([1, 2, 3])`。

- 元组解包：将元组中的元素分别赋值给多个变量，例如`a, b, c = (1, 2, 3)`。
    - 利用元组解包可以快速实现两个变量交换，例如`a, b = b, a`。

- 元组操作：
    - `+`：拼接元组，例如`(1, 2, 3) + (4, 5, 6) = (1, 2, 3, 4, 5, 6)`。
    - `*`：重复元组，例如`(1, 2, 3) * 2 = (1, 2, 3, 1, 2, 3)`。
    - `<`：比较元组，类似还有`>`、`<=`、`>=`，逐个元素比较，直到找到不同的元素或比较完所有元素。例如`(1, 2, 3) < (4, 5, 6)`。
    - `in`：检查元素是否在元组中，例如`1 in (1, 2, 3)`。
    - `not in`：检查元素是否不在元组中，例如`1 not in (1, 2, 3)`。

````python
one1 = 1,               # 元组 (1,)
one2 = (1,)             # 元组 (1,)
one3 = (1)              # 整数 1

(1, 2, 3) + (4, 5, 6)   # (1, 2, 3, 4, 5, 6)
(1, 2, 3) * 2           # (1, 2, 3, 1, 2, 3)
(1, 2, 3) > (1, 2)      # True
````

### 2.列表

- 列表是一种**可变序列类型**，可以就地修改列表，向其中添加新元素，删除或替换已有元素。在列表中，值可以是任意类型，同样的值可以多次出现。
    - 创建方式1：使用方括号`[]`和逗号`,`，例如`[1, 2, 3]`。
    - 创建方式2：使用`list()`函数将其他序列类型转换为列表，例如`list((1, 2, 3))`。

- 列表方法：
    - `append()`：在列表末尾添加一个元素。
    - `insert()`：在指定位置插入一个元素。
    - `extend()`：在列表末尾添加多个元素，类似于`+`运算符的连接操作。
    - `remove()`：删除第一个匹配的元素。
    - `pop()`：删除并返回指定位置的元素，默认删除最后一个元素。
    - `clear()`：清空列表。
    - `index()`：返回指定元素在列表的第一个索引位置。如果元素不在列表中，抛出ValueError异常。
    - `count()`：返回指定元素在列表中出现的次数。
    - `sort()`：对列表进行就地排序，该操作会修改原列表。Python3**不允许**不同元素类型之间直接排序。
    - `sorted()`：返回一个新的排序后的列表，不会修改原列表。
    - `reverse()`：将列表的元素顺序就地反转，该操作会修改原列表。
    - `len()`：返回列表的元素数量。
    - `copy()`：返回一个新的列表，该列表是原列表的浅拷贝。
        - `copy.deepcopy()`：返回一个新的列表，该列表是原列表的深拷贝。
    - `zip()`：将多个可迭代对象打包成一个元组序列，每个元组包含来自每个可迭代对象的对应元素。
        - 如果可迭代对象的长度不同，最短的可迭代对象会决定元组序列的长度。

- 列表推导式：`[表达式 for 变量 in 可迭代对象 if 条件]`，其中for...in...可以嵌套使用。
    - 注意：不存在元组推导式，`(number for number in range(10))`，是一个生成器表达式，会返回一个生成器对象，而不是一个元组。

### 3.元组与列表对比

| 特性     | 元组                   | 列表                                 |
| -------- | ---------------------- | ------------------------------------ |
| 语法     | `(1, 2, 3)`            | `[1, 2, 3]`                          |
| 可变性   | **不可变**             | **可变**                             |
| 添加元素 | ❌                      | `append()` / `insert()` / `extend()` |
| 删除元素 | ❌                      | `del` / `remove()` / `pop()`         |
| 修改元素 | ❌                      | ✅ `list[i] = new_value`              |
| 排序     | ❌                      | `sort()` / `sorted()`                |
| 字典键   | ✅ 可以                 | ❌ 不可以                             |
| 性能     | 略快                   | 略慢                                 |
| 内存     | 更小                   | 更大                                 |
| 推导式   | ❌（用 `tuple()` 包裹） | ✅ `[expr for ...]`                   |
| 复制     | 无需深复制             | `copy()` / `deepcopy()`              |

### 4.本章小结

```text
序列容器：元组与列表
    │
    ├── 元组 (tuple) — 不可变
    │     ├── 创建：() 逗号 / tuple()
    │     ├── ⚠️ 单元素必须加逗号 (1,)
    │     ├── 索引/切片：与字符串相同
    │     ├── 不可修改，但可拼接/重复（产生新元组）
    │     ├── 元组解包：a, b, c = tuple_ ⭐
    │     └── 用途：字典键、函数多返回值、常量集合
    │
    ├── 列表 (list) — 可变 ⭐
    │     ├── 创建：[] / list() / split()
    │     ├── 索引/切片：同字符串规则
    │     ├── 嵌套：列表中的列表
    │     │
    │     ├── 添加：append / insert / extend / +=
    │     │     └── ⚠️ append([1,2]) vs extend([1,2])
    │     │
    │     ├── 删除：del / remove / pop / clear
    │     │     └── del按位置 / remove按值 / pop按位置并返回
    │     │
    │     ├── 查找：index / in / count
    │     │
    │     ├── 排序：sort(原地) vs sorted(新列表)
    │     │     └── key=参数自定义排序
    │     │
    │     ├── 反转：reverse(原地) vs reversed(迭代器)
    │     │
    │     └── 复制：copy(浅) vs deepcopy(深) ⭐
    │           └── 浅复制：嵌套元素仍共享引用
    │
    ├── 列表推导式 [expr for item in iterable if cond] ⭐⭐
    │     ├── 带条件过滤
    │     ├── 多重循环
    │     └── 原则：一眼看不懂就改用普通循环
    │
    └── (expr for ...) = 生成器，不是元组推导式
```
