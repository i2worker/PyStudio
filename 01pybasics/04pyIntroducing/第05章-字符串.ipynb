## 第5章 字符串

### 1.字符串创建

- 字符串是字符序列，**是不可变的**，无法就地修改字符串。
- 字符串通过单引号`'`或双引号`"`创建，或者三个单引号`'''`或双引号`"""`创建**多行字符串**(其中的换行符、空白符等会被保留)，还通过`str()`函数创建。
    - `f''`：f字符串，用于格式化的字符串。
    - `r''`：原始字符串，字符串中出现的转义序列不会被解释。
    - `fr''`：原始f字符串，用于格式化的原始字符串。
    - `u''`：Unicode字符串，用于表示非ASCII字符。
    - `b''`：表示bytes类型的值。
- 字符串通过`\`进行转义，例如`\\n`表示换行符，`\\t`表示制表符，`\\r`表示回车符，`\\b`表示退格符，`\\f`表示换页符。

### 2.字符串操作

- 拼接：`+`运算符可以用于拼接字符串，也可以将两个字面字符串相邻放置实现拼接效果。
    - 换行拼接：如果有大量字符串，则可以将其放入圆括号`()`内，以此避免转义换行符。
- 重复：`*`运算符可以用于重复字符串。

````python
# 两个字符串前后放置实现拼接
s1 = 'My word! ' 'A gentleman caller!'
print(s1)

# 通过圆括号实现多行字符串的拼接
s2 = (
    'My word! '
    'A gentleman caller!'
)
print(s2)
````

### 3.字符串切片

- 语法结构：`s[start:stop:step]`，从start提取至end-1，每隔step个字符提取一个字符。
    - 范围规则：`start`为起始偏移(包含)，`end`为结束偏移(不包含)，省略`start`默认从0开始，省略`end`默认到字符串末尾。
    - 步长规则：`step`为可选参数，表示每隔step个字符取一个，正步长从左向右取，负步长从右向左取。
    - 越界处理：切片对越界偏移更宽容，不会抛出异常。**小于起始偏移会视为0，大于结束偏移会视为字符串末尾**。

````python
letters = 'abcdefghijklmnopqrstuvwxyz'

print(letters[:])               # 复制整个字符串
print(letters[20:])             # 从第20个字符开始提取到字符串结束
print(letters[12:15])           # 从第12个字符开始提取到第14个字符结束
print(letters[-3:])             # 提取最后3个字符
print(letters[18:-3])           # 从第18个字符开始提取到最后第4个字符结束
print(letters[-6:-2])           # 从最后6个字符开始提取到最后第3个字符结束
print(letters[-1::-1])          # 反向提取字符串
print(letters[::-1])            # 反向提取字符串
print(letters[-50:])            # 整个字符串，letters[0:]
print(letters[-51:-50])         # ''， letters[0:0]
print(letters[:70])             # 整个字符串，letters[0:-1]
print(letters[70:71])           # ''， letters[-1:-1]
````

### 4.字符串方法

- `len()`：返回字符串的长度。
- `split()`：将字符串按指定分隔符分割成子字符串列表。
- `join()`：使用指定分隔符将子字符串列表连接成一个字符串。
- `replace()`：替换字符串中的子字符串。
- `strip()`：移除字符串首尾的指定字符(默认空白字符)。`rstrip()`、`lstrip()`分别移除字符串右侧、左侧的指定字符(默认空白字符)。
- `startswith()`、`endswith()`：检查字符串是否以指定字符开头或结尾。
- `find()`：返回子字符串首次出现的索引，如果未找到则返回 **-1**。`rfind()`、`lfind()`分别从字符串右侧、左侧开始查找。
- `index()`：返回子字符串首次出现的索引，如果未找到则抛出**异常**。`rindex()`、`lindex()`分别从字符串右侧、左侧开始查找。
- `count()`：返回子字符串在字符串中出现的次数。
- `capitalize()`：将字符串的第一个字符转换为大写，其他字符转换为小写。
- `title()`：将字符串的每个单词的第一个字符转换为大写，其他字符转换为小写。
- `upper()`：将字符串转换为大写。
- `lower()`：将字符串转换为小写。
- `swapcase()`：将字符串中的大写字母转换为小写字母，小写字母转换为大写字母。
- `center()`：将字符串居中对齐，用指定字符填充(默认为空格)。`rjust()`、`ljust()`分别将字符串右对齐、左对齐。

### 5.字符串格式化

- **%**：`string % (value1, value2, ...)`。
- **format()**：`string.format(value1, value2, ...)`。
    - format()方法可以使用命名参数或位置参数。
- **f字符串**：`f"string {value1} {value2} ..."`。Python3.6以上版本支持。
    - 从Python3.8开始，如果想打印变量名及其值，可以把`=`放在f字符串的{}内的名称之后。

- **格式字符串**：
    - 起始冒号(':')。
    - 可选的填充字符(默认为'')，在不足最小宽度时作填充之用。
    - 可选的对齐字符，默认为左对齐。'<'表示左对齐，'>'表示右对齐，'^'表示居中对齐
    - 可选的数字符号，如果忽略，则表示只在负数前添加负号('-')​。
    - 可选的最小宽度。可选的点号（'.'）用于分隔最小宽度和最大字符数。
    - 可选的最大字符数。
    - 转换类型。

````python
thing = 'wereduck'
place = 'werepond'

# 直接打印变量值，并格式化
# ->6.3，表示右对齐，最大宽度为6，最小宽度为3，用-填充不足部分。
# #<5.2，表示左对齐，最大宽度为5，最小宽度为2，用#填充不足部分。
# 输出：thing[-4:] = ---duc, place.title() = We###
print(f'{thing[-4:] = :->6.3}, {place.title() = :#<5.2}')
````
