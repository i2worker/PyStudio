## 第6章 while循环和for循环

- **循环语句**：
    - `while`：while condition: ...。
    - `for`：for item in iterable: ...。
    - `break`：立即跳出当前整个循环。
    - `continue`：跳过当前循环，继续下次循环。
    - `else`：如果while/for循环正常结束(没有执行break)，则执行else语句。

- **range()函数**：range(start, stop, step)，返回指定范围内的系列数字。
    - `start`：起始值，如果忽略从0开始计数。
    - `stop`：结束值，**不包含在生成的序列中**。
    - `step`：步长，默认为1，也可以为负数。

- **本章小结**

```text
循环：让代码重复执行
    │
    ├── while 循环（条件循环）
    │     ├── while condition: ...
    │     ├── while True: ... break    （无限循环+break模式）
    │     └── while...else             （没被break→执行else）
    │
    ├── for 循环（遍历循环）
    │     ├── for item in iterable: ...
    │     ├── for...else               （没被break→执行else）
    │     └── 遍历：字符串、列表、字典(键)等
    │
    ├── 循环控制
    │     ├── break    — 终止整个循环
    │     └── continue — 跳过本次，继续下一轮
    │
    └── range() — 生成数字序列 ⭐
          ├── range(stop)
          ├── range(start, stop)
          ├── range(start, stop, step)
          ├── 语法与切片一致：start含，stop不含
          └── 惰性生成，节省内存
```
