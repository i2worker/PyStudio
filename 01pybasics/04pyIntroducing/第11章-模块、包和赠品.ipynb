## 第11章 模块、包和赠品

### 1.模块

- 模块就是一个包含Python代码的.py文件(或由其他语言编写的动态链接库)。模块可以定义函数、类、变量，也可以包含可执行代码。
- 导入模块：
    - 导入整个模块：`import module`，导入后使用模块名调用模块中的函数或变量(`module.function()`)。
    - 导入时使用别名：`import module as alias`，导入后使用别名调用模块中的函数或变量(`alias.function()`)。下面的导入方式也可以使用别名。
    - 导入模块中特定函数或变量：`from module import 函数名, 变量名`，导入后直接调用函数或变量，无需使用模块名。
    - 导入模块中**所有不以下划线开头**的函数或变量：`from module import *`，尽量避免使用，因为它会导入模块中的所有函数或变量，导致命名冲突。
- 导入语句放置位置：
    - 如果导入的代码可能会用在不止一处，应从函数外部导入，放在代码顶部。
    - 如果导入的代码只在函数内部使用，应在函数内部导入，同时可以实现延迟导入，提供启动速度。
- 搜索路径：Python会按以下路径顺序查找导入的模块名，**当前脚本目录 → 环境变量PYTHONPATH中列出的目录 → 标准库目录 → 第三方库目录**。
    - 这些路径存放在`sys.path`列表中。
    - 可以动态添加路径，如：`sys.path.insert(0, '/my/modules')`，将/my/modules目录添加到搜索路径的最前面优先查找。
- 相对导入：用于在包内部导入同一包下的其他模块，使用点号(`.`)表示当前目录，两个点号(`..`)表示父目录。
    - 当前目录导入：`from . import 模块名`，导入当前目录下的模块。
    - 子目录导入：`from .submodule import 模块名`，导入子目录下的模块。
    - 上一级目录导入：`from .. import 模块名`，导入上一级目录下的模块。
    - 相对导入的限制：
        - 只能在包内部使用，即模块必须属于一个包。
        - 当模块作为顶层脚本直接运行时，相对导入会失败，因为无法确定相对路径。
        - 要执行包内的脚本，应使用`python -m mypackage.module_b`方式运行。
- 重新加载模块：模块导入后会缓存在`sys.modules`中，修改源文件后再次导入不会生效。
    - 可以使用`importlib.reload(module)`函数强制重新加载模块，使模块中的代码生效。
    - **注意**：`reload()`函数不会自动更新已经引用的名称，需要手动更新。
- 模块与对象：
    - 模块中的所有类、函数和全局变量都可以在外部访问。对象可以使用属性和双下划线命名方式隐藏或控制对于其数据特性的访问。
    - 程序导入模块时，只会在程序运行时执行一次，后续调用时不会重复导入。类可以创建多个对象，每个对象都有自己的属性和方法。

````python
import sys
from pathlib import Path

sys.path.insert(0, fr"{Path.cwd()}\mod")    # 将当前目录下的mod目录添加到sys.path中

for path in sys.path: print(path)           # 其中出现的空白路径是当前目录

import fast             # 因为之前将mod目录添加到sys.path中，所以这里可以直接导入fast模块
print(fast.pick())
````

### 2.包

- 包是一种特殊的模块，是一个包含多个模块的目录。传统包是一个包含`__init__.py`文件的目录，Python3.3+允许没有`__init__.py`文件的命名空间包。
    - `__init__.py`文件的作用：
        - 标识该目录是一个Python包。
        - 可以在此文件中执行包级别的初始化代码。
        - 定义`__all__`列表控制`from package import *`的行为。

````python
from mod import fast as fa      # 从包中导入模块并使用别名
from mod import advice as ad

print("Let's go to", fa.pick())
print("Should we take out?", ad.give())
````

### 3.标准库

- collections：
    - defaultdict(): 用于创建一个字典，可以指定其中每个键的默认值。
    - Counter(): 用于计数可哈希对象。
    - deque(): 用于实现双端队列，支持从两端高效插入和删除元素。
- itertools：
    - chain(): 用于将多个迭代对象连接起来，形成一个迭代对象。
    - cycle(): 用于无限循环迭代一个可迭代对象。
    - accumulate(): 用于计算可迭代对象的累计和。
- pprint：
    - pprint(): 用于以格式化的方式打印对象，使输出更易读。
- random：
    - choice(): 用于随机选择一个元素。
    - randint(): 用于生成一个随机整数。
    - shuffle(): 用于随机打乱一个列表的元素。
    - sample(): 用于随机选择多个元素，不重复选择。
