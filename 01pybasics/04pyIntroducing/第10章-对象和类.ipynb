## 第10章 对象和类

### 1.基本概念

- 类：是创建对象的蓝图或模板，包含了该类对象所共有的属性(数据)和方法(行为)。
- 对象：根据类创建的具体实例，每个对象都有自己独立的属性值，并可以使用类中定义的方法。
    - 关系：类是对象的模板，对象是类的具体实例。
- 属性：对象的特征或状态，是对象的内部数据。
- 方法：对象的行为或操作，是对象的外部接口。
- 类与模块的区别：类可以同时存在多个实例，每个实例的属性可以不同；而模块全局仅有一个实例。

### 2.类的定义

- 语法：
    ```python
    class 类名:
        属性1 = 值1
        属性2 = 值2
        ...
        方法1(self):
            方法体1
        方法2(self):
            方法体2
        ...
    ```
- 说明：
    - 类名：类的唯一标识符，用于引用类。
    - 属性1, 值1, 属性2, 值2, ...：类的属性，用于存储对象的状态。
    - 方法1(self), 方法2(self), ...：类的方法，用于定义对象的行为。
    - self：方法的第一个参数，用于引用对象本身。
- 初始化：
    - 当创建对象时，会自动调用类的初始化方法(`__init__`)来设置对象的初始状态。
    - 可以在初始化方法中，定义对象的属性值，也可以在类的定义中直接赋值。

### 3.类的继承

- 语法：
    ```python
    class 子类名(父类名1, 父类名2, ...):
        属性1 = 值1
        属性2 = 值2
        ...
        方法1(self):
            方法体1
        方法2(self):
            方法体2
        ...
    ```
- 说明：
    - 子类名：子类的唯一标识符，用于引用子类。
    - 父类名1, 父类名2, ...：父类的唯一标识符，用于引用父类。
- 调用父类方法：
    - 子类中可以使用`super()`函数来调用父类的方法。
    - 语法：super().方法名()。
- 多重继承：
    - 每个Python类都有一个特殊方法`mro()`，返回一个类列表，用于查找该类对象的方法或特性。
    - 还有一个类似的特性`__mro__`，是由继承类组成的元组。

### 4.类的特性

- 在Python中，对象特性和方法通常是公开的，可以直接访问。
    - 可以约定下划线`_`开头的属性或方法视为私有，不建议外部直接访问。
    - 可以使用双下划线`__`开头的属性或方法，来表示更私有的属性或方法(需通过重整命名来访问，例如`实例._类名__属性名`)。
- getter和setter方法：
    - getter方法：用于获取对象的属性值。
    - setter方法：用于设置对象的属性值。
- 特性：
    - 特性：用于将方法转换为属性，使代码更简洁。
    - 语法1：`特性名 = property(getter, setter)`。
    - 语法2：`@property`装饰器放在getter方法上, `@特性名.setter`装饰器放在setter方法上。

### 5.类的属性和方法

- 类属性：
    - 定义方式：定义在类本身上的变量，直接写在类的主体中，不在任何方法内。
    - 访问方式：可以通过`类名.属性名`或`实例.属性名`访问。
    - 存储方式：属于类对象，在内存中只有一份，被该类的所有实例共享。
    - 修改行为：通过类名修改类属性，会影响所有实例。通过实例修改类属性，实际上不会修改类属性，而是动态创建一个同名的对象属性，从此该实例访问到的将是新创建的对象属性，而类属性本身不变。
    - 使用场景：适合存放所有实例共用的数据或常量。
- 对象(实例)属性：
    - 定义方式：通常通过`self.属性名`在`__init__`方法(或其他实例方法)中定义。  
    - 访问方式：只能通过实例访问(`实例.属性名`)，不能通过类名直接访问。
    - 存储方式：属于每个实例，每个实例拥有独立的一份副本，互不影响。
    - 修改行为：直接修改对象属性，不会影响其他对象实例的属性值。
    - 使用场景：适合存放每个实例特有的数据。

````python
class Student:
    school = "清华大学"          # 类属性

s1 = Student()
s2 = Student()

# 通过类名修改类属性 → 所有实例受影响
Student.school = "北京大学"
print(s1.school)                # 北京大学
print(s2.school)                # 北京大学

# 通过实例“修改”类属性 → 实际是创建了实例属性
s1.school = "复旦大学"           # 为s1创建了一个实例属性 school
print(s1.school)                # 复旦大学（实例属性覆盖了类属性）
print(s2.school)                # 北京大学（类属性）
print(Student.school)           # 北京大学（类属性）

del s1.school                   # 删除实例属性后，s1重新看到类属性
print(s1.school)                # 北京大学（类属性）
````

- 类的方法：
    - 实例方法：方法没有前置装饰器，第一个参数是`self`，引用对象本身。
        - 可以访问实例和类的属性。
        - 必须通过实例调用，或者通过类调用并手动传入实例。
        - 最常用的方法类型，用于处理每个对象特有的行为。
    - 类方法：方法有前置装饰器`@classmethod`，第一个参数是`cls`​，引用类本身。
        - 只能访问类的属性，不能访问实例属性。
        - 可以通过类或实例调用，都会自动传入类对象。
        - 常用于：替代构造函数，返回不同配置的实例；操作类级别的状态。
    - 静态方法：方法有前置装饰器`@staticmethod`，第一个参数不能是对象或类。
        - 不能访问实例属性，也不能直接访问类属性(除非用类名显式引用)。
        - 可以通过类或实例调用，行为与普通函数完全一致。
        - 常用于：工具函数，与类相关但不需要依赖类或实例的状态。

````python
class MyClass:
    count = '类属性'

    def __init__(self, name):
        self.name = name

    # 实例方法
    def instance_method(self):
        return f"实例方法, 访问实例属性：{self.name}，类属性：{self.count}"

    # 类方法
    @classmethod
    def class_method(cls):
        return f"类方法, 只能访问类属性：{cls.count}，不能访问实例属性"

    # 静态方法
    @staticmethod
    def static_method():
        return "静态方法, 与普通函数类似，不自动接收self或cls"

obj = MyClass('实例对象')

# 实例方法
print(obj.instance_method())            # 实例调用
print(MyClass.instance_method(obj))     # 类调用（手动传实例）

# 类方法
print(MyClass.class_method())           # 类调用
print(obj.class_method())               # 实例调用（自动传入类）

# 静态方法
print(MyClass.static_method())          # 类调用
print(obj.static_method())              # 实例调用
````

### 6.魔术方法

- 用于比较的魔术方法：`__eq__`(==)、`__ne__`(!=)、`__gt__`(>)、`__ge__`(>=)、`__lt__`(<)、`__le__`(<=)。
- 用于算术运算的魔术方法：`__add__`(+)、`__sub__`(-)、`__mul__`(*)、`__truediv__`(/)、`__floordiv__`(//)、`__mod__`(%)、`__pow__`(**)。
- 其他魔术方法：`__str__`(对象的字符串表示)、`__repr__`(对象的官方字符串表示)、`__len__`(对象的长度)、`__iter__`(可迭代对象)、`__next__`(迭代器的下一个元素)等。

### 7.具名元组

- 具名元组是Python中`collections`模块提供的一个工厂函数`namedtuple()`创建的轻量级、不可变的数据结构。
    - 语法：`namedtuple(类名, 字段名列表)` 或者 `namedtuple(类名, 使用空格分隔的字段名字符串)`。
    - 特性：
        - 与普通元组一样，创建后不能修改、添加或删除字段。
        - 既支持属性访问`.字段名`，也支持索引访问`[下标]`。
        - 支持元组的解包、迭代、切片、成员测试等操作。

````python
from collections import namedtuple

# 创建具名元组
Person = namedtuple('Person', ['name', 'age'])  # 使用列表
Person = namedtuple('Person', 'name age')       # 使用空格分隔的字符串

# 创建实例
p1 = Person(name='张三', age=30)     # 使用命名参数
p2 = Person('李四', 25)              # 使用位置参数

# 访问方式
print(p1.name)       # 通过属性访问
print(p2[1])         # 通过索引访问
````

### 8.数据类

- 数据类是Python 3.7+通过标准库`dataclasses`模块提供的一个装饰器`@dataclass`，用于自动生成常见特殊方法，从而大幅减少编写简单数据容器类的样板代码。
    - 语法：`@dataclass`。
    - 自动生成的方法：`__init__`、`__repr__`、`__hash__`、`__eq__`等比较方法。
    - 可变性：数据类是可变的，可以修改字段值。如果要创建不可变的类，需要使用`frozen=True`参数。

````python
from dataclasses import dataclass

# 创建数据类
@dataclass
class Point:
    x: float
    y: float
    z: float = 0.0  # 指定默认值为0.0

# 实例化数据类
p1 = Point(1.0, 2.0)
p2 = Point(3.0, 4.0, 5.0)

print(p1)        # Point(x=1.0, y=2.0, z=0.0)
print(p1 == p2)  # False
````
