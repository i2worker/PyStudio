## 第2章 数据：类型、值、变量和名称

- **对象**：Python将每个数据值(包括整数、浮点数、字符串、布尔值，甚至大型数据结构、函数和程序)，都包装为内存中的**对象**，每个对象至少包含以下内容：
    - 类型：定义对象能做什么。`type()`函数可以查看对象的类型。`isinstance()`函数可以检查对象是否是某种类型的实例。
    - 标识符：区分其他对象的唯一标识符。`id()`函数可以查看对象的标识(内存地址)。
    - 数据值：与其类型一致的数据。
    - 引用计数：跟踪该对象被使用的频率。`sys.getrefcount()`函数可以查看对象的引用计数。

````python
import sys
x = 500000
print(type(x))                  # <class 'int'>
print(isinstance(x, int))       # True
print(id(x))                    # 对象标识(内存地址)
print(sys.getrefcount(x))       # 引用计数，注意：通过getrefcount()函数调用会增加引用计算，所以实际引用计数比实际值大1
````

- **可变性**：类型还决定了对象包含的值是可以改变的(可变 mutable)，还是恒定不变的(不可变 immutable)。
    - 不可变对象：创建后**值不能被修改**的类型。像一个密封的透明盒子，你能看到内部的值，但不能改变它。"修改"不可变对象的唯一方式是**创建新对象**。
    - 可变对象：创建后**值可以被原地修改**的类型。像一个带盖子的盒子，你不仅能看到值，还能改变它；但你不能改变它的类型。

- **值**：在Python中，有两种指定数据值的方式：字面值和变量。
    - 字面值：直接在代码中指定的值，包括：数字序列(整数)、包含小数点的数字(浮点数)、引号包围的文本(字符串)、布尔值(True/False)等。
    - 变量：已赋值的变量，例如a、b等。

- **变量**：在程序中为了引用内存中的值而为其取的**名称**。
    - <strong style="color: red;">变量仅仅是名称而已，名称是对事物的引用，而非事物本身。</strong>
    - <strong style="color: red;">Python是动态类型语言，变量名称可以指代任何事物，通过“顺着绳子找盒子”来获取值和类型。</strong>
    - 变量命名规则如下：
        - 只能包含：小写字母、大写字母、数字(0-9)、下划线(_)。
        - 区分大小写.
        - 必须以字母或下划线开头，不能以数字开头。
        - 以下划线开头的名称有特殊待遇。

- **赋值**：使用`=`将右侧的值赋给左侧的变量。**赋值操作不会复制值，而只是为包含数据的对象附加一个名称**。
    - 初始化要求：右侧必须是有值的(已初始化的字面值或变量)，如果使用未初始化的变量，Python会抛出异常。
    - 执行逻辑：
        - 在内存中创建右侧的对象。
        - 创建左侧的名字。
        - 将名字**引用/绑定**到对象。
    - 重新赋值：只是让名称指向新的对象，旧对象的引用计数减少了，新对象的引用计数增加了。

- **复制**：将变量`a`赋值给变量`b`，仅仅是让`b`指向了`a`所指的同一个对象。
    - 不可变对象的情况：如果对象是不可变的(如整数)，改变其中一个变量的指向(如`a = 29`)，只会让该变量指向新对象，不影响另一个仍指向原对象的变量(如`b`依然指向5)。
    - 可变对象的陷阱：如果两个名称指向同一个可变对象(如列表)，通过任何一个名称修改对象的值，另一个名称访问时也会看到改变后的值。因为修改的是盒子里的内容，而两个标签仍然系在同一个盒子上。
