## 第3章 数字

### 1.布尔值

- 在Python中，任何值都可解释为布尔值，包括`True`和`False`，等价于1和0。`bool()`函数能将任意的Python数据类型转换为布尔值。
    - 真值：非零的数被视为`True`。
    - 假值：值为0的数被视为`False`，包括0、None、空字符串、空列表、空元组、空字典、空集合等。**注意：空序列、空映射虽然为假值，但不等于False**。


````python
print(True + False + 2)         # 3
print([] == False)              # False
print(bool([]) == False)        # True
````

### 2.整数

- 整数为无小数、无分数的全数字类型，还可能包括前缀符号(正负号、多进制)。禁止以0开头后接1-9的数字。
    - 数位分隔符：用下划线`_`表示，不是逗号`,`(这样将变成元组)。
    - 基数：`0x / 0X`表示16进制，`0o / 0O`表示8进制，`0b / 0B`表示2进制。
    - 范围：在Python3中，`int`支持**任意精度**，无整数溢出，可表示超大数。

### 3.浮点数

- 浮点数是带小数点的数字，可以在字母`e`之后包含一个十进制数形式的指数，还可以使用下划线`_`分隔数位。
    - 若需高精度小数计算，可使用`decimal`模块，避免浮点数精度误差。

### 4.算术运算符

- Python中，算术运算符主要包括：`+`、`-`、`*`、`/`、`//`(整除)、`%`(取余)、`**`(指数)。
    - `/`：浮点数除法，最后结果总是浮点数。
    - `//`：向下取整法，注意负数情况，向负无穷方向取整。如：`10 // 3 = 3`，`-10 // 3 = -4`。
    - `divmod()`函数：返回一个元组，包含商和余数。
    - <strong style="color: red;">注意：浮点数运算时，需要考虑精度问题，永远不要用 `==` 比较两个浮点数是否相等。如：`0.1 + 0.2 == 0.3`返回False。</strong>
    - 对象比较：`is`比较对象id(身份)，`==`比较对象值；Python内部会缓存`[-5, 256]`范围内的整数，这些整数的id相同。

### 5.类型转换

- Python是**动态强类型语言**：动态指运行时检查数据类型，强类型指未经强制转换时变量类型不会改变。
- `int()`函数：将其他类型的数据转换为整数，还可以指定基数。浮点数转整数时直接截断小数部分，不做四舍五入。要四舍五入用`round()`函数。
    - `bin()`函数将整数转换为2进制字符串，`oct()`函数转换为8进制字符串，`hex()`函数转换为16进制字符串。
    - `chr()`函数将整数转换为与其对等的单字符字符串，`ord()`函数将单字符字符串的ASCII码转换为整数。
- `float()`函数：将其他类型的数据转换为浮点数。
- `str()`函数：将其他类型的数据转换为字符串。
- `bool()`函数：将其他类型的数据转换为布尔值。

    │
    ├── 浮点型 (float)
    │       ├── 科学计数法：1.0e8
    │       ├── 精度陷阱：0.1 + 0.2 ≠ 0.3 ← ⭐
    │       ├── 混合运算自动提升：bool→int→float
    │       └── 解决方案：decimal模块 / 容差比较

    └── 类型转换全景
            bool() / int() / float() / str() 互相转换
            关键：int(98.6)=98（截断），round(98.6)=99（四舍五入）
