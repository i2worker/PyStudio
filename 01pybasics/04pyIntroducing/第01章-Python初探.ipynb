## 第1章 Python初探

Python是一种动态类型语言，具有如下优势：

- **易读易学**：语法接近英语，代码即文档。
- **简洁**：完成同样功能，Python代码通常比C/Java短得多。
- **"自带电池"**：标准库极为丰富，开箱即用。
- **跨平台**：Windows、macOS、Linux均可用。
- **免费开源**：无许可费用，社区活跃。
- **多范式**：面向对象、函数式、过程式编程。
- **社区生态**：PyPI上有海量第三方包，支持扩展功能。

同时，Python也有其不足之处，例如：

- **对运行速度要求极高的场景**：如高频交易、实时3D渲染、底层驱动等。
- **移动端开发**：Python不是iOS/Android的主流开发语言。
- **内存受限的嵌入式系统**：Python解释器占用的资源相对较多。


**Python之禅**：`import this`。

- Beautiful is better than ugly.                                            美胜于丑
- Explicit is better than implicit.                                         显胜于隐
- Simple is better than complex.                                            简胜于繁
- Complex is better than complicated.                                       繁胜于杂
- Flat is better than nested.                                               扁平胜于嵌套
- Sparse is better than dense.                                              稀疏胜于密集
- Readability counts.                                                       可读性很重要
- Special cases aren't special enough to break the rules.                   特殊情况不足以打破规则
- Although practicality beats purity.                                       但实用性胜过纯粹性
- Errors should never pass silently.                                        错误不应被静默忽略
- Unless explicitly silenced.                                               除非被显式地静默
- In the face of ambiguity, refuse the temptation to guess.                 面对歧义，拒绝猜测
- There should be one--and preferably only one--obvious way to do it.       应有一种(最好只有一种)显而易见的方式
- Although that way may not be obvious at first unless you're Dutch.        虽然除非你是荷兰人，否则那方式一开始可能不明显
- Now is better than never.                                                 现在胜于永不
- Although never is often better than *right* now.                          但永不往往胜过"立刻"
- If the implementation is hard to explain, it's a bad idea.                难以解释的实现是坏主意
- If the implementation is easy to explain, it may be a good idea.          容易解释的实现可能是好主意
- Namespaces are one honking great idea--let's do more of those!            命名空间是个绝妙主意，多用！
