## 第4章 if语句

### 1.注释

- 注释是程序中会被Python解释器忽略的文本片段，可用于说明代码逻辑、记录待办事项、标注版权信息等，不会对程序运行产生任何影响。
- Python使用`#`字符标记注释，从#开始到当前行结束的部分都属于注释。可以把注释作为单独的一行，也可以把注释写在代码的后面。

- Python使用续行字符`\`(反斜线)放在行尾，表示后续内容仍在同一行。
    - 隐式续行：如果将代码放在括号`()`、方括号`[]`、花括号`{}`等内部，Python会自动续行，无需加`\`。建议尽量使用隐式续行。

````python
# 用\续行
sum1 = 1 + \
       2 + \
       3
print(sum1)

# 用括号续行
sum2 = (1 +
        2 +
        3)
print(sum2)
````

### 2.代码块

- 在Python中，代码块由缩进定义，通常使用`4个空格`缩进1个层级，这与其他编程语言使用`{}`、`begin/end`等关键字标记代码块不同。

### 3.if语句

- 语句结构：`if` condition: statement `elif` condition: statement `else`: statement。
- 比较运算符：`==`(等于)、`!=`(不等于)、`>`(大于)、`<`(小于)、`>=`(大于等于)、`<=`(小于等于)。
- 逻辑运算符：`and`(且)、`or`(或)、`not`(非)。
    - 注意短路效应：`and`和`or`运算符在计算时，会根据操作数的真假来决定是否继续计算，而不是直接返回结果。
- 链式比较：Python支持连续比较写法，如`5 < x < 10`等价于`5 < x and x < 10`，也支持更长的链式比较，如`1 < x < 3 < y < 5`。

- 假值：`False`、`None`、`0`、`0.0`、`''`(空串)、`[]`(空列表)、`()`(空元组)、`{}`(空字典)、`set()`(空集)。
- 真值：`True`、非空值。

- 成员运算符：`in`(在)、`not in`(不在)。注意：对于字典，in查看的是键​，而不是值。

- 海象运算符：`:=`(赋值并返回值)，仅支持Python3.8及以上版本。
- 三目运算符语法：`x if condition else y`。

````python
limit = 280
tweet = 'Hello' * 60
if (diff := limit - len(tweet)) >= 0:  # 海象运算符，先计算，在赋值diff，然后判断
    print('tweet is valid')
else:
    print(f'tweet is invalid, {diff} characters too long')

age = 18
status = '成年' if age >= 18 else '未成年'  # 类似三目运算符语法
print(status)
````
