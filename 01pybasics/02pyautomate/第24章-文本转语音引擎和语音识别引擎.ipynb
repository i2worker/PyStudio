## 第24章 文本转语音引擎和语音识别引擎

### 1.使用pyttsx3模块进行文本转语音

- 初始化：`engine = pyttsx3.init()`。
- 设置音量：`engine.setProperty("volume", 0.5)`。
- 设置语速：`engine.setProperty("rate", 150)`。
- 设置音调：`engine.setProperty("pitch", 100)`。
- 设置语音：`engine.setProperty("voice", voice_id)`。
- 文本转语音：`engine.say(text)`。
- 保存语音：`engine.save_to_file(text, "filename.wav")`。
- 播放或保存：`engine.runAndWait()`。

### 2.使用openai-whisper模块进行语音识别

- 载入语音识别模型：`model = whisper.load_model("base")`。
- 识别音频：`result = model.transcribe(filename, language="zh")`。
- 创建字幕文件：`whisper.utils.get_writer("srt", "字幕文件路径")`。
