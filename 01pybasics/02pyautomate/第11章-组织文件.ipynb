### 1.shutil模块

- **复制文件**：`shutil.copy(source, destination)`，如果目标是目录，则将文件复制到该目录下并保留原文件名；如果目标是包含文件名的完整路径，则复制并重命名文件。
    - `shutil.copytree(source, destination)`，递归复制目录树，将所有文件和子目录复制到目标目录。

```python
import shutil
from pathlib import Path
path = Path.cwd()
shutil.copy(path / "res/spam/file1.txt", path / "res")
shutil.copy(path / "res/spam/file1.txt", path / "res/spam/file2.txt")
shutil.copytree(path / "res/spam", path / "res/spam_backup")
```

- **移动文件**：`shutil.move(source, destination)`。如果目标是目录，则将文件移动到该目录下并保留原文件名；如果目标是包含文件名的完整路径，则将文件重命名为该文件名。

```python
import shutil
from pathlib import Path
path = Path.cwd()
(path / "res/spam2").mkdir(exist_ok=True)
shutil.move(path / "res/spam/file1.txt", path / "res/spam2")
shutil.move(path / "res/spam2/file1.txt", path / "res/spam/file1.txt")
```

- **删除文件**：
    - `shutil.rmtree(path)`：删除path处整个文件夹树，包括所有文件和子文件夹。
    - `Path.unlink()`：删除path处单个文件。
    - `Path.rmdir()`：删除path处空目录。
    - `send2trash.send2trash(path)`：将path处文件或目录删除到回收站。

### 2.遍历目录

- **遍历目录树**：
    - `os.walk()`：递归遍历目录树，返回每个目录、子目录和文件的元组。
    - `os.listdir()`：返回目录下所有文件和子目录的列表。
    - `Path.iterdir()`：返回目录下所有文件和子目录的迭代器。

```python
import os, shutil
from pathlib import Path

# 创建测试文件夹结构
h = Path.cwd() / "res"
(h / "spam").mkdir(exist_ok=True)
(h / "spam/eggs").mkdir(exist_ok=True)
(h / "spam/eggs2").mkdir(exist_ok=True)
(h / "spam/eggs/bacon").mkdir(exist_ok=True)
for f in ['spam/file1.txt', 'spam/eggs/file2.txt', 'spam/eggs/file3.txt', 'spam/eggs/bacon/file4.txt']:
    with open(h / f, "w", encoding="utf-8") as file:
        file.write("Hello")

# 遍历文件夹结构
for folder, subfolders, files in os.walk(h / "spam"):
    print(f"The current folder is {folder}")

    # 子文件夹
    for subfolder in subfolders:
        print(f"SUBFOLDER OF {folder}: {subfolder}")

    # 文件
    for file in files:
        print(f"FILE INSIDE {folder}: {file}")
        p = Path(folder)
        shutil.move(p / file, p / file.upper()) # 文件名改为大写
    print("")
```

### 3.压缩文件(zipfile模块)

- **创建压缩文件**：
    - 以写入模式打开文件：`zipfile.ZipFile('文件名.zip', 'w')`。
    - 添加文件：`zip_obj.write('文件名', compress_type=zipfile.ZIP_DEFLATED, compresslevel=9)`，`compress_type`为压缩类型，`compresslevel`为压缩等级。

```python
import os, zipfile
from pathlib import Path

# 创建测试文件
h = Path.cwd() / "res"
with open(h / "test.txt", "w", encoding="utf-8") as file:
    file.write("Hello" * 10000)

# 压缩文件
os.chdir(h)
with zipfile.ZipFile("test.zip", "w") as zip:
    zip.write("test.txt", compress_type=zipfile.ZIP_DEFLATED, compresslevel=9)
os.chdir(h.parent)
```

- **读取压缩文件**：
    - `ZipFile.namelist()`方法返回一个列表，获取压缩包内所有文件名的列表。
    - `ZipFile.getinfo('文件名')`方法返回一个`ZipInfo`对象，包含文件的属性信息。
        - `ZipInfo.filename`：文件名。
        - `ZipInfo.file_size`：文件大小(字节)。
        - `ZipInfo.compress_size`：压缩后的文件大小(字节)。
        - `ZipInfo.compress_type`：压缩类型。
        - `ZipInfo.compress_level`：压缩等级。
        - `ZipInfo.date_time`：文件修改时间(时间戳)。

```python
import os, zipfile
from pathlib import Path
os.chdir(Path.cwd() / "res")
with zipfile.ZipFile("test.zip") as zip:
    print(zip.namelist())
    print(zip.getinfo("test.txt").file_size)
os.chdir(Path.cwd().parent)
```

- **解压压缩文件**：
    - `ZipFile.extractall()`：解压压缩包内的所有文件到当前目录。
    - `ZipFile.extract('文件名')`：解压指定文件到当前目录。
