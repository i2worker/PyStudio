## 第23章 控制键盘和鼠标

Python中使用**PyAutoGUI**模块来模拟键盘和鼠标操作。

### 1.操作鼠标

- 模拟鼠标移动：`pyautogui.moveTo(x, y, duration=0)`，`pyautogui.move(xOffset, yOffset, duration=0)`。
- 获取鼠标位置：`pyautogui.position()`。
- 模拟鼠标点击：`pyautogui.click(x, y, button='left')`。
    - 双击鼠标左键：`pyautogui.doubleClick(x, y)`。
    - 单击鼠标右键：`pyautogui.rightClick(x, y)`。
    - 单击鼠标中键：`pyautogui.middleClick()`。
- 模拟鼠标拖拽：`pyautogui.dragTo(x, y, duration=0)`，`pyautogui.drag(xOffset, yOffset, duration=0)`。
- 模拟鼠标滚动：`pyautogui.scroll(clicks)`。
- 获取鼠标信息：`pyautogui.mouseInfo()`。

### 2.操作屏幕

- 截屏：`pyautogui.screenshot()`。
- 获取屏幕指定像素的颜色：`pyautogui.pixel(x, y)`。
- 判断指定像素位置的颜色是否相同：`pyautogui.pixelMatchesColor(x, y, (r, g, b))`。
- 判断指定图像是否在屏幕中：`pyautogui.locateOnScreen(image)`。

### 3.操作窗口

- 获取活动窗口：`windows = pyautogui.getActiveWindow()`，返回一个Window对象。
- 获取所有窗口：`windows = pyautogui.getAllWindows()`，返回一个Window对象列表。
- 窗口最大化：`window.maximize()`。
- 窗口最小化：`window.minimize()`。
- 窗口恢复：`window.restore()`。
- 窗口关闭：`window.close()`。

### 4.操作键盘

- 模拟键盘输入：`pyautogui.write(msg, interval)`。
- 模拟快捷键：`pyautogui.hotkey(key1, key2, ...)`。
