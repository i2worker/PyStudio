## 第18章 CSV、JSON和XML文件

### 1.使用csv库

- 读取CSV文件：
    - 只读打开文件：`file = open("filename.csv")`。
    - 创建读取对象：`reader = csv.reader(file)`，**注意：不能把表示文件名的字符串作为参数传递给reader()**。
        - `csv.reader(file, delimiter=",")`，`delimiter`参数指定分隔符。
    - 获取其中的值：`data = list(reader)`，然后利用索引(如data[row][col])获取其中的值。**注意：这样将会载入所有数据而占用较大内存。**
    - 遍历每行数据：`for row in reader:`，每次迭代返回一行数据。**注意：reader对象只能迭代一次。**
    - 关闭CSV文件：`file.close()`。

- 写入CSV文件：
    - 写模式打开文件：`file = open("filename.csv", "w", newline="")`。**注意：Windows系统中需设置newline参数为空字符串，否则每行后会多出一行空行。**
    - 创建写入对象：`writer = csv.writer(file)`。
        - `csv.writer(file, delimiter="\t", lineterminator="\n\n")`，`delimiter`指定分隔符，默认为逗号。`lineterminator`指定行终止符，默认为回车符。
    - 写入一行数据：`writer.writerow(["name", "age", "gender"])`。
    - 关闭CSV文件：`file.close()`。

- 处理含有标题行的CSV文件：
    - 只读打开文件：`file = open("filename.csv")`。
    - 创建DictReader对象：`reader = csv.DictReader(file)`。
        - 创建DictReader对象时，可以指定列名：`reader = csv.DictReader(file, ["name", "age"])`。
    - 迭代DictReader对象：`for row in reader:`，每次迭代会返回一个字典，其中键是标题行中的列名，值是该列中的数据。
    - 创建DictWriter对象：`writer = csv.DictWriter(file, ["name", "age"])`。
    - 写入标题行：`writer.writeheader()`。
    - 写入数据：`writer.writerow({"name": "Alice", "age": 18})`。**注意：字典中键值对的顺序无关紧要，自动按照DictWriter对象中指定的顺序写入。**
    - 关闭文件：`file.close()`。

```python
import csv

# 读取CSV文件
file = open("res/demo.csv", "r")
reader = csv.reader(file, delimiter=",")
for row in reader:
    print(f"Row #{reader.line_num}: {row}")  # 通过列表索引访问数据
file.close()

# 写入CSV文件
file = open("res/demo2.csv", "w", newline="")  # 必须指定newline参数为空字符串防止空行
writer = csv.writer(file, delimiter=",")
writer.writerow(['spam', 'eggs', 'bacon', 'ham'])
writer.writerow(['Hello, world!', 'eggs', 'bacon', 'ham'])
writer.writerow([1, 2, 3.1415926, 4])
file.close()

print()

# 读取含标题行的CSV文件
file = open("res/header.csv", "r")
reader = csv.DictReader(file, delimiter=",")
for row in reader:
    print(f"Row #{reader.line_num}: {row['Timestamp']} {row['Fruit']} {row['Quantity']}")  # 通过字典键索引访问数据
file.close()

# 写入含标题行的CSV文件
file = open("res/header2.csv", "w", newline="")
writer = csv.DictWriter(file, delimiter=",", fieldnames=['Name', 'Pet', 'Phone'])
writer.writeheader()  # 写入标题行
writer.writerow({'Name': 'Alice', 'Pet': 'cat', 'Phone': '555-1234'})
writer.writerow({'Name': 'Bob', 'Phone': '555-9999'})  # 可以缺少某些列
writer.writerow({'Phone': '555-5555', 'Name': 'Carol', 'Pet': 'dog'})  # 顺序可以不一致
file.close()
```

### 2.使用json库

- 读取json数据：`json.loads()` 或 `json.load()`，返回一个包含数据的字典。
- 写入json数据：`json.dumps()`，将Python值转换为包含JSON数据的字符串。
    - indent参数：指定缩进，默认为None。

```python
import json

# 读取JSON数据
string = '''{
  "name": "Alice Doe",
  "age": 30,
  "car": null,
  "programmer":true,
  "address": {"street": "100 Larkin St.", "city": "San Francisco", "zip":"94102"},
  "phone": [{"type": "mobile", "number": "415-555-7890"}, {"type":"work", "number": "415-555-1234"}]
}'''
data = json.loads(string)
print(data['name'])  # 通过字典访问数据

print()

# 转换为JSON数据
data = {
  "name": "Alice Doe",
  "age": 30,
  "car": None,
  "programmer":True,
  "address": {"street": "100 Larkin St.", "city": "San Francisco", "zip":"94102"},
  "phone": [{"type": "mobile", "number": "415-555-7890"}, {"type":"work", "number": "415-555-1234"}]
}
string = json.dumps(data, indent=2)  # 指定缩进
print(string)
```

### 3.使用xml库

- 读取xml文件：
    - 从字符串解析xml：`xml.etree.ElementTree.fromstring()`。
    - 从文档中解析xml：`xml.etree.ElementTree.parse()`。
    - 获取元素标签名：`element.tag`。
    - 获取元素文本：`element.text`。
    - 遍历子元素：`element.iter()`。
- 写入xml文件：
    - 创建根元素：`xml.etree.ElementTree.Element()`。
    - 创建子元素：`xml.etree.ElementTree.SubElement()`。
    - 添加文本：`element.text = 'text'`。
    - 添加属性：`element.set()`。
    - 将元素转换为字符串：`xml.etree.ElementTree.tostring().decode()`。
