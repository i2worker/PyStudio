### 1.文件路径

- **标准化文件路径**：Windows系统中路径使用反斜杠 `\` 分隔不同文件夹名称，而macOS和Linux系统中使用的路径分隔符为正斜杠 `/`。
    - pathlib.Path()：能够根据不同操作系统做不同处理，<strong style="color:red;">最佳实践是在Python代码中使用正斜杠。</strong>

```python
from pathlib import Path
files = ["accounts.txt", "details.csv", "invite.docx"]
for file in files:
    print(Path(r"C:/Users/AI" + file))
```

- **合并路径**：运算符 `/` 可以将Path对象与字符串合并为一个路径，可以取代`os.path.join()`。
    - <strong style="color:red;">注意：在表达式的前两个值中，必须至少有一个是Path对象。</strong>

```python
from pathlib import Path
print(Path(r"C:/Users/AI") / "accounts.txt")
print(Path(r"C:/Users/AI") / "Claude/accounts.txt")
```

- **当前工作目录**：`Path.cwd()`获取当前工作目录(可以取代`os.getcwd()`方法)，`os.chdir()`修改当前工作目录(如果目录不存在会报错)。
    - `Path.home()`：获取用户主目录。

```python
from pathlib import Path
import os
print(Path.home())
print(Path.cwd())
os.chdir(r"E:/Downloads")
print(Path.cwd())
```

- **创建目录**：
    - `Path.mkdir()`：创建一个新文件夹。指定 `parents=True` 时创建所有必要的的父目录，指定 `exist_ok=True` 时检查目录是否已存在。
    - `os.makedirs()`：递归创建所有必要的中间文件夹。


```python
from pathlib import Path
import os
os.makedirs(r"E:/Downloads/PyStudio/01pybasics")
Path(r"E:/Downloads/PyStudio/02pypro").mkdir(parents=True)
```

- **绝对路径**：从根目录开始的完整路径。`Path.is_absolute()`判断是否为绝对路径。
- **相对路径**：从当前工作目录开始的路径。
- 相对路径转为绝对路径：表示相对路径的Path对象与`Path.cwd()`或`Path.home()`进行拼接，或者使用`Path.absolute()`方法。

```python
from pathlib import Path
p = Path(r"my/relative/path")
print(p.is_absolute())
print(Path.cwd() / p)
print(p.absolute())
```

- **路径属性**：
    - `Path.anchor`：路径的锚点(根目录)。
    - `Path.drive`：驱动器字母(仅限Windows系统)。
    - `Path.parent`：父目录(Path对象)。
    - `Path.parents`：所有父目录(Path对象列表)。
    - `Path.parts`：路径的各部分(字符串元组，包含各级目录名和文件名)。
    - `Path.name`：文件名(包含扩展名)。
    - `Path.stem`：文件名(不包含扩展名)。
    - `Path.suffix`：文件扩展名。

```python
from pathlib import Path
p = Path(r"C:/Users/AI/spam.txt")
print(p.anchor)
print(p.drive)
print(p.parent)
print(p.parents[0])
print(p.parts)
print(p.name)
print(p.stem)
print(p.suffix)
```

- **文件属性**：`Path.stat()`方法返回一个`os.stat_result`对象，包含文件的属性信息。
    - `st_size`：文件大小(字节)。
    - `st_mtime`：文件修改时间(时间戳)。
    - `st_ctime`：文件创建时间(时间戳)。
    - `st_atime`：文件最近访问时间(时间戳)。

```python
from pathlib import Path
p = Path(r"C:/Windows/System32/calc.exe")
print(p.stat().st_size)  # 文件大小
print(p.stat().st_mtime)  # 文件修改时间
print(p.stat().st_ctime)  # 文件创建时间，已废弃，被st_birthtime替换
print(p.stat().st_atime)  # 文件最近访问时间
```

- **查找文件**：`Path.glob()`方法可以查找指定模式的文件。
    - `Path.glob()`方法返回一个生成器对象，包含所有匹配的文件路径。如需要查看文件路径Path对象，需使用`list()`函数将其转换为列表。

```python
from pathlib import Path
p = Path.cwd()
print(list(p.glob("*"))[0])
for name in p.glob("*"):
    print(name)
```

- **路径检查**：
    - `Path.exists()`：检查路径是否存在。与`os.path.exists()`方法相同。
    - `Path.is_file()`：检查路径是否为文件。与`os.path.isfile()`方法相同。
    - `Path.is_dir()`：检查路径是否为目录。与`os.path.isdir()`方法相同。


### 2.读写文件

- **打开文件**：`Path.open()` 打开文件，返回一个文件对象。默认是只读模式。`encoding='utf-8'`参数指定编码。
- **读取内容**：`file.read()` 读取文件内容，返回一个字符串。也可以使用`file.readlines()`读取文件内容，返回一个列表，每个元素对应一行内容。还可以使用：
    - `Path.read_text()`：读取文件内容，返回一个字符串。
    - `Path.readlines()`：读取文件内容，返回一个列表，每个元素对应一行内容。
- **写入内容**：`file.write()` 写入内容到文件(不自动添加换行符)，打开文件时必须指定`w`(写入)或`a`(追加)模式。如果文件不存在，会创建一个新的文件。还可以使用：
    - `Path.write_text()`：写入内容到文件。
- **关闭文件**：`file.close()` 关闭文件，释放资源。

```python
from pathlib import Path

file = open(Path.cwd() / "res" / "bacon.txt", "w", encoding="utf-8")  # 写入文件
file.write("Hello, world!\n")
file.close()

file = open(Path.cwd() / "res" / "bacon.txt", "a", encoding="utf-8")  # 追加写入文件
file.write("Bacon is not a vegetable.")
file.close()

file = open(Path.cwd() / "res" / "bacon.txt", "r", encoding="utf-8")  # 读取文件
print(file.read())
file.close()
```

- **with语句**：`with open()` 打开文件，确保文件在使用后自动关闭。

```python
from pathlib import Path
with open(Path.cwd() / "res" / "bacon.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# 生成随机的考卷文件
import random
from pathlib import Path

# 考卷内容
capitals = {
    'Alabama': 'Montgomery', 'Alaska': 'Juneau', 'Arizona':'Phoenix',
    'Arkansas': 'Little Rock', 'California': 'Sacramento', 'Colorado':'Denver',
    'Connecticut': 'Hartford', 'Delaware': 'Dover', 'Florida':'Tallahassee',
    'Georgia': 'Atlanta', 'Hawaii': 'Honolulu', 'Idaho': 'Boise',
    'Illinois': 'Springfield', 'Indiana': 'Indianapolis', 'Iowa': 'Des Moines',
    'Kansas': 'Topeka', 'Kentucky': 'Frankfort', 'Louisiana': 'Baton Rouge',
    'Maine': 'Augusta', 'Maryland': 'Annapolis', 'Massachusetts': 'Boston',
    'Michigan': 'Lansing', 'Minnesota': 'Saint Paul', 'Mississippi': 'Jackson',
    'Missouri': 'Jefferson City', 'Montana': 'Helena', 'Nebraska': 'Lincoln',
    'Nevada': 'Carson City', 'New Hampshire': 'Concord', 'New Jersey': 'Trenton',
    'New Mexico': 'Santa Fe', 'New York': 'Albany', 'North Carolina': 'Raleigh',
    'North Dakota': 'Bismarck', 'Ohio': 'Columbus', 'Oklahoma': 'Oklahoma City',
    'Oregon': 'Salem', 'Pennsylvania': 'Harrisburg', 'Rhode Island': 'Providence',
    'South Carolina': 'Columbia', 'South Dakota': 'Pierre', 'Tennessee':'Nashville',
    'Texas': 'Austin', 'Utah': 'Salt Lake City', 'Vermont':'Montpelier',
    'Virginia': 'Richmond', 'Washington': 'Olympia', 'WestVirginia':'Charleston',
    'Wisconsin': 'Madison', 'Wyoming': 'Cheyenne'
}

# 生成35份试卷文件
for num in range(35):
    # 创建试卷文件和答案文件
    quiz_file = open(Path.cwd() / "res" / f"quiz_{num+1}.txt", "w", encoding="utf-8")
    answer_file = open(Path.cwd() / "res" / f"answer_{num+1}.txt", "w", encoding="utf-8")

    # 写入试卷卷首信息
    quiz_file.write("Name:\n\nDate:\n\nPeriod:\n\n")
    quiz_file.write((" " * 20) + f"State Capitals Quiz (Form{num+1})")
    quiz_file.write("\n\n")

    # 打乱各州排列顺序
    states = list(capitals.keys())
    random.shuffle(states)

    # 生成50道题
    for i in range(50):
        # 正确答案
        correct = capitals[states[i]]

        # 错误答案
        wrong = list(capitals.values())  # 复制所有答案
        del wrong[wrong.index(correct)]  # 删除正确答案
        wrong = random.sample(wrong, 3)  # 随机选择3个错误答案

        # 构建答题选项
        options = wrong + [correct]
        random.shuffle(options)

        # 写入题目和选项
        quiz_file.write(f"{i+1}. Capital of {states[i]}：\n")
        for j in range(4):
            quiz_file.write(f"{"ABCD"[j]}. {options[j]}\n")
        quiz_file.write("\n")

        # 写入答案
        answer_file.write(f"{i+1}. {"ABCD"[options.index(correct)]}\n")

    # 关闭文件
    quiz_file.close()
    answer_file.close()
```
