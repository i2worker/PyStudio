## 第12章 设计并部署命令行程序

### 1.使用终端

- **查看环境变量**：`echo %PATH%`(Windows系统) 或 `echo $PATH`(macOS和Linux系统)。
- **修改环境变量**：macOS系统中修改.zshrc文件，Linux系统中修改.bashrc文件，添加以下内容：`export PATH=...:$PATH`
- **定位程序**：`where <program>`(Windows系统) 或 `which <program>`(macOS和Linux系统)。


### 2.虚拟环境

- **创建虚拟环境**：`python -m venv <env-name>`。
- **激活虚拟环境**：`source <env-name>/bin/activate`(macOS和Linux系统) 或 `<env-name>\Scripts\activate`(Windows系统)。
- **退出虚拟环境**：`deactivate`。

### 3.使用pip

- **运行pip**：`python -m pip`。
- **列出已安装包**：`python -m pip list`。
- **安装包**：`python -m pip install <package>`。
- **升级包**：`python -m pip install --upgrade <package>`。
- **卸载包**：`python -m pip uninstall <package>`。


### 4. Python及系统相关信息

- `__file__`：当前.py文件的路径。
- `sys.executable`：Python解释器的完整路径。
- `sys.version`：Python解释器的版本信息。
- `sys.version_info`：Python解释器的版本信息，可以传递给函数list()。
- `sys.platform`：当前操作系统的标识信息。


### 5. 介绍几个包

- argparse：用于解析命令行参数的库。
- pyperclip：用于复制和粘贴文本的库。如：pyperclip.copy()、pyperclip.paste()。
- Bext：用于在命令行中显示彩色文本的库。如：Bext.fg()、Bext.bg()、Bext.clear()。
- playsound3：用于播放音频文件的库。如：playsound3.playsound()。
- pyinstaller：用于将Python脚本打包为独立执行文件的库。
