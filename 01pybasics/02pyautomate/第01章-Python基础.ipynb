### 第1章　Python基础

- **表达式**：由变量、常量、运算符、函数等组合而成，用于计算出一个结果。

- **运算符**：`**`(乘方)、`%`(求余)、`//`(整除)、`/`(除法)、`*`(乘法、字符串复制)、`-`(减法)、`+`(加法、字符串连接)。
    <p><img src="img/002.jpg" width="650"></p>

- **数据类型**：`int`(整数)、`float`(浮点数)、`str`(字符串)、`bool`(布尔值)、`None`(空值)。
    - <strong style="color: red;">🔔注意：整数和浮点数的数学运算结果为浮点数，整数的除法结果也为浮点数。</strong>
    <p><img src="img/003.jpg" width="650"></p>

- **变量**：用于存储数据的容器，是与值相关联的标签，变量名的命名规则如下：
    - 变量名只能包含字母、数字和下划线(_)，不能有空格、特殊字符等。
    - 变量名不能以数字开头。
    - 变量名不能使用Python保留字。
    - 变量名区分大小写，并且尽量使用小写的变量名。
    <p><img src="img/004.jpg" width="650"></p>

- **函数**：用于执行特定任务的代码块，常用函数如下：
    - `print()`：输出信息到控制台。
    - `input()`：获取用户输入的字符串。
    - `len()`：获取字符串的长度。
    - `int()`：将字符串转换为整数。
    - `float()`：将字符串转换为浮点数。
    - `str()`：将其他数据类型转换为字符串。
    - `type()`：获取变量的数据类型。
    - `round()`：获取四舍五入后的结果(注意 **.5** 结尾的浮点数将取整为最接近的偶数)。
    - `abs()`：获取绝对值。

```python
print("Hello, World!")
print("What's your name?")
name = input(">")
print(f"It is good to meet you, {name}!")
print(f"The length of your name is {len(name)} characters.")
print("What is your age?")
age = input(">")
print(f"You will be {int(age)+1} years old in a year.")
```
