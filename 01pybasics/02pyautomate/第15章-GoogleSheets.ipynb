## 第15章 Google Sheets

### 1.使用ezsheets模块

- 创建实例：`ezsheets.Spreadsheet(URL/ID)`。
- 查看所有电子表格：`ezsheets.listSpreadsheets()`。
- 上传电子表格：`ezsheets.upload("filename.xlsx")`。
- 下载电子表格：`Spreadsheet.downloadAsExcel("filename.xlsx")`。

- 新建工作表格：`Spreadsheet.Sheet("Sheet1")`。
- 复制电子表格：`Spreadsheet.copyTo()`。
- 删除电子表格：`Spreadsheet.delete()`。
- 刷新电子表格：`Spreadsheet.refresh()`。

- 读写电子表格数据：`sheet["A1"] = "Hello, World!"`。
- 转换行列地址：`ezsheets.convertAddress("A1")`、`ezsheets.getColumnLetterOf()`、`ezsheets.getColumnNumberOf()`。
- 读写整行/列：`sheet.getRow()`、`sheet.getColumn()`、`sheet.updateRow()`、`sheet.updateColumn()`。
