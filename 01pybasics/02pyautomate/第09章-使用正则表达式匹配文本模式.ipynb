## 第9章 使用正则表达式匹配文本模式

### 1.创建正则表达式

- 导入模块：`import re`
- 创建对象：`re.compile(pattern)`
- 搜索匹配：`re.search(pattern, string)`，**只返回第一个匹配结果**
- 获取结果：`match.group()`

```python
import re
p = re.compile(r'\d{3}-\d{3}-\d{4}')
m = p.search("My number is 415-555-4242.")
if m:
    print(m.group())    # 返回全部匹配的字符串
```

### 2.匹配模式分组

- 使用圆括号 `()` 可以在正则表达式中创建分组，提取匹配内容的特定部分。
    - `group()` / `group(0)`：返回整个匹配的字符串。
    - `group(1)`、`group(2)`：返回对应的第1、第2个分组内容。
    - `groups()`：返回一个包含所有分组内容的元组。

```python
import re
p = re.compile(r'(\d{3})-(\d{3}-\d{4})')    # 通过括号进行分组
m = p.search("My number is 415-555-4242.")
if m:
    print(m.group())    # 返回全部匹配的字符串
    print(m.group(0))   # 返回全部匹配的字符串
    print(m.group(1))   # 返回第一个括号匹配的字符串
    print(m.group(2))   # 返回第二个括号匹配的字符串
    print(m.groups())   # 返回所有括号匹配的字符串，以元组形式返回
```

### 3.匹配所有内容

- `findall()`：查找字符串中所有匹配的子串，而且返回的**匹配内容不会相互重叠**。
    - 注意：如果模式中没有分组，返回匹配字符串的列表；如果有分组，则返回包含分组元组的列表。

```python
import re
p =re.compile(r'(\d{3})-(\d{3})-(\d{4})')   # 包含编组的正则表达式
print(p.findall("Cell: 415-555-9999 Work: 212-555-0000"))  # 返回元组列表

p = re.compile(r'\d{3}')
print(p.findall("1234"))  # 不重复匹配，返回“123”
```

### ４.正则表达式语法

- 转义符 (`\`)：用于匹配具有特殊含义的字符本身(如 `\.`, `\?`, `\*`, `\\` 等)。
- 管道符 (`|`)：表示“或”操作，匹配多个表达式中的一个。
- 通配符 (`.`)：匹配除换行符外的任意单个字符。
- 开头锚定 (`^`)：匹配字符串的开头。
- 结尾锚定 (`$`)：匹配字符串的结尾。

- 字符集 (`[]`)：匹配方括号中的任意一个字符，例如 `[aeiouAEIOU]` 匹配所有元音字母。
    - 范围 (`-`)：指定字符的范围，如 `[a-z]` 表示匹配小写字母。
    - 取反 (`^`)：在字符集开头使用脱字符表示匹配不在集内的字符。例如 `[^aeiouAEIOU]` 匹配所有非元音字符。
    - 预定义字符集：如 `\d` 所有数字，`\w` 所有字母、数字和下划线，`\s` 所有空格字符等，`\D`, `\W`, `\S` 表示对应的非字符(如 `\D` 匹配非数字)。

- 量词：用于指定字符出现的次数范围。
    - `?`：匹配零个或一个字符。
    - `*`：匹配零个或多个字符。
    - `+`：匹配一个或多个字符。
    - `{n}`：匹配 n 个字符。
    - `{n,}`：匹配 n 个或多个字符。
    - `{,m}`：匹配 0 到 m 个字符。
    - `{n,m}`：匹配 n 到 m 个字符。注意：**第二数字式包含在范围内。**
    - `.*`：匹配任何内容。


- 贪婪匹配：默认模式，尽可能匹配最长的字符串。
    - 非贪婪匹配：在量词后加上问号 `?`，尽可能匹配最短的字符串。

```python
import re
p1 = re.compile(r"(Ha){3,5}")  # 贪婪匹配
m1 = p1.search("HaHaHaHaHa")
print(m1.group())  # type: ignore

p2 = re.compile(r"(Ha){3,5}?")  # 非贪婪匹配
m2 = p2.search("HaHaHaHaHa")
print(m2.group())  # type: ignore
```

### 5. 其他参数

- `re.compile()`方法中第二个参数是一个标志位，可以改变正则表达式的的行为。
    - `re.DOTALL`：使用`.`时也匹配换行符。
    - `re.IGNORECASE`：忽略大小写。
    - `re.VERBOSE`：允许在正则表达式中使用多行、注释和空格。
