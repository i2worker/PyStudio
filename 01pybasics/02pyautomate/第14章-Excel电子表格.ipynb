## 第14章 Excel电子表格

### 1.读取Excel文件

- 打开工作簿：`openpyxl.load_workbook('filename.xlsx')`。
- 获取工作表：`wb.sheetnames` 获取所有工作表名称，`wb['Sheet1']` 获取指定工作表，`wb.active` 获取当前活动工作表。
- 获取单元格：`sheet['A1']` 获取A1单元格，或者 `sheet.cell(row=1, column=1)` 获取A1单元格。
- 获取单元格值：`sheet['A1'].value` 获取A1单元格的值。
- 获取单元格坐标：`sheet['A1'].coordinate` 获取A1单元格的坐标。
- 获取单元格行号：`sheet['A1'].row` 获取A1单元格的行号。
- 获取单元格列号：`sheet['A1'].column` 获取A1单元格的列号。
- 获取单元格最大行号：`sheet.max_row` 获取工作表的最大行号。
- 获取单元格最大列号：`sheet.max_column` 获取工作表的最大列号。
- 字母转换为列号：`openpyxl.utils.get_column_letter(1)`，将列编号1转换为字母A。
- 列号转换为字母：`openpyxl.utils.column_index_from_string("A")` 将字母A转换为列编号1。
- 获取指定范围的单元格：`sheet['A1':'C3']` 获取A1到C3的单元格。

```python
import openpyxl, pprint

# 打开工作簿
print("Open workbook...")
wb = openpyxl.load_workbook("res/censuspop.xlsx")
sheet = wb["Population by Census Tract"]
countys = {}

# 读取单元格数据
print("Read rows...")
for row in range(2, sheet.max_row + 1):
    state = sheet[f"B{row}"].value
    county = sheet[f"C{row}"].value
    pop = sheet[f"D{row}"].value

    # 设置默认值
    countys.setdefault(state, {})
    countys[state].setdefault(county, {"tracts": 0, "pop": 0})

    # 计算县的数量和人口数量
    countys[state][county]["tracts"] += 1
    countys[state][county]["pop"] += pop

# 写入结果
print("Write results...")
with open("res/census2010.py", "w") as f:
    f.write(f"allData = {pprint.pformat(countys)}")

# 关闭工作簿
sheet.close()
wb.close()
print("Done!")
```

### 2.写入Excel文件

- 创建工作簿：`wb = openpyxl.Workbook()`，只有一个工作表，工作表的名称是"Sheet"。
- 保存工作簿：`wb.save()` 或 `wb.save("filename.xlsx")`，指定名称时保存为另外一个文件。
- 创建工作表：`wb.create_sheet(index=0, title="Sheet2")`，`index`是工作表的索引，从0开始，`title`是工作表的名称。
- 删除工作表：`del wb["Sheet2"]`，删除工作表"Sheet2"。
- 单元格写入值：`sheet["A1"] = "Hello, world!"`。

### 3.设置单元格样式

- 字体：`sheet["A1"].font = openpyxl.styles.Font(bold=True, size=24)`，设置单元格A1的字体为加粗24号。
- 公式：`sheet["A1"] = "=SUM(B1:B10)"`，设置单元格A1的公式为B1到B10的和。注意：仅当在Excel中打开文件并执行公式计算保存结果后，才能在Python中看到计算结果。
- 行高：`sheet.row_dimensions[1].height = 100`，设置第1行的高为100像素。
- 列宽：`sheet.column_dimensions["A"].width = 20`，设置第A列的宽为20个字符。
- 合并：`sheet.merge_cells("A1:B10")`，合并A1到B10的单元格。
- 拆分：`sheet.unmerge_cells("A1:B10")`，拆分A1到B10的单元格。
- 冻结：`sheet.freeze_panes = "C2"`，解冻：`sheet.freeze_panes = None`。
