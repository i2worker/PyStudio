## 第5章 调试

- **引发异常**：语法格式为: `raise Exception("错误信息")`。

```python
def box_print(symbol, width, height):
    if len(symbol) != 1:
        raise Exception("Symbol must be a single character string.")  # 引发异常
    if width <= 2:
        raise Exception("Width must be greater than 2.")  # 引发异常
    if height <= 2:
        raise Exception("Height must be greater than 2.")  # 引发异常
    print(symbol * width)
    for _ in range(height - 2):
        print(symbol + " " * (width - 2) + symbol)
    print(symbol * width)

try:
    box_print("#", 20, 3)
    box_print("##", 20, 3)  # 引发异常
except Exception as e:
    print(f"Error: {e}")
```

- **断言**：用于检查代码是否符合预期，不符合则引发 AssertionError 异常。语法为: `assert condition, "message"`。

```python
ages = [26, 57, 92, 54, 22, 15, 17, 80, 47, 73]
ages.sort()
assert ages[0] >= ages[-1], "Ages are not sorted in descending order"
```

- **日志**：可以使用 `logging` 模块创建日志，记录程序运行时的信息，如错误、警告、调试等。
    - 启用日志：`logging.basicConfig(level=logging.DEBUG, format='%(asctime)s - %(levelname)s - %(message)s')`。
    - 日志文件：`logging.basicConfig(filename='app.log', level=logging.DEBUG, format='%(asctime)s - %(levelname)s - %(message)s')`。
    - 记录日志：`logging.debug()`、`logging.info()`、`logging.warning()`、`logging.error()`。
    - 禁用日志：`logging.disable(logging.CRITICAL)`。
    - 日志级别：`DEBUG`、`INFO`、`WARNING`、`ERROR`。

```python
import logging

# 配置日志记录
logging.basicConfig(level=logging.DEBUG, format='%(asctime)s - %(levelname)s - %(message)s')
logging.debug("Start of program")

def factorial(n):
    logging.debug(f"Start of factorial({n})")
    total = 1
    for i in range(1, n + 1):
        total *= i
        logging.debug(f"i = {i}, total = {total}")
    logging.debug(f"End of factorial({n})")
    return total

print(factorial(5))
logging.debug("End of program")
```
