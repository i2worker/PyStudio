## 第20章 发送电子邮件、短信和通知

### 1.使用ezgmail模块处理Gmail

- 初始化：`ezgmail.init()`。
- 发送邮件：`ezgmail.send(to, subject, message)`。
- 获取最近20篇未读邮件：`emails = ezgmail.unread(maxResults=20)`，或者 `emails = ezgmail.recent(maxResults=20)`。返回值是`GmailThread`对象列表。
- 获取邮件列表内容摘要：`ezgmail.summary(emails)`。
- 获取指定邮件内容：`emails[0].messages`。messages是列表，其中元素是`GmailMessage`对象，该对象有描述邮件的属性：subject、body、timestamp、sender、recipient等。
- 搜索邮件：`emails = ezgmail.search(query)`。
- 下载附件：`emails[0].messages[0].downloadAttachment('attachment.pdf')`下载指定附件，`emails[0].messages[0].downloadAllAttachments()`下载所有附件。
