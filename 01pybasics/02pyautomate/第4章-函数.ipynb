## 第4章 函数

- **函数**：由关键字 `def` 定义，包括函数名、参数、函数体和返回值。

- **参数**：函数定义时指定的变量叫**形参**，调用函数时传递的值叫**实参**。

- **返回值**：函数执行完成后返回的结果，用 `return` 语句指定，如果没有 `return` 语句，函数返回 `None`。

- **位置参数**：根据参数定义顺序传递实参，顺序必须与形参一致。

- **具名参数**：也称为关键字参数，根据参数名称传递实参，名称必须与形参一致。


- **变量作用域**：变量在程序中有效的作用范围，根据变量的定义位置和使用位置，分为全局作用域和局部作用域，也称为**全局变量**和**局部变量**，在函数内可通过 `global` 关键字定义全局变量。
    - 作用域规则：
        - 全局作用域内的代码不能访问局部作用域内的变量。
        - 一个函数内的代码不能访问其他局部作用域内的变量。
        - 局部作用域内的代码可以访问全局作用域内的变量。
        - 不同局部作用域的变量可以同名，互不干扰。
        - 全局变量可以在任何作用域访问，局部变量只能在定义它的作用域访问。
    - 如何确定全局变量和局部变量：
        - 全局作用域内的变量总是全局变量；
        - 在函数中，使用 `global` 关键字定义的变量是全局变量。
        - 在函数中，使用赋值语句定义的变量是局部变量。
        - 在函数中，在使用前未被赋值的变量是全局变量。

- **异常处理**：用`try-except-else`语句处理异常，其中 `try` 语句捕获异常，`except` 语句处理异常，`else` 语句如果没有异常时执行的代码块。


```python
# 之字形运动
# 注意：该程序会无限运行，直到用户按下Ctrl+C退出，并且不能在Jupyter Notebook中运行
import time, sys

indent = 0  # 缩进空格数
indent_increasing = True  # 是否增加缩进空格

try:
    while True:
        print(" " * indent, end="")
        print("********")
        time.sleep(0.1)

        if indent_increasing:
            indent += 1
            if indent >= 20:  # 改变方向
                indent_increasing = False
        else:
            indent -= 1
            if indent == 0:  # 改变方向
                indent_increasing = True
except KeyboardInterrupt:
    sys.exit()
```

```python
# 绘制尖峰
# 注意：该程序会无限运行，直到用户按下Ctrl+C退出，并且不能在Jupyter Notebook中运行
import time, sys

try:
    while True:
        for i in range(1, 9):
            print("-" * (i * i))
            time.sleep(0.1)

        for i in range(7, 1, -1):
            print("-" * (i * i))
            time.sleep(0.1)
except KeyboardInterrupt:
    sys.exit()
```
