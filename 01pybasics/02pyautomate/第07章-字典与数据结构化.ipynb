## 第7章 字典与数据结构化

- **字典**：使用**键值对**存储数据的**可变数据类型**，用 **{}** 表示，通过键访问值，如`user['name']`。
    - **键**在字典中是唯一的，可以是整数、字符串等不可变数据类型，而**值**可以是任何数据类型。
    - 与列表相比，**列表是有序的**，相同元素顺序不同则列表不同，而**字典是无序的**，只要键值对相同则字典相同。

- **获取键值对**：`key()`获取字典中所有键的列表，`value()`获取字典中所有值的列表，`items()`获取字典中所有键值对的列表。
    - <strong style="color:red;">🔔注意：这些列表都是视图，不能直接修改字典中的数据。</strong>如果要转换为列表，需要使用`list()`函数。
    - 遍历：`for key in dict.keys():`遍历键，`for value in dict.values():`遍历值，`for key, value in dict.items():`遍历键值对。

- **检查键是否存在**：`if key in dict:` 或 `if key in dict.keys():` 或 `dict.get(key, default)`。

- **设置默认值**：`dict.setdefault(key, default)`，如果键(key)不存在，则设置默认值(default)。如果键(key)存在，则返回对应的值。


```python
msg = "It was a bright cold day in April, and the clocks were striking thirteen."
count = {}

for char in msg:
    count.setdefault(char, 0)  # 如果键(char)不存在，则设置默认值(0)，如果存在，则不修改
    count[char] += 1

print(count)
```

```python
guests = {
    'Alice': {'apples': 5, 'pretzels': 12},
    'Bob': {'ham sandwiches': 3, 'apples': 2},
    'Carol': {'cups': 3, 'apple pies': 1}
}

def total_brought(guests, item):
    numbers = 0
    for _, value in guests.items():
        numbers += value.get(item, 0)  # 如果item不存在，则返回0
    return numbers

print('Number of things being brought:')
print(f' - Apples         {total_brought(guests, "apples")}')
print(f' - Cups           {total_brought(guests, "cups")}')
print(f' - Cakes          {total_brought(guests, "cakes")}')
print(f' - Ham Sandwiches {total_brought(guests, "ham sandwiches")}')
print(f' - Apple Pies     {total_brought(guests, "apple pies")}')
```
