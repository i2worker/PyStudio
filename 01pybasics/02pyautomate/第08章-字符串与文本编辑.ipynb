## 第8章 字符串与文本编辑

- **转义字符**：字符串使用双引号或单引号包裹，当字符串内部包含特殊字符时，可使用反斜杠`\`进行转义。如：`\n`(换行)、`\t`(制表符)、`\r`(回车)等。

- **原始字符串**：使用前缀 `r` 表示字符串，不进行转义，常用于文件路径或正则表达式。如：`r'C:\Users\Al\Desktop'`，表示`C:\Users\Al\Desktop`。

- **多行字符串**：使用三引号 `"""` 或 `'''` 表示多行字符串，常用于编写多行文本或程序的文档字符串。

- **f-字符串**：在引号前加`f`，并在大括号`{}`中直接嵌入变量或表达式。若需输出大括号本身，使用双层大括号`{{}}`。如：`f'Hello, {name}!'`。

- **字符串格式化**：使用 `%` 或 `format()` 方法对字符串格式化。如：`'My name is %s.' % name` 或 `'My name is {}'.format(name)`。


- **字符串常用方法**：
    - `upper()` / `lower()`：将字符串全部转为大写/小写，返回新字符串(字符串不可变)。
    - `title()`：将字符串每个单词首字母大写，其余小写，返回新字符串(字符串不可变)。
    - `isupper()` / `islower()`：判断字符串是否至少有一个字母且所有字母均为大写/小写。
    - `istitle()`：是否每个单词首字母大写且其余小写。
    - `isalpha()`：是否只包含字母。
    - `isdecimal()`：是否只包含数字。
    - `isalnum()`：是否只包含字母和数字。
    - `isspace()`：是否只包含空格、制表符、换行等空白字符。
    - `startswith()` / `endswith()`：判断字符串是否以指定的字符串开头(或结尾)。
    - `join()`：将字符串列表连接成一个字符串。
    - `split()`：将字符串按指定分隔符拆分为列表，默认按空白字符拆分，如按`\n`拆分时可用于处理多行文本。
    - `rjust(width, fillchar)` / `ljust(width, fillchar)`：右/左对齐，并使用指定字符填充至指定宽度。
    - `center(width, fillchar)`：居中对齐并填充。
    - `strip()`：删除字符串两端指定的字符(默认删除空白字符)。
    - `lstrip()` / `rstrip()`：删除字符串左侧(或右侧)指定字符(默认删除空白字符)。
    - `ord()` / `chr()`：将字符转换为对应的ASCII码值，或将ASCII码值转换为对应的字符。
    - `pyperclip.copy(text)` / `pyperclip.paste()`：将文本复制到剪贴板，或将剪贴板中的文本粘贴到当前位置。

```python
print('Hello'.upper())
print('Hello'.lower())
print('hello'.title())
print('hello'.isupper())
print('hello'.islower())
print('hello'.istitle())
print('hello'.isalpha())
print('12345'.isdecimal())
print('hello123'.isalnum())
print('  '.isspace())
print('hello'.startswith('he'))
print('hello'.endswith('lo'))
print(' '.join(['My', 'name', 'is', 'Simon']))
print('My name is Simon'.split())
print('Hello'.rjust(10, "="))
print('Hello'.ljust(10, "="))
print('Hello'.center(10, "="))
print('SpamSpam'.strip('mSa'))
print('SpamSpam'.lstrip('mSa'))
print('SpamSpam'.rstrip('mSa'))
```
