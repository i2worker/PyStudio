## 第6章 列表

- **列表**：由多个值组成的有序集合，用 **[]** 表示，列表中的值称为**元素**，元素之间用**逗号**分隔，每个元素可以是**任意**类型，还可以包括其他列表，形成嵌套列表。

- **索引**：列表中每个元素的唯一标识符，可以从0开始(正向索引)，也可以从-1开始(反向索引)。

- **切片**：从列表中提取多个元素形成新的列表，语法为 `列表[开始索引:结束索引:步长]`，**注意不包含结束索引**。如省略开始索引，从列表开头开始切片；如省略结束索引，切片到列表末尾。`列表[:]` 复制列表。

- **列表操作**：包括修改、删除、连接、重复等。
    - 修改元素：使用**索引赋值**修改列表中的元素，例如 `nums[0] = 100`。
    - 删除元素：使用 `del` 语句删除列表中的元素，例如 `del nums[0]`。
    - 连接列表：使用 `+` 运算符将多个列表连接起来，例如 `nums = [1, 2, 3] + [4, 5, 6]`。
    - 重复列表：使用 `*` 运算符将列表重复多次，例如 `nums = [1, 2, 3] * 3`。

- **列表方法**：包括 `index()`、`append()`、`insert()`、`remove()`、`pop()` 等。
    - `index()`：查找列表中第一个匹配的元素的索引。
    - `append()`：在列表末尾添加一个元素，不会返回新列表。
    - `insert()`：在指定位置插入一个元素，会将后续元素向后移动。
    - `remove()`：删除第一个匹配的元素，不会返回新列表。
    - `sort()`：对列表进行升序排序，可指定 `reverse=True` 参数实现降序排列。
        - **直接在原列表上进行排序，不返回新列表**。
        - **无法对同时包含数字和字符串的列表进行排序**。
        - 对字符串排序时会根据ASCII值排序，如想按字母顺序排序，可使用 `key=str.lower` 参数。
    - `reverse()`：反转列表元素的顺序。
    - `pop()`：删除并返回最后一个元素。
    - `clear()`：清空列表。
    - `count()`：返回元素出现的次数。
    - `copy()`：返回列表的浅拷贝。

- **列表循环**：`for i in range(len(list))` 遍历列表中每个元素，`for index, value in enumerate(list)` 遍历列表中每个元素并获取元素的索引和值。

- **列表运算符**：`in` 检查元素是否在列表中，`not in` 检查元素是否不在列表中。

- **增强赋值运算符**：`+=`，`-=`，`*=`，`/=`，`%=`。

- **多重赋值**：用列表可以同时给多个变量赋值，每个变量之间用逗号隔开，并且变量数量与列表长度必须相同。如：`a, b, c = [1, 2, 3]`

- **常用函数和方法**：
  - `len()`：返回列表的长度。
  - `max()`：返回列表中最大的元素。
  - `min()`：返回列表中最小的元素。
  - `sum()`：返回列表中所有元素的总和。
  - `sorted()`：返回一个排序后的列表，不会修改原列表。
  - `reversed()`：返回一个反向迭代器，可以用于遍历列表中的元素。
  - `random.shuffle()`：随机重新排列列表中的元素。
  - `random.choice()`：随机选择一个列表中的元素。

- **序列数据类型**：列表、元组、字符串。其中列表是**可变数据类型**，元组和字符串是**不可变数据类型**。
    - 如想修改字符串，可以使用**切片**复制既有字符串的一部分，再使用**拼接**操作创建新字符串。

- **元组**：不可变序列数据类型，用 **()** 表示，元素之间用逗号隔开。如果元组只有一个元素，需要在元素后面添加**逗号**。
    - list()：将元组转换为列表。
    - tuple()：将列表转换为元组。

- **引用**：<strong style="color:red;">变量存储的是指向值的引用，而不是值本身。赋值操作复制的是引用，而不是值。</strong>
    - 函数调用时，参数是引用传递，而不是值传递。当传递的是可变数据类型时，函数内部修改参数值，会影响调用函数的变量。
    - 如果不希望函数内部修改参数值，可以是copy模块的 `copy.copy()`(浅拷贝) 和 `copy.deepcopy()`(深拷贝)进行复制。

```python
# 黑客帝国屏保
import random, sys, time

WIDTH = 70  # 列数
try:
    columns = [0] * WIDTH
    while True:
        for i in range(WIDTH):  # 遍历所有列
            if random.random() < 0.02:
                columns[i] = random.randint(4, 14)  # 设置当前列的计数

            if columns[i] == 0:  # 如果当前列的计数为0，打印空格
                print(" ", end="")
            else:  # 否则打印0或1，计数减1
                print(random.choice([0, 1]), end="")
                columns[i] -= 1
        print()  # 行末换行
        time.sleep(0.1)
except KeyboardInterrupt:
    sys.exit()
```
