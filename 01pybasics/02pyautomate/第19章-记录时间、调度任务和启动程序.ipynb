## 第19章 记录时间、调度任务和启动程序

### 1.time模块

- `time.time()`：获取UNIX纪元时间戳，从世界标准时间(UTC)1970年1月1日起过去了多少秒。
- `time.ctime()`： 获取当前时间的字符串描述，具体格式根据系统不同而不同。
- `time.sleep()`： 暂停当前线程，等待指定的时间。

```python
import time
string = time.ctime(time.time())  # 转换为系统格式的时间字符串
print(string)

string = time.strftime("%Y-%m-%d %H:%M:%S", time.localtime())  # 转换为指定格式的时间字符串
print(string)
```

### 2.datetime模块

- datetime对象：表示日期和时间的对象，包含年份、月份、日、小时、分钟、秒和毫秒。
    - `datetime.datetime.now()`：获取表示当前时间的datetime对象。
    - `datetime.datetime.fromtimestamp()`：将时间戳转换成datetime对象。
    - `datetime.datetime.timestamp()`：将datetime对象转换成时间戳。
- timedelta对象：表示时间间隔的对象。
    - `datetime.timedelta()`：创建一个timedelta对象。
    - `datetime.datetime.now() + datetime.timedelta(days=10)`：获取当前时间之后的10天，可以通过datatime对象与timedelta对象进行时间计算。
- 时间与字符串相互转换：
    - `datetime.datetime.strftime("时间格式字符串")`，将datetime对象转换成字符串。
    - `datetime.datetime.strptime("时间字符串", "时间格式字符串")`，将字符串转换成datetime对象。
    <p><img src="img/007.jpg" width="650"></p>

```python
import datetime
import time

# 获取当前时间的datatime对象
dt = datetime.datetime.now()
print(dt.year, dt.month, dt.day, dt.hour, dt.minute, dt.second, dt.microsecond)

# 将时间戳转换为datatime对象
dt = datetime.datetime.fromtimestamp(time.time())
print(dt)

# 将datatime对象转换为时间戳
stamp = dt.timestamp()
print(stamp)

# 时间比较运算
dt2026 = datetime.datetime(2026, 10, 31, 0, 0, 0)
dt2027 = datetime.datetime(2027, 1, 1, 0, 0, 0)
print(dt2026 > dt2027)
print(dt2026 < dt2027)

# 创建时间间隔对象
delta = datetime.timedelta(days=11, hours=10, minutes=9, seconds=8)
print(delta.days, delta.seconds, delta.microseconds, delta.total_seconds(), str(delta))

# 时间算术运算
print(dt + datetime.timedelta(days=1000))  # 加上1000天
print(dt - datetime.timedelta(days=365 * 30))  # 减去30年

# 格式化时间
print(dt.strftime("%Y-%m-%d %H:%M:%S"))  # datetime转字符串
dt2 = datetime.datetime.strptime('2026/10/21 16:29:00', '%Y/%m/%d %H:%M:%S')  # 字符串转datetime
```

### 3.subprocess模块

- 运行程序：`subprocess.run(["ls", "-l"])`，参数必须是列表，第一个元素为命令或程序，后面的元素为运行命令或程序的参数。**注意：该函数会阻塞，直到程序运行结束。**
    - `subprocess.Popen()`：不阻塞运行程序，调用后返回Popen对象，`popen.wait()`等待程序运行结束(阻塞)，`popen.poll()`检查程序是否运行结束(不阻塞)，`open.kill()`结束程序。
- 获取程序输出：`subprocess.run(["ls", "-l"], capture_output=True)`，参数`capture_output`为True时，将程序输出保存在`stdout`和`stderr`属性中。
- 默认程序打开文件：`subprocess.run(["start", "test.txt"], shell=True)`。**注意：Windows系统必须指定`shell=True`，而MacOS、Linux系统中第一个参数是`open`。
