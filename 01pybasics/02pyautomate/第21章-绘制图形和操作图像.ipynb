## 第21章 绘制图形和操作图像

### 1.计算机图像基础知识

- 颜色：用`RGBA`值表示，具体为一个四元组，分别代表红、绿、蓝和透明度，如(255, 0, 0, 255)表示红色。
    - 可以通过`pillow`模块中的`ImageColor.getcolor()`方法获取颜色名称对应的颜色值，如：`ImageColor.getcolor('red', 'RGBA')`。
- 位置：图像中像素的位置通过矩形元组`(left, top, right, bottom)`来表示，如(0, 0, 100, 100)表示左上角为(0, 0)，右下角为(100, 100)的矩形。
    - **注意：左坐标、上坐标包含在矩形内，但右坐标、下坐标不包含在矩形内。**

### 2.使用pillow模块绘制图像

- 载入图像：`image = Image.open(filename)`，返回一个`Image`对象，对图像的操作都是在这个对象上进行。
- 图像大小：`image.size`，返回表示图像宽度和高度的元组。
- 图像格式：`image.format`，还可以用`image.format_description`查看图像格式的详细描述。
- 显示图像：`image.show()`。
- 保存图像：`image.save(filename)`，可以保存为另外的图像格式。

```python
from PIL import Image
cat = Image.open("res/zophie.png")
print("Width: %s, Height: %s" % cat.size)
print(f"Format: {cat.format}")
cat.save("res/zophie.jpg")
cat.close()
```

- 新建图像：`Image.new(cmode, size, bgcolor)`，cmode为图像颜色模式，size为图像大小，bgcolor为背景颜色。
- 裁剪图像：`Image.crop(box)`，box为裁剪区域的矩形元组，**注意：不改变原图像。**
- 复制图像：`Image.copy()`，返回一个图像的副本。
- 粘贴图像：`Image.paste(im, box)`，im为粘贴的图像，box为粘贴区域的矩形元组。
- 缩放图像：`Image.resize(size)`，size为缩放后的大小(为整数元组)。
- 旋转图像：`Image.rotate(angle)`，angle为旋转角度。
- 翻转图像：`Image.transpose(method)`，method为翻转方法。
- 修改像素：`Image.putpixel((x, y), value)`，(x,y)为像素坐标，value为像素颜色值。`Image.getpixel((x, y))`，返回像素颜色值。

```python
from PIL import Image

# 新建图像
Image.new("RGBA", (100, 200), "red").save("res/red.png")

# 裁剪图像
cat = Image.open("res/zophie.png")
cat2 = cat.crop((140, 140, 250, 250))

# 复制/粘贴图像
cat3 = cat.copy()
cat3.paste(cat2, (0, 0))
cat3.paste(cat2, (230, 350))
cat3.save("res/zophie_paste.png")

# 缩放图像
cat.resize((int(cat.size[0] / 2), int(cat.size[1] / 2))).save("res/zophie_resize.png")

# 旋转图像
cat.rotate(90).save("res/zophie_rotate.png")

# 翻转图像
cat.transpose(Image.FLIP_LEFT_RIGHT).save("res/zophie_transpose.png") # type: ignore

# 修改像素
for x in range(cat.size[0]):
    for y in range(int(cat.size[1] / 2)):
        cat.putpixel((x, y), (210, 210, 210))
for x in range(cat.size[0]):
    for y in range(int(cat.size[1] / 2), cat.size[1]):
        cat.putpixel((x, y), (169, 169, 169))
cat.save("res/zophie_putpixel.png")
```

- 绘制图形：通过`ImageDraw`模块的`Draw`类绘制图形，`draw = ImageDraw.Draw(img)`。
    - 绘制点：`draw.point(xy, fill)`。
    - 绘制直线：`draw.line(xy, fill, width)`。
    - 绘制矩形：`draw.rectangle(xy, fill, outline, width)`。
    - 绘制椭圆：`draw.ellipse(xy, fill, outline, width)`。
    - 绘制多边形：`draw.polygon(xy, fill, outline, width)`。
    - 绘制文字：`draw.text(xy, text, fill, font)`，font参数是`ImageFont.truetype(font_file, size)`生成的字体对象。

```python
from PIL import Image, ImageDraw, ImageFont

# 创建绘画对象
im = Image.new('RGBA', (200, 200), 'white')
draw = ImageDraw.Draw(im)

# 绘制直线
draw.line([(0, 0), (199, 0), (199, 199), (0, 199), (0, 0)], fill='black')

# 绘制矩形
draw.rectangle((20, 30, 60, 60), fill='blue')

# 绘制圆形
draw.ellipse((120, 30, 160, 60), fill='red')

# 绘制多边形
draw.polygon(((57, 87), (79, 62), (94, 85), (120, 90), (103, 113)), fill='brown')

for i in range(100, 200, 10):
    draw.line([(i, 0), (200, i - 100)], fill='green')

# 绘制文字
draw.text((20, 150), 'Hello', fill='purple')
font = ImageFont.truetype('arial.ttf', 32)
draw.text((90, 150), 'Howdy', fill='gray', font=font)

# 保存图片
im.save("res/draw.png")
```

### 3.使用matplotlib绘制图像

```python
import matplotlib.pyplot as plt
x = [0, 1, 2, 3, 4, 5]
y1 = [10, 13, 15, 18, 16, 20]
y2 = [9, 11, 18, 16, 17, 19]

# 绘制折线图
plt.plot(x, y1)
plt.plot(x, y2)

# 绘制散点图
plt.scatter(x, y1)
plt.scatter(x, y2)
plt.show()
```

```python
import matplotlib.pyplot as plt
categories = ['Cats', 'Dogs', 'Mice', 'Moose']
values = [100, 200, 300, 400]

# 绘制柱状图
plt.bar(categories, values)
plt.show()
```

```python
import matplotlib.pyplot as plt
slices = [100, 200, 300, 400]
labels = ['Cats', 'Dogs', 'Mice', 'Moose']

# 绘制饼状图
plt.pie(slices, labels=labels, autopct='%.1f%%')
plt.show()
```
