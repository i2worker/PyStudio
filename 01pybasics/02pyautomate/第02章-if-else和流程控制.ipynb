## 第2章 if-else和流程控制

- **比较运算符**：`==`(等于)、`!=`(不等于)、`>`(大于)、`<`(小于)、`>=`(大于等于)、`<=`(小于等于)。
    - 整数与浮点数数值相等时，比较运算符的返回值为`True`，如`1 == 1.0`。
    - 运算符`<`、`>`、`<=`、和`>=`只能用于整数值和浮点数值，不能用于字符串。
    <p><img src="img/005.jpg" width="650"></p>

- **布尔运算符**：`and`(且)、`or`(或)、`not`(非)。

- **if-else语句**：`if` 条件1: 代码块1 `elif` 条件2: 代码块2 `else:` 代码块3。
    - <strong style="color: red;">🔔注意：`elif`语句可以有多个，`else`语句可以没有；在有一系列`elif`语句时，将执行其中一条子句或不执行任何子句，elif语句的排列顺序很重要。</strong>

```python
today = True  # 不管设置为True或False，结果都是："Today is not Opposite Day."

if today:
    opposite = True
else:
    opposite = False

if today:
    opposite = not opposite

if opposite:
    print("Today is Opposite Day.")
else:
    print("Today is not Opposite Day.")
```

```python
print("Enter TB or GB for the advertised unit:")
unit = input(">")

# 计算宣传的容量与实际容量的比值
if unit.upper() == "TB":
    discrepancy = 1_000_000_000_000 / 1_099_511_627_776
elif unit.upper() == "GB":
    discrepancy = 1_000_000_000 / 1_073_741_824
else:
    print("Invalid unit. Please enter TB or GB.")
    exit()

# 输入宣传的容量
print("Enter the advertised capacity:")
capacity = float(input(">"))

# 计算实际容量
real_capacity = round(capacity * discrepancy, 2)
print(f"The actual capacity is {real_capacity} {unit}.")
```
