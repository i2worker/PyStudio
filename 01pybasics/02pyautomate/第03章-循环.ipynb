## 第3章 循环

- while循环：`while 条件: 循环体`。

- break语句：用于跳出循环。

- continue语句：用于跳出当前循环，继续下一次循环。

- for循环：`for 变量 in range(次数): 循环体`。

- 在编程中，区间通常是以 **“开，闭”** 表示的。即包含起始值，不包含结束值。

- range()函数：用于生成一个整数序列。语法：`range(起始值, 结束值, 步长)`。

- import语句：用于导入模块。语法：`import 模块名`。
    - from语句：用于从模块中导入函数或类。语法：`from 模块名 import 函数名, 类名`。
    - <strong style="color: red;">🔔注意：请不要将任何Python内置函数的名称用作文件名。</strong>

```python
# 猜数游戏
import random

number = random.randint(1, 20)
print('I think of a number between 1 and 20.')

for i in range(1, 7):
    print("Take a guess.")
    guess = int(input(">"))
    if guess < number:
        print("Your guess is too low.")
    elif guess > number:
        print("Your guess is too high.")
    else:
        break

if guess == number:
    print(f"Good job! You got it in {i} guesses!")
else:
    print(f"Nope. The number I was thinking of was {number}.")
```

```python
# 剪刀、石头、布游戏
import random, sys

print("ROCK, PAPER, SCISSORS!")

wins = 0
losses = 0
ties = 0

while True:
    print(f"Wins: {wins}, Losses: {losses}, Ties: {ties}")

    # 玩家输入
    while True:
        print("Enter your move: (r)ock (p)aper (s)issors or (q)uit")
        player = input(">").upper()
        if player == "Q":
            print("Thanks for playing!")
            sys.exit()
        if player in "RPS":
            break
        print("Type one of r, p, s, or q.")

    # 显示玩家选择
    if player == "R":
        print("You chose ROCK.")
    elif player == "P":
        print("You chose PAPER.")
    elif player == "S":
        print("You chose SCISSORS.")

    # 显示电脑选择
    computer = random.choice("RPS")
    choice = {'R': 'ROCK', 'P': 'PAPER', 'S': 'SCISSORS'}[computer]
    print(f"Computer chose {choice}.")

    # 判断游戏结果
    if player == computer:
        print("It's a tie!")
        ties += 1
    elif player == "R" and computer == "S":
        print("You win!")
        wins += 1
    elif player == "P" and computer == "R":
        print("You win!")
        wins += 1
    elif player == "S" and computer == "P":
        print("You win!")
        wins += 1
    else:
        print("You lose!")
        losses += 1
```
