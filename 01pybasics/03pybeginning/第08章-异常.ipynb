## 第8章 异常

### 1.创建异常

- **异常**：Python使用异常对象表示异常状态，是某个异常类的实例。当错误发生时，程序会引发异常。如果异常未被处理(捕获)，程序将终止并显示错误消息。
    - 传播机制： 函数内未处理的异常，会向上传播到调用函数的地方，直到被捕获或到达主程序。
- **引发异常**：可以使用 `raise` 语句手动引发异常，参数是异常类或异常实例。
    - 如：`raise Exception('Hyperdrive Overload')`。
- **自定义异常**：可以创建自己的异常类，继承自 `Exception` 或其他异常类。
    - 如：`class HyperdriveOverload(Exception):  pass`。
- **内置异常类**：Python内置了多个异常类：

| 类名                  | 描述                     |
| :-------------------- | :-----------------------|
| **Exception**         | 几乎所有异常类的基类。     |
| **AttributeError**    | 访问不存在的属性。         |
| **OSError**           | 操作系统错误。            |
| **IndexError**        | 序列索引越界。            |
| **KeyError**          | 字典键不存在。            |
| **NameError**         | 变量名未定义。            |
| **SyntaxError**       | 语法错误。                |
| **TypeError**         | 类型不匹配的操作。         |
| **ValueError**        | 类型正确但值不合适。       |
| **ZeroDivisionError** | 除以零。                 |

### 2.捕获异常

- **基本语法**：`try...except...`，可能引发异常的代码放在 `try` 中，处理异常的代码放在 `except` 中。
    - 重新引发异常：在 `except` 块中，可以使用不带参数的 `raise` 重新引发当前捕获的异常。
    - 异常上下文：如果在 `except` 块中引发另一个异常，原异常会被作为上下文存储。
        - 可以使用 `raise ... from ...` 指定上下文，或使用 `raise ... from None` 禁用上下文。
- **捕获多个异常**：可以使用多个 `except` 块捕获不同的异常。
    - 也可以使用一个 `except` 子句捕获多种异常，方法是将多个异常类放在元组中，如 `except (ZeroDivisionError, TypeError, NameError):`。
- **访问异常对象**：用 `as` 关键字将异常对象赋值给一个变量，可以在 `except` 块中使用该变量来访问异常对象。
    - 如：`except ZeroDivisionError as e:`。
- **捕获所有异常**：
    - `except`：捕获所有异常(包括KeyboardInterrupt、SystemExit)，极其危险，会隐藏一切错误。
    - `except Exception as e:`：捕获Exception及其子类的异常，稍微安全点，不会捕获KeyboardInterrupt等异常。
- **else**：`else` 子句在没有异常被捕获时执行。
    - 常用于循环输入场景。只有在输入有效(无异常)时，才通过 `break` 跳出循环。
- **finally**：`finally` 子句无论是否有异常被捕获，都会执行。即使异常未被捕获，`finally` 也会先执行，然后异常继续传播。
    - 主要用于执行清理工作，如关闭文件、数据库连接等资源。

````python
try:
    print(1 / 0)
except ZeroDivisionError:
    raise ValueError from None  # 禁用上下文，抛出 ValueError 异常
````

````python
try:
    print(1 / 0)
except ZeroDivisionError:
    print("Division by zero!")  # 捕获到指定异常时执行
else:
    print("That went well!")    # 没有异常时执行
finally:
    print("Cleaning up.")       # 无论是否有异常，都执行
````

### 3. warnings模块

- `warn(message, category=None)`：发出警告。默认只显示一次，类别必须是Warning的子类。
- `filterwarnings(action, category=Warning)`：过滤警告。
    - action：ignore(忽略警告)、error(抛出异常)、always(总是显示警告)、default(默认行为，同类警告只显示一次)。
    - category：警告类别，默认Warning。

````python
from warnings import warn

warn("This is a warning message.")      # 发出警告
````

````python
from warnings import warn, filterwarnings

filterwarnings("ignore")                # 忽略所有警告
warn("Anyone out there?")               # 无输出
````

````python
from warnings import warn, filterwarnings

filterwarnings("error")                 # 将警告转为异常
warn("Something is very wrong!")        # 抛出异常
````

````python
from warnings import warn, filterwarnings

filterwarnings("default")                                   # 恢复默认行为
filterwarnings("ignore", category=DeprecationWarning)       # 忽略 DeprecationWarning 警告
warn("This function is really old...", DeprecationWarning)  # 被忽略
warn("Something else.")                                     # 发出警告
````

### 4. 本章小结

- **知识脉络**：

```text
异常处理
│
├── 引发异常
│   ├── raise Exception('message')
│   ├── raise（无参数：重新引发当前异常）
│   ├── raise X from Y（设置异常上下文）
│   └── raise X from None（禁用上下文）
│
├── 自定义异常
│   └── class MyError(Exception): pass
│
├── 捕获异常
│   ├── try/except 基本结构
│   ├── 多个 except 子句
│   ├── except (A, B, C) 元组捕获
│   ├── except E as e 捕获对象 ⭐
│   ├── except Exception as e 安全兜底 ⭐
│   ├── else 子句（无异常时执行）
│   └── finally 子句（总是执行，清理用）⭐
│
├── 异常传播
│   └── 函数内未处理 → 向上传播到调用方
│
├── 编程哲学
│   ├── EAFP：先做后处理 ⭐（Python推荐）
│   └── LBYL：先查后做（不推荐）
│
└── 警告（warnings）
    ├── warn() 发出警告
    └── filterwarnings() 过滤警告
```

- **异常处理语法**：

```text
try:
    # 可能引发异常的代码
except SomeError as e:
    # 处理 SomeError
except (OtherError, AnotherError) as e:
    # 处理多种异常
except Exception as e:
    # 安全兜底（推荐替代裸except）
else:
    # 没有异常时执行
finally:
    # 总是执行（清理工作）
```
