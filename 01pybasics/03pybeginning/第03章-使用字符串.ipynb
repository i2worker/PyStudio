## 第3章 使用字符串


### 1.设置字符串格式

- **设置方法**：
    - **百分号(%)运算符**：在字符串中用`%s`(字符串)、`%d`(整数)、`%f`(浮点数)等**转换说明符**标记值的插入位置及格式，然后再跟随`%`运算符，最后是值的元组。
        - 例如：`"Hello, %s! You are %d years old." % ("Alice", 30)`。输出：Hello, Alice! You are 30 years old.
        - 注意：值的顺序必须与转换说明符的顺序一致。
    - **str.format()方法**：使用花括号`{}`作为占位符，然后调用`format()`方法，将值插入到占位符中。
        - 无索引/自动编号：`"{} {}".format("first", "second")`。
        - 有索引(索引可不按顺序排列)：`"{3} {0} {2} {1} {3} {0}".format("be", "not", "or", "to")`。输出：to be or not to be
        - 命名字段：`"{name} is {value}".format(name="π", value=3.14)`。
    - **f-字符串**：在Python 3.6及以上版本中，使用`f`或`F`前缀的字符串，直接在字符串中嵌入表达式，无需调用`format()`方法。
        - 例如：`f"Hello, {name}! You are {age} years old."`。

- **设置格式**：
    - 完整格式：`{字段名!转换标志:格式说明符}`。
        - 字段名：可以是索引或命名字段，指出要设置哪个值的格式并使用结果来替换该字段。
        - 转换标志：跟在`!`后面，使用指定的函数将对象转换为字符串，当前支持的字符包括`r`(表示repr)​、`s`(表示str)​和`a`(表示ascii)。
        - 格式说明符：跟在`:`后面，指定要应用的格式，包括格式类型​，字段宽度和数的精度，如何显示符号和千位分隔符，以及各种对齐和填充方式。
            - 顺序为：`{:<填充><对齐><符号><宽度><,><.精度><类型>`。

- **字段名**：
    - 第1种方式：只提供要设置其格式的未命名参数，并在格式字符串中使用未命名字段。此时，**将按顺序将字段和参数配对**。
    - 第2种方式：给参数指定名称，这种参数将被用于相应的替换字段中。
    - 第3种方式：可混合使用这两种方法，还可通过索引来指定要在哪个字段中使用相应的未命名参数。

```python
print("{foo} {} {bar} {}".format(1, 2, bar=4, foo=3))          # 3 1 4 2
print("{foo} {1} {bar} {0}".format(1, 2, bar=4, foo=3))        # 3 2 4 1
```

- **转换标志**：
    - `r`：使用repr()函数将对象转换为字符串，创建给定值的Python表示。
    - `s`：使用str()函数将对象转换为字符串，创建外观普通的字符串版本。
    - `a`：使用ascii()函数将对象转换为ASCII字符串。

```python
print("{pi!s} {pi!r} {pi!a}".format(pi="π"))        # π 'π' '\u03c0'
```

- **类型说明符**：指定要转换的值是哪种类型。

```python
print("The number is {num:b}".format(num=42))        # The number is 101010
print("The number is {num:f}".format(num=42))        # The number is 42.000000
print("The number is {num:x}".format(num=42))        # The number is 2a
```

- **宽度、精度、千位分隔符**：
    - 宽度：用整数指定，不足时填充空格。
    - 精度：用小数点 + 整数指定。
    - 分隔符：用逗号指定，**放在宽度和精度之间**。

```python
print("{:10}".format(3.14))             # '      3.14'
print("{:10.2f}".format(3.1415926))     # '      3.14'
print("{:15,.2f}".format(1000000))      # '   1,000,000.00'
```

- **填充、对齐、符号**：
    - 填充：用指定字符(默认为空格)填充不足的宽度。
        - `=`：表示填充字符放在符号与数字之间。
    - 对齐：指定对齐方式，包括左对齐(`<`)、右对齐(`>`)、居中对齐(`^`)。
    - 符号：`-`仅负数显示符号，`+`强制正数显示符号，`空格`正数前加空格(与负数的-对齐)。
    - `#`：显示进制前缀，如`0b`(二进制)、`0o`(八进制)、`0x`(十六进制)。

```python
print("{:0=+15.2f}".format(3.1415926))   # '+00000000003.14'
print("{:=-15.2f}".format(-3.1415926))   # '-          3.14'
print("{:=^15}".format(" WIN "))         # '===== WIN ====='
print("{:-<15}".format("LOSE"))          # 'LOSE-----------'
print("{:*>15}".format("COUNT"))         # '**********COUNT'
print("{:#b}".format(42))                # '0b101010'
```

### 2.字符串方法

- `str.center(width, [fillchar])`：两侧添加填充字符(默认空格)使字符串居中对齐，宽度为width，填充字符为fillchar。相关方法：ljust、rjust、zfill。
- `str.find(sub, [start, [end]])`：在字符串指定搜索范围中查找子串，返回其**首次出现的索引**；如果未找到，返回-1。相关方法：rfind、index、rindex、count、startswith、endswith。
- `sep.join(sequence)`：将列表中的元素用sep连接起来，返回一个新的字符串。**注意：序列元素必须全是字符串类型**。相关方法：split、partition、rpartition、splitlines。
- `str.lower()`：将字符串转换为小写。相关方法：islower、isupper、istitle、capitalize、casefold、swapcase、upper。
- `str.replace(old, new, [count])`：将字符串中所有子串old替换为new，返回一个新的字符串。
- `str.split([sep, [maxsplit]])`：将字符串按sep分隔，返回一个列表。
- `str.strip([chars])`：移除字符串首尾指定字符(默认为空格)，返回一个新的字符串。相关方法：lstrip、rstrip。
- `str.translate(table)`：根据指定的翻译表将字符串中的字符替换为其他字符，返回一个新的字符串。与replace的区别：translate只能替换单个字符，但可同时替换多个字符，效率更高。
- 判断字符串是否满足特定条件：`isalnum`(字母或数字)、`isalpha`(字母)、`isdigit`(数字)、`isspace`(空白)、`islower`(全小写)、`isupper`(全大写)、`istitle`(词首大写)等。
