## 第5章 条件、循环及其他语句

### 1.再谈print和import

- **print**：
    - 打印多个参数：`print(参数, 参数, 参数, ...)`。
        - `sep`参数：指定输出结果参数之间的分隔符。默认是空格。
        - `end`参数：指定输出结果行末尾的结束符，默认是换行符。
- **import**：
    - 导入时重命名：`import mod as alias` 或 `from mod import func as alias`。

### 2.赋值魔法

- **多重赋值**：同时给多个变量赋值，每个变量之间用逗号`,`分隔。
    - 实际执行的操作是**序列解包**：将序列(或可迭代对象)的值赋给多个变量。**注意：序列的长度必须与变量的数量相等。**
        - 可以使用星号`*`收集序列中的剩余元素，将它们赋值给一个列表。
- **链式赋值**：将同一个值赋给多个变量。例如：`x = y = somefunction()`。
- **增强赋值**：将运算符移到赋值运算符前。如：`+=`、`-=`、`*=`、`/=`、`//=`、`%=`等。

````python
name = "Albus Percival Wulfric Brian Dumbledore"
first, *middle, last = name.split()
print(first, middle, last)          # Albus ['Percival', 'Wulfric', 'Brian'] Dumbledore
````

### 3.代码块

- 代码块通过**缩进**(通常4个空格)创建，不用`{}`或`begin/end`。
- 冒号`:`标识代码块开始。有时候为了方便，会将简单语句放在同一行，如`if a == 1: print('a is 1')`。
- 同一代码块内缩进量必须相同，否则会报错。

### 4.条件和条件语句

- **布尔值**：任何Python值都可解释为布尔值，包括`True`和`False`，分别表示真和假，等价于1和0。
    - 假值：`False`、`None`、数值`0`(整数、浮点数)、空序列(空字符串`""`、空列表`[]`、空元组`()`)、空映射(空字典`{}`)等。**注意：空序列、空映射虽然为假值，但不等于False**。
    - 真值：假值以外的其他值都视为真值，还包括`True`。

````python
print(True + False + 42)        # 43
print([] != False)              # True
print(bool([]) == False)        # True
````

- **条件语句**：
    - if：根据条件执行不同的代码块。语法：`if 条件: 代码块`。
    - elif：根据多个条件执行不同的代码块。语法：`elif 条件: 代码块`。
    - else：如果以上条件都不满足，执行该代码块。语法：`else: 代码块`。
- **条件表达式**：语法：`变量 = 表达式1 if 条件 else 表达式2`。
    - 说明：如果条件为True，则变量的值为表达式1，否则为表达式2。

````python
status = "friend" if 1>2 else "stranger"
print(status)       # stranger
````

- **比较运算符**：包括`==`、`!=`、`>`、`<`、`>=`、`<=`、`is`、`is not`、`in`、`not in`。
    - 支持链式比较：可同时使用多个比较运算符，如`0 < age > 100`。
    - `is`运算符：`is`用来检查两个对象是否相同(是同一个对象)，`==`用来检查两个对象是否相等(值)​。**注意：不要将`is`用于数和字符串等不可变的基本值。**
    - 字符串是根据字符的字母排列顺序进行比较的。 列表、元组、集合等可迭代对象也可以进行比较，根据每个元素的比较结果进行判断。

````python
x = y = [1, 2, 3]
z = [1, 2, 3]
print(x == y)               # True
print(x == z)               # True
print(x is y)               # True, x和y指向同一个列表对象
print(x is z)               # False, x和z指向不同的列表对象

print("alpha" < "beta")     # True，按照字符的字母排列顺序比较
print([1, 2] > [1, 1])      # True，按照每个元素的比较结果进行判断
````

- **布尔运算符**：包括`and`、`or`、`not`。
    - and：逻辑与，两个操作数都为True时返回True，否则返回False。
    -  or：逻辑或，至少有一个操作数为True时返回True，否则返回False。
    - not：逻辑非，返回操作数的反面值。
- **短路逻辑**：
    - `x and y`：若x为假 → 返回x(不再计算y)；若x为真 → 返回y。
    - `x or y`： 若x为真 → 返回x(不再计算y)；若x为假 → 返回y。

````python
name = input('Please enter your name: ') or '<unknown>'
print(f'Hello, {name}!')    # 如果用户输入为空，则name为'<unknown>'。
````

### 5.循环

- **while循环**：
    - 语法：`while 条件: 循环体`。
    - 功能：当条件为True时，重复执行循环体中的语句。
    - 注意：循环条件必须在循环体中更新，否则会形成无限循环。

- **for循环**：
    - 语法：`for 变量 in 可迭代对象: 循环体`。常与`range()`函数配合使用(`range(start, stop, step)`)。
    - 功能：遍历可迭代对象中的每个元素，将每个元素赋值给变量，然后执行循环体中的语句。
    - 遍历字典：`for key in d`遍历键(也可以使用`d.keys()`)；`d.values()`遍历值；`d.items()`遍历键值对，可配合序列解包得到键和值。

- **迭代工具**：
    - `zip()`：将多个可迭代对象打包成一个元组序列，用于同时遍历多个可迭代对象。**注意：最短序列用完即停止**。
    - `enumerate()`：为可迭代对象生成索引-值对。
    - `reversed()`：返回一个可迭代对象的逆序版本。
    - `sorted()`：返回一个排序后的列表，原列表保持不变。

````python
names = ['anne', 'beth', 'george', 'damon']
ages = [12, 45, 32, 102]

# 并行迭代
for name, age in zip(names, ages):
    print(f'{name} is {age} years old.')

# 迭代时获取索引
for index, name in enumerate(names):
    print(f'{index}: {name}')

# 反向迭代
for name in reversed(names):
    print(name)

# 排序迭代
for name in sorted(names):
    print(name)
````

- **跳出循环**：
    - `break`：立即跳出循环，不执行循环体中的剩余代码。
    - `continue`：跳过当前循环，继续下一次循环。
- **else**：
    - 当循环正常结束时(没有执行`break`时)，执行的代码块。

````python
from math import sqrt
for i in range(99, 81, -1):
    root = sqrt(i)
    if root == int(root):
        print(i)
        break
else:  # 循环正常结束时执行
    print("Didn't find it!")
````

### 6.列表推导式

- **列表推导**：
    - 基本形式：`[表达式 for 变量 in 可迭代对象]`。
    - 添加条件：`[表达式 for 变量 in 可迭代对象 if 条件]`。
    - 多个for：`[表达式 for 变量1 in 可迭代对象1 for 变量2 in 可迭代对象2]`。
    - 多个for+if：`[表达式 for 变量1 in 可迭代对象1 for 变量2 in 可迭代对象2 if 条件]`。

````python
# [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
squares = [x**2 for x in range(10)]
print(squares)

# [0, 4, 16, 36, 64]
squares = [x**2 for x in range(10) if x % 2 == 0]
print(squares)

# [(0, 0), (0, 1), (0, 2), (1, 0), (1, 1), (1, 2)]
squares = [(x, y) for x in range(2) for y in range(3)]
print(squares)

# [(0, 1), (0, 2), (1, 0), (1, 2)]
squares = [(x, y) for x in range(2) for y in range(3) if x != y]
print(squares)

# 字典推导
# {0: '0', 1: '1', 2: '4', 3: '9', 4: '16', 5: '25', 6: '36', 7: '49', 8: '64', 9: '81'}
squares = {i: f"{i**2}" for i in range(10)}
print(squares)
````

### 7.其他语句

- **pass语句**：空语句，用作占位符，使代码块不为空。
- **del语句**：删除变量或数据结构成员(如列表元素、字典键)，但不删除值本身(值由垃圾回收处理)。
- **exec()函数**：将字符串作为Python代码执行，可提供命名空间避免污染全局变量。
- **eval()函数**：计算字符串表达式的值并返回结果。

````python
exec("print('Hello, world!')")      # Hello, world!

from math import sqrt
scope = {}
exec('sqrt = 1', scope)
print(sqrt(4))                      # 2.0
print(scope['sqrt'])                # 1

eval("6 + 18 * 2")                  # 42
````
