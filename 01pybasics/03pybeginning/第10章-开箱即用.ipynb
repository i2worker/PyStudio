## 第10章 开箱即用

### 1.模块

- 模块是Python代码复用、逻辑拆分的基础单元，**任何Python程序都可作为模块导入**，本质是包含Python定义和语句的.py文件，**文件名即为模块名**。
    - **模块查找**：模块需位于解释器的搜索路径中，默认搜索当前目录、标准库目录等，也可自定义搜索路径：
        - 通过修改`sys.path`添加自定义目录，如`sys.path.append(dir)`。
        - 通过环境变量`PYTHONPATH`添加自定义目录，如`export PYTHONPATH=$PYTHONPATH:~/python`。
    - **模块导入**：
        - 首次导入模块时会执行模块内的所有代码，后续重复导入不会执行代码(可避免循环导入问题)。
        - 模块导入后会生成`__pycache__`缓存目录，存储编译后的字节码以加速后续导入。
    - **模块重载**：若需应用修改后的模块代码，可使用`importlib.reload(module)`重载模块。
        - **注意**：重载不会更新已基于旧模块实例化的对象，需重新创建对象。
    - **特殊用法**：可通过`python -m 模块名`直接运行可作为模块的脚本。

````python
import sys
sys.path.append('mod')              # 告诉解释器去哪找模块

import hello  # type: ignore        # 第一次导入时执行模块代码，输出：hello world!  后续导入时无输出

import importlib
hello = importlib.reload(hello)     # 强制重新加载模块，输出：hello world!
````

- 模块的核心价值是拥有**独立作用域**，内部定义的类、函数、变量均为模块的属性，可实现代码复用，避免多项目重复编写相同逻辑。
    - 在模块中定义函数：导入后可通过`模块名.函数名`调用，适合多项目复用的工具类代码。
    - 在模块中添加测试代码：用`if __name__ == '__main__':`条件判断，主程序运行模块时，`__name__`值为`'__main__'`，执行测试代码；模块导入时，`__name__`值为模块名，不执行测试代码。

````python
import mod.hello2 as hello2     # 正常导入模块
hello2.hello()                  # 调用模块中的函数
hello2.test()

!python mod/hello2.py           # 运行程序，通过__name__判断，执行测试代码
!python -m mod.hello2           # 将模块作为程序运行，通过__name__判断，执行测试代码
````

### 2.包

- 包是包含其他模块的的特殊模块，用于组织复杂项目的模块结构。
    - 实现方式：包是一个目录，需包含`__init__.py`文件，内容可为空也可包含代码，用于初始化包，Python3.3及以上版本可以省略。

### 3.研究模块

- 使用`dir()`函数：列出模块所有属性(函数、类、变量等)。如：`[n for n in dir(copy) if not n.startswith('_')]`。
- 查看`__all__`变量：列出模块定所有公开接口。使用`from module import *`时，仅导入`__all__`内的名称；若未定义`__all__`，则导入所有非下划线开头的全局名称。
- 使用`help()`函数：查看模块详细帮助信息(函数签名、文档字符串、功能说明等)。如：`help(copy.deepcopy)`。
- 查看`__doc__`属性：查看模块或函数的文档字符串。如：`copy.deepcopy.__doc__`。
- 查看源代码：通过模块的`__file__`属性获取模块源码路径。如：`copy.__file__`。

| 方法                | 用途           |
| ------------------- | ------------- |
| `dir(module)`       | 列出所有属性   |
| `module.__all__`    | 查看公有接口   |
| `help(module.func)` | 获取帮助信息   |
| `module.__doc__`    | 查看文档字符串 |
| `module.__file__`   | 定位源文件     |

### 4.常用标准库

- sys模块：与Python解释器交互。

| 变量/函数                 | 作用                               |
| ------------------------- | ---------------------------------- |
| `sys.argv`                | 命令行参数列表，第一个元素为脚本名                     |
| `sys.path`                | 模块搜索路径                       |
| `sys.platform`            | 平台标识（`'win32'`、`'linux'`等） |
| `sys.version`             | Python版本字符串                   |
| `sys.exit()`              | 退出程序                           |
| `sys.stdin/stdout/stderr` | 标准输入/输出/错误流               |
| `sys.modules`             | 已加载模块字典                     |

- os模块：与操作系统交互。

| 变量/函数           | 作用                     |
| ------------------ | ----------------------- |
| `os.environ`       | 环境变量字典             |
| `os.system(cmd)`   | 执行系统命令             |
| `os.sep`           | 路径中文件夹分隔符（`/` 或 `\`） |
| `os.pathsep`         | 多条路径分隔符，UNIX为`:`，Windows为`;`        |
| `os.linesep`         | 文本行分隔符，UNIX为`\n`，Windows为`\r\n`  |
| `os.urandom(n)`      | 返回n字节强加密随机数据     |

- fileinput模块：迭代多个文件的行。

| 函数                      | 作用                   |
| ------------------------- | ---------------------- |
| `fileinput.input()`       | 返回可迭代的行对象     |
| `fileinput.filename()`    | 返回当前处理的文件名      |
| `fileinput.lineno()`      | 返回累计行号（跨文件不重置）   |
| `fileinput.filelineno()`  | 返回当前文件的行号（跨文件重置    |
| `fileinput.isfirstline()` | 判断是否为当前文件的第一行 |
| `fileinput.isstdin()`     | 判断最后一行是否来自标准输入   |
| `fileinput.nextfile()`    | 关闭当前文件，跳到下一个文件    |
| `fileinput.close()`       | 关闭整个文件链                |


- 集合：内置`set`类型，用于存储不重的元素集合，创建用`set()`或`{1， 2， 3， ...}`(空集合不能用`{}`，那是空字典)。

| 操作   | 运算符   | 方法                        |
| ------ | -------- | --------------------------- |
| 并集   | `a \| b`  | `a.union(b)`                |
| 交集   | `a & b`  | `a.intersection(b)`         |
| 差集   | `a - b`  | `a.difference(b)`           |
| 对称差 | `a ^ b`  | `a.symmetric_difference(b)` |
| 子集   | `a <= b` | `a.issubset(b)`             |
| 父集   | `a >= b` | `a.issuperset(b)`           |


- 堆：`heapq`模块，是一种优先队列，最小元素始终在顶部。

| 函数                   | 作用         |
| ---------------------- | ------------ |
| `heapify(list)`        | 让列表具有堆特征   |
| `heappush(heap, item)` | 向堆中添加元素     |
| `heappop(heap)`        | 从堆中弹出最小元素 |
| `heapreplace(heap, item)` | 从堆中弹出最小元素，再添加新元素 |
| `nlargest(n, seq)`     | 返回seq中n个最大的元素    |
| `nsmallest(n, seq)`    | 返回seq中n个最小的元素    |


- 双端队列:`collections.deque`模块，是一种在两端都可以高效插入和删除元素的队列。

| 操作     | 方法            |
| -------- | --------------- |
| 右端添加 | `append(x)`     |
| 左端添加 | `appendleft(x)` |
| 右端弹出 | `pop()`         |
| 左端弹出 | `popleft()`     |
| 旋转     | `rotate(n)`     |

- time模块：时间处理。时间表示方式有两种：新纪元(UNIX为1970-1-1 0:0:0)后的秒数(实数)、9元素时间元组(年、月、日、时、分、秒、星期、儒略日、夏令时)。

| 函数                   | 作用                        |
| ---------------------- | --------------------------- |
| `time.time()`          | 返回当前UTC时间（新纪元后秒数） |
| `time.asctime([t])`    | 时间元组→可读字符串         |
| `time.localtime([ts])` | 时间戳→时间元组             |
| `time.mktime(t)`       | 时间元组→时间戳             |
| `time.strptime(string[, format])` | 字符串转时间元组   |
| `time.sleep(secs)`     | 暂停指定秒数                |

- random模块：伪随机数生成。

| 函数                                      | 作用                                       |
| ----------------------------------------- | ------------------------------------------ |
| `random.random()`                         | 返回0~1（含）的随机实数                    |
| `random.getrandbits(n)`                   | 返回n位随机二进制数（长整数）              |
| `random.uniform(a, b)`                    | 返回a~b（含）的均匀随机实数                |
| `random.randrange([start,] stop[, step])` | 从`range(start, stop, step)`中随机选一个数 |
| `random.choice(seq)`                      | 从序列中随机选一个元素                     |
| `random.shuffle(seq)`                     | 就地打乱序列                               |
| `random.sample(seq, n)`                   | 从序列中选n个不重复的随机元素              |


- re模块：正则表达式。

**正则表达式语法速查：**

| 语法     | 含义               | 示例                  |
| -------- | ------------------ | --------------------- |
| `.`      | 任意字符（除换行） | `a.b` → `acb`         |
| `^`      | 字符串开头         | `^Hello`              |
| `$`      | 字符串结尾         | `world$`              |
| `*`      | 前一项0次或多次    | `ab*c`                |
| `+`      | 前一项1次或多次    | `ab+c`                |
| `?`      | 前一项0次或1次     | `ab?c`                |
| `{n,m}`  | 前一项n到m次       | `a{2,4}`              |
| `[...]`  | 字符集             | `[a-z]`               |
| `[^...]` | 补集               | `[^0-9]`              |
| `\d`     | 数字               | 等价于 `[0-9]`        |
| `\w`     | 单词字符           | 等价于 `[a-zA-Z0-9_]` |
| `\s`     | 空白字符           | 空格、制表符等        |
| `\b`     | 单词边界           | `\bword\b`            |
| `(...)`  | 编组（捕获组）     | `(ab)+`               |
| `\1`     | 反向引用编组1      | `(\w+)\s+\1`          |

**贪婪 VS 非贪婪：**

| 运算符             | 行为                 | 示例                      |
| ------------------ | -------------------- | ------------------------- |
| `*` `+` `{n,m}`    | 贪婪（尽量多匹配）   | `<.*>` 匹配整个 `<a>b<c>` |
| `*?` `+?` `{n,m}?` | 非贪婪（尽量少匹配） | `<.*?>` 只匹配 `<a>`      |

**核心函数**

| 函数                                   | 作用                                       |
| ------------------------------------- | ------------------------------------------ |
| `re.compile(pattern)`                 | 编译正则表达式为模式对象，提升匹配效率       |
| `re.search(pattern, string)`             | 查找第一个匹配的子串，返回MatchObject |
| `re.match(pattern, string)`             | 匹配字符串开头，返回MatchObject |
| `re.split(pattern, string)`             | 按正则表达式分割字符串，返回列表 |
| `re.findall(pattern, string)`           | 返回所有匹配子串的列表       |
| `re.sub(pattern, repl, string)`          | 替换匹配的子串 |

- 其他模块：
    - `argparse`：命令行参数解析，快速构建功能完备的CLI界面
    - `cmd`：命令行解释器框架，可自定义交互式命令
    - `csv`：CSV文件读写，处理表格数据交换
    - `datetime`：更直观的日期时间对象，比`time`模块更易用

    - `enum`：枚举类型支持
    - `functools`：函数工具，如`partial`偏函数
    - `hashlib`：哈希签名计算，加密与安全场景使用
    - `itertools`：迭代器工具，创建、合并迭代器
    - `logging`：日志管理，支持多优先级日志输出
    - `statistics`：统计计算，如均值、中位数、标准差
    - `timeit`/`profile`/`trace`：性能测试、代码覆盖率分析工具
