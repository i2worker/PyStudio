## 第1章 快速上手：基础知识

- 程序：是一组有序的Python语句，用于执行特定任务。
    - 算法：告诉计算机如何完成特定任务的步骤。
    - 语句：让计算机执行的操作指令，Python中每个语句后**不需要分号结尾**。如一行中有多个语句，每个语句之间用分号隔开。
    - 表达式：简单表达式由运算符、函数、变量、字面值等组成。

- 变量：是表示或指向特定值的名称，使用变量前必须先**赋值**。
    - 变量名：只能包含**字母、数字和下划线**，且不能以数字开头、不能使用Python保留字。

- 数值：包括整数、浮点数、复数(需导入`cmath`模块后使用)等。
    - 算术运算符：`+`加、`-`减、`*`乘、`/`除(浮点数结果)、`//`整除(向下取整)、`%`取余、`**`指数等。
    - 数值的进制：Python支持不同进制的整数表示，如`0b`二进制、`0o`八进制、`0x`十六进制等。
    - `round()`：取整到最接近的整数，如果两个整数一样近时取整到偶数，如`round(2.5)`返回`2`，`round(3.5)`返回`4`。

- 字符串：可以使用单引号(`'`)和双引号(`"`)表示字符串。
    - 转义字符：使用反斜杠`\`处理特殊符号，如 `\'`、`\"`、`\n`等。
    - 原始字符串：使用`r`前缀表示字符串，不处理转义字符。**注意：原始字符串不能以单个反斜杠结尾(除非转义)**。
    - 多行字符串：使用三引号(`'''`或`"""`)表示多行字符串。
    - 拼接字符串：使用 `+` 运算符拼接字符串。
    - `str()`：以用户友好方式转换为字符串。
    - `repr()`：返回合法的Python表达式表示。

- 运行Python脚本：命令行中输入`python myscript.py`。
    - 让Python脚本像普通程序：先在第一行添加`#!/usr/bin/env python3`，然后设置可执行权限：`chmod +x myscript.py`。

- 注释：Python支持单行注释（`#`）、多行注释（`'''`或`"""`）和文档字符串（`"""`或`'''`）。
