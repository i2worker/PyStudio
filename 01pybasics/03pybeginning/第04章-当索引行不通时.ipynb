## 第4章 当索引行不通时

### 1.创建字典

- **常用方式**：用花括号`{}`创建，其中由**键**及其相应的**值**组成，这种键-值对称为**项​**，每个键与值之间用冒号`:`分隔，项之间用逗号`,`分隔。
    - 例如：`{'name': '张三', 'age': 18, 'gender': '男'}`。`{}`表示空字典。
    - 注意：**键必须是唯一的**，值可以是任意类型。字典中值不按顺序存储，只能通过键来访问。
- **函数方式**：通过`dict()`函数从键-值对序列、关键字实参、其他映射(如其他字典)创建字典。
    - 例如：`dict([('name', '张三'), ('age', 18), ('gender', '男')])` 或 `dict(name='张三', age=18, gender='男')`。

- **字典与列表的区别**：列表是**顺序排列、编号访问**的数据结构，字典是**键值对应、按名访问**的数据结构。
    - 键的类型：字典的键必须是唯一的，且可以是任意不可变类型。列表的索引必须是整数。
    - 自动添加：字典中给不存在的键赋值将自动创建新项，而列表中给不存在的索引赋值将出现异常。
    - 成员资格：`in`运算符，字典中检查的是键，而列表中检查的是值。

### 2.字典方法

- `clear()`：删除字典所有项，就地执行，返回None。
- `copy()`：返回字典的浅拷贝，修改副本的可变值会影响原字典。如需深拷贝，可使用`copy.deepcopy()`。
- `fromkeys()`：创建新字典，包含指定的所有键，默认值为None，也可自定义默认值。
- `get()`：返回指定键的值，若键不存在则返回默认值None，也可以自定义默认值。
- `setdefault()`：指定键不存在时，返回指定的值并相应地更新字典；指定键存在时，返回其值，并保持字典不变。
- `items()`：返回包含所有字典项的列表，每个元素为(key, value)形式。
- `keys()`：返回包含所有键的列表。
- `values()`：返回包含所有值的列表。
- `pop()`：删除指定键的项，并返回该项的值。
- `popitem()`：随机删除一个键值对，并返回该对。
- `update()`：用另一个字典/键-值对序列/关键字参数更新当前字典，键相同则替换值，键不同则添加新项。
