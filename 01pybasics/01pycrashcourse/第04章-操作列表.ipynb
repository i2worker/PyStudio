## 第4章 操作列表

### 4.1 遍历列表

- 通过 **for...in...** 循环遍历列表中所有元素。

```python
magicians = ['alice', 'david', 'carolina']
for magician in magicians:
    print(f'{magician.title()}, that was a great trick!')
    print(f'I can\'t wait to see your next trick, {magician.title()}.\n')
print('Thank you, everyone. That was a great magic show!')
```

### 4.2 数值列表

- 通过 **range(start, stop, step)** 函数创建数列(不含stop)，通过 **list()** 函数将 range() 函数返回的数列转换成列表。

```python
numbers = list(range(2, 11, 2))
print(numbers)
```

- 可以对数值列表进行简单的统计计算。

```python
numbers = list(range(1, 10))
numbers.append(0)
print(numbers)
print(min(numbers))
print(max(numbers))
print(sum(numbers))
```

- 列表推导式：将 **for** 循环的结果保存在一个列表中。

```python
squares = [value ** 2 for value in range(1, 11)]
print(squares)
```

### 4.3 切片

- 获取列表中的部分元素称为切片。

```python
players = ['charles', 'martina', 'michael', 'florence', 'eli']
print(players[0:3])
print(players[1:4])
print(players[:4])
print(players[2:])
print(players[-3:])
```

- 复制列表：通过创建一个包含整个列表的切片来复制列表。

```python
foods1 = ['apple', 'pear', 'orange', 'banana']
foods2 = foods1[:]
foods1.append('coconut')
foods2.append('mango')
print(foods1)
print(foods2)
```

### 4.4 元组

- 元组：用 **(,)** 创建，其中的元素不可以修改。

```python
dimensions = (100, 200)
print(dimensions)
print(dimensions[0])
print(dimensions[1])
```

```python
# 练习4.1：比萨
print("练习4.1：比萨")
pizzas = ["pepperoni", "hawaiian", "veggie"]
for pizza in pizzas:
    print(pizza)

for pizza in pizzas:
    print(f"I like {pizza} pizza.")

print("I really love pizza!")

# 练习4.2：动物
print("\n练习4.2：动物")
animals = ["dog", "cat", "rabbit"]
for animal in animals:
    print(animal)

for animal in animals:
    print(f"A {animal} would make a great pet.")

print("Any of these animals would make a great pet!")
```

```python
# 练习4.3：数到20
print("练习4.3：数到20")
for number in range(1, 21):
    print(number,end=" ")
print("")

# 练习4.5: 100万求和
print("\n练习4.5: 100万求和")
numbers = list(range(1, 1000001))
print(f"最小值：{min(numbers)}")
print(f"最大值：{max(numbers)}")
print(f"总和：{sum(numbers)}")

# 练习4.6：奇数
print("\n练习4.6：奇数")
for number in range(1, 21, 2):
    print(number, end=" ")
print("")

# 练习4.7: 3的倍数
print("\n练习4.7: 3的倍数")
for number in range(3, 31, 3):
    print(number, end=" ")
print("")

# 练习4.8：立方
print("\n练习4.8：立方")
numbers = []
for number in range(1, 11):
    numbers.append(number**3)
for number in numbers:
    print(number, end=" ")
print("")

# 练习4.9：立方推导式
print("\n练习4.9：立方推导式")
numbers = [number**3 for number in range(1, 11)]
for number in numbers:
    print(number, end=" ")
print("")
```
