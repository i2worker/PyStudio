## 第7章 用户输入与while循环

### 7.1 获取用户输入

- **input()**：获取用户输入

```python
message = input('Please enter your name: ')
print(f'Hello, {message.title()}')
```

```python
# 通过int()函数将输入的字符串转换为数字
age = input("How old are you? ")
age = int(age)
if age >= 18:
    print("You are old enough to vote!")
elif age >= 16:
    print("You can learn to drive.")
else:
    print("You can't vote or learn to drive yet.")
```

### 7.2 while循环

- 一般使用

```python
prompt = input('Enter "quit" to exit. ')

# 设置退出标志
active = True

while active:
    message = input(prompt)

    # 设置退出条件
    if message.lower() == 'quit':
        active = False
    else:
        print(message)
```

- **break**：跳出当前循环

```python
prompt = "Please enter the name of a city you have visited:"
prompt += "\n(Enter 'quit' when you are finished.) "
while True:
    city = input(prompt)
    if city == 'quit':
        break  # 结束循环
    else:
        print(f"I'd love to go to {city.title()}!")
```

- **continue**：跳过当前循环，继续下一次循环

```python
number = 0
while number < 10:
    number += 1
    if number % 2 == 0:
        continue # 结束当前循环，继续下一次循环
    print(number)
```

- 循环列表：用 **while** 遍历列表的同时修改列表

```python
users1 = ['alice', 'brian', 'candace']
users2 = []

# 通过pop()方法将列表中的元素逐个取出来，然后通过append()方法添加到另一个列表中
while users1:
    user = users1.pop()
    print(f'Verifying user: {user.title()}')
    users2.append(user)

for user in users2:
    print(user.title())
```

```python
pets = ['dog', 'cat', 'dog', 'goldfish', 'cat', 'rabbit', 'cat']
print(pets)

# 删除列表中所有指定的值
while 'cat' in pets:
    pets.remove('cat') # 该方法只删除第一个匹配项
print(pets)
```

- 循环字典

```python
responses = {}
active = True
while active:
    name = input("\nWhat is your name? ")
    response = input("Which mountain would you like to climb someday? ")
    responses[name] = response
    repeat = input("Would you like to let another person respond? (yes/no) ")
    if repeat.lower() == 'no':
        active = False
        print("\n--- Poll Results ---")

for name, response in responses.items():
    print(name + " would like to climb " + response + ".")
```

```python
# 练习7.8：熟食店
print("练习7.8：熟食店")
sandwich1= ['tuna', 'chicken', 'veggie']
sandwich2 = []
while sandwich1:
    sandwich = sandwich1.pop()
    print(f"I'm working on your {sandwich} sandwich.")
    sandwich2.append(sandwich)
for sandwich in sandwich2:
    print(f'I made a {sandwich} sandwich.')

# 练习7.9：五香烟熏牛肉卖完了
print("\n练习7.9：五香烟熏牛肉卖完了")
sandwich= ['tuna', 'pastrami', 'chicken', 'pastrami', 'veggie', 'pastrami']
print("I'm sorry, we're all out of pastrami today.")
while 'pastrami' in sandwich:
    sandwich.remove('pastrami')
print(sandwich)
```

```python
# 练习7.10：梦想中的度假胜地
name_prompt = "What is your name? "
place_prompt = "If you could visit one place in the world, where would it be? "
continue_prompt = "Would you like to let someone else respond? (yes/no) "

responses = {}

while True:
    name = input(name_prompt)
    place = input(place_prompt)
    responses[name] = place
    repeat = input(continue_prompt)
    if repeat.lower() == 'no':
        break

for name, place in responses.items():
    print(f"{name.title()} would like to visit to {place.title()}.")
```
