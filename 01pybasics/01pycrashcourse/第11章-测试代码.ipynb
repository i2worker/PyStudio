## 第11章 测试代码

Python中通常使用 pytest 包进行测试，可以在命令行中输入 `pytest` 来运行测试代码。

### 11.1 测试函数

- 使用 pytest 进行测试时，文件名必须以 **test_** 开头，函数名也必须以 **test_** 开头。
- 断言：使用`assert`语句，假定满足某个特定条件，如果条件不满足，将触发测试失败。
- [详见：test/test_survey.py](test/test_survey.py)

```python
# 切换至测试目录
%cd -q test

# 运行函数测试
!python -m pytest test_name.py -v

# 切换回主目录
%cd -q ..
```
