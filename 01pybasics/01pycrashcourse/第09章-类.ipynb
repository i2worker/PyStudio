### 9.1 创建类

- 类：由关键字 `class` 定义，包含 **属性(变量)** 和 **方法(函数)**，并通过 `.` 访问属性和方法。根据约定，类名首字母**大写**，属性和方法名首字母**小写**。
  - `__init__()`：特殊的方法，当创建实例时自动调用，用于初始化实例。
  - `self`：表示当前实例，用于访问实例的属性和方法。

```python
# 定义类
class Dog:
    """文档注释：一个模拟狗的类"""

    # 初始化方法
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def sit(self):
        print(f"{self.name.title()} is now sitting.")

    def roll_over(self):
        print(f"{self.name.title()} rolled over!")

# 创建实例
my_dog = Dog('willie', 6)

# 访问属性
print(f"My dog's name is {my_dog.name.title()}.")
print(f"My dog is {my_dog.age} years old.")

# 调用方法
my_dog.sit()
my_dog.roll_over()
```

### 9.2 使用类

- 修改属性：既直接修改属性的值，也可以通过方法来修改属性的值。

```python
# 定义类
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0  # 该属性不在形参中定义，初始值设为0

    def get_descriptive_name(self):
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can can't roll back an odometer!")

    def increment_odometer(self, miles):
        self.odometer_reading += miles

my_new_car = Car('audi', 'a4', 2024)
print(my_new_car.get_descriptive_name())

my_new_car.odometer_reading = 23  # 直接修改属性的值
my_new_car.read_odometer()

my_new_car.update_odometer(24)  # 通过方法修改属性的值
my_new_car.read_odometer()

my_new_car.increment_odometer(100)  # 通过方法让属性的值递增
my_new_car.read_odometer()
```

### 9.3 继承

- 继承：基于已有的类创建新的类，已有类称为父类，新类称为子类。子类继承父类的属性和方法，也可以添加自己的属性和方法，还可以重写父类的方法。
  - 创建子类时，必须在类名后面的括号中指定父类的名称。
  - super()：特殊的函数，用于调用父类的方法。

```python
# 继承：电动汽车
class ElectricCar1(Car):  # 在括号中指定父类的名称
    def __init__(self, make, model, year):
        super().__init__(make, model, year)  # 通过super()函数调用父类的方法
        self.battery_size = 40  # 添加子类特有的属性

    def describe_battery(self):  # 添加子类特有的方法
        print(f"This car has a {self.battery_size}-kWh battery.")

my_leaf = ElectricCar1('nissan', 'leaf', 2023)
print(my_leaf.get_descriptive_name())
my_leaf.describe_battery()
```

- 组合：将一个类作为另一个类的属性，实现类之间的关联。

```python
class Battery1:
    def __init__(self, battery_size=40):
        self.battery_size = battery_size

    def describe_battery(self):
        print(f"This car has a {self.battery_size}-kWh battery.")

    def get_range(self):
        if self.battery_size == 40:
            range = 150
        elif self.battery_size == 65:
            range = 225
        print(f"This car can go about {range} miles on a full charge.")

class ElectricCar2(Car):
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery = Battery1()  # 组合：将Battery类作为ElectricCar1类的属性

my_leaf2 = ElectricCar2('nissan', 'leaf', 2023)
print(my_leaf2.get_descriptive_name())
my_leaf2.battery.describe_battery()
my_leaf2.battery.get_range()
```

### 9.4 导入类

- 导入单个类：`from module_name import class_name`
- 导入多个类：`from module_name import class_name1, class_name2`
- 导入整个模块：`import module_name`
- 导入所有类：`from module_name import *`

```python
from mod.car import Car as Cr  # 导入单个类并使用别名
from mod.car import ElectricCar as Ec
import mod.car as mc  # 导入整个模块并使用别名

my_new_car = Cr('audi', 'a4', 2024)
print(my_new_car.get_descriptive_name())

my_new_car2 = mc.Car('audi', 'a4', 2024)
print(my_new_car2.get_descriptive_name())

my_leaf = Ec('nissan', 'leaf', 2024)
print(my_leaf.get_descriptive_name())
```

### 9.5 类编程风格

- 类名采用驼峰命名法，即类名中每个单词的首字母大写，并且不使用下划线。
- 实例名和模块名都采用全小写格式，并在单词之间加上下划线。
- 类中用一个空行分隔方法，模块中用两个空行分隔类。
- 先导入标准库模块，再添加一个空行，然后导入自己编写的模块。

```python
# 练习9.1：餐馆
class Restaurant:
    def __init__(self, restaurant_name, cuisine_type):
        self.restaurant_name = restaurant_name.title()
        self.cuisine_type = cuisine_type

    def describe_restaurant(self):
        print(f"{self.restaurant_name} serves wonderful {self.cuisine_type}.")

    def open_restaurant(self):
        print(f"{self.restaurant_name} is now open!")

restaurant = Restaurant("the mean queen", "pizza")
print(restaurant.restaurant_name)
print(restaurant.cuisine_type)
restaurant.describe_restaurant()
restaurant.open_restaurant()
```

```python
# 练习9.3：用户
class User:
    def __init__(self, first_name, last_name, age, city):
        self.first_name = first_name.title()
        self.last_name = last_name.title()
        self.age = age
        self.city = city.title()

    def describe_user(self):
        print(f"{self.first_name} {self.last_name}")
        print(f"  Age: {self.age}")
        print(f"  City: {self.city}")

    def greet_user(self):
        print(f"Hello, {self.first_name}!")

eric = User("eric", "matthes", 30, "seattle")
eric.describe_user()
eric.greet_user()
```

```python
# 练习9.4：就餐人数
print("练习9.4：就餐人数")
class Restaurant:
    def __init__(self, name, cuisine_type):
        self.name = name.title()
        self.cuisine_type = cuisine_type
        self.number_served = 0

    def set_number_served(self, number):
        self.number_served = number

    def increment_number_served(self, number):
        self.number_served += number

restaurant = Restaurant('the restaurant', 'pizza')
print(f"Number served: {restaurant.number_served}")

restaurant.number_served = 100  # 直接修改属性的值
print(f"Number served: {restaurant.number_served}")

restaurant.set_number_served(200)  # 通过方法修改属性的值
print(f"Number served: {restaurant.number_served}")

restaurant.increment_number_served(50)  # 通过方法让属性的值递增
print(f"Number served: {restaurant.number_served}")
```
