## 第10章 文件和异常

### 10.1 读取文件

- 读取文件内容：需要 `pathlib.Path` 对象，主要方法有：
    - 读取整个文件内容：`read()`
    - 读取文件内容的每一行：`readlines()`
    - 读取文件内容并将其转换为字符串：`read_text()`

```python
from pathlib import Path

path = Path('res/pi.txt')
content = path.read_text().rstrip()  # 读取文件内容转换为字符串，并去掉字符串末尾空白符
print(content)

lines = content.splitlines()  # 按行分割字符串
pi_string = ''
for line in lines:
    pi_string += line.lstrip()
print(pi_string)
print(len(pi_string))
```

### 10.2 写入文件

- 写入文件内容：`write_text(content)`方法，将字符串写入文件，**覆盖文件内容**。

```python
from pathlib import Path

path = Path('res/program.txt')
path.write_text("I love programming!")
```

### 10.3 存储数据

- 存储数据：`json.dumps()` 方法将Python对象转为JSON字符串存储，`json.loads()` 方法将JSON字符串转换为Python对象。

```python
from pathlib import Path
import json

# 将列表转换为JSON字符串并存储到文件中
numbers1 = [2, 3, 5, 7, 11, 13]
path = Path('res/numbers.json')
contents1 = json.dumps(numbers1)
path.write_text(contents1)

# 从文件中读取JSON字符串并将其转换为列表
contents2 = path.read_text()
numbers2 = json.loads(contents2)
print(numbers2)
```

```python
from pathlib import Path
import json

path = Path('res/username.json')
if path.exists():
    contents = path.read_text()
    username = json.loads(contents)
    print(f"Welcome back, {username}!")
else:
    username = input("What is your name? ")
    contents = json.dumps(username)
    path.write_text(contents)
    print(f"We'll remember you when you come back, {username}!")
```

```python
# 重构
from pathlib import Path
import json

def get_stored_username():
    path = Path('res/username.json')
    if path.exists():
        contents = path.read_text()
        username = json.loads(contents)
        return username
    else:
        return None

def get_new_username(path):
    username = input("What is your name? ")
    contents = json.dumps(username)
    path.write_text(contents)
    return username

def greet_user():
    username = get_stored_username()
    if username:
        print(f"Welcome back, {username}!")
    else:
        username = get_new_username(path)
        print(f"We'll remember you when you come back, {username}!")

greet_user()
```

### 10.4 异常

- 异常：程序运行时出现的错误，导致程序无法继续执行，使用`try-except-else`语句处理异常。
  - `try`：尝试执行代码块，如果出现异常，跳转到except语句。
  - `except`：处理异常的代码块，可以使用多个except语句处理不同类型的异常，也可以使用 `pass` 语句静默异常。
  - `else`：如果没有异常，执行的代码块。

```python
print("Give me two numbers, and I'll divide them.")
print("Enter 'q' to quit.")

while True:
    firt_number = input("\nFirst number: ")
    if firt_number == 'q':
        break
    second_number = input("Second number: ")
    if second_number == 'q':
        break

    try:
        answer = int(firt_number) / int(second_number)
    except ZeroDivisionError:  # 处理异常
        print("You can't divide by zero!")
    else:
        print(answer)  # 没有异常时执行
```
