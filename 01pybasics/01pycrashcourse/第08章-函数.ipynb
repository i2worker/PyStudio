## 第8章 函数

### 8.1 定义函数

- **函数**：一段带名字的代码块，用关键字 `def` 定义，主要包括函数名、参数列表、函数体、返回值等。
- **参数**：函数定义时指定的变量叫**形参**，函数调用时传递的值叫**实参**。

```python
# 定义函数
def greet_user(username):
    """文档注释：显示简单的问候语"""
    print(f"Hello, {username.title()}!")

# 调用函数
greet_user('jesse')
```

### 8.2 传递参数

- **位置实参**：根据函数定义中的形参顺序，按照顺序传递实参。**注意**：位置实参顺序不能改变。

```python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")

# 位置实参
describe_pet('dog', 'willie')
```

- **关键字实参**：根据函数定义中的**形参名称**和需传递的**值**，按照**名值对**的形式传递实参，顺序可以任意。**注意**：务必准确地指定函数定义中的形参名。

```python
def describe_pet(animal_type, pet_name):
    print(f"I have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")

# 关键字实参
describe_pet(pet_name='willie', animal_type='dog')
```

- **默认值**：在函数定义时可以指定形参的值，当没有提供实参时，使用默认值。**注意**：当使用默认值时，必须在形参顺序中最后指定。

```python
# 默认值
def describe_pet(pet_name, animal_type='dog'):
    print(f"I have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")

describe_pet('willie')
```

- **任意数量位置实参**：通常用 `*args` 表示，将所有位置实参打包成一个**元组**，必须放在所有普通实参后面。

```python
# *args: 任意数量位置实参
def make_pizza(size, *args):
    print(f"Making a {size}-inch pizza with the following toppings:")
    for topping in args:
        print(f"- {topping}")

make_pizza(16, "mushrooms", "green peppers", "extra cheese")
make_pizza(12, "pepperoni", "green peppers", "extra cheese")
```

- **任意数量关键字实参**：通常用 `**kwargs` 表示，将所有关键字实参打包成一个**字典**，必须放在所有普通实参后面。

```python
def build_profile(first, last, **kwargs):
    profile = {'first_name': first, 'last_name': last}
    for key, value in kwargs.items():
        profile[key] = value
    return profile

users = build_profile('albert', 'einstein', location='princeton', field='physics')
print(users)
```

### 8.3 返回值

- 返回值：使用 `return` 语句返回一个值或一组值。

```python
# 返回一个值，并通过默认值让实参可选
def get_formatted_name(first_name, last_name, middle_name=''):
    if middle_name:
        full_name = f"{first_name} {middle_name} {last_name}"
    else:
        full_name = f"{first_name} {last_name}"
    return full_name.title()

musician = get_formatted_name('jimi', 'hendrix')
print(musician)
musician = get_formatted_name('jimi', 'hendrix', 'lee')
print(musician)


# 返回一个字典
def build_person(first_name, last_name, age=None):
    person = { 'first_name': first_name, 'last_name': last_name}
    if age:
        person['age'] = age
    return person

musician = build_person('jimi', 'hendrix', age=27)
print(musician)
```

### 8.4 传递列表

- 如将列表传递给函数，函数就能直接访问列表中的内容。**注意**：函数对列表的任何修改都会影响主程序中的列表，如不希望函数修改，可通过切片操作 `[:]` 将列表复制给函数。

```python
def print_models(unprinted_designs, completed_models):
    while unprinted_designs:
        current_design = unprinted_designs.pop()  # 该操作会修改原列表
        print(f"Printing model: {current_design}")
        completed_models.append(current_design)

def show_completed_models(completed_models):
    print("\nCompleted models:")
    for completed in completed_models:
        print(completed)

unprinted_designs = ['iphone case', 'robot pendant', 'dodecahedron']
completed_models = []
print_models(unprinted_designs, completed_models)
show_completed_models(completed_models)

# print_models(unprinted_designs[:], completed_models)  # 通过切片复制避免修改原列表
# show_completed_models(completed_models)
```

```python
# 练习8.6：城市名
print("练习8.6：城市名")
def city_country(city, country):
    return f"{city.title()}, {country.title()}"

print(city_country('beijing', 'china'))
print(city_country('shanghai', 'china'))
print(city_country('New York', 'USA'))


# 练习8.7：专辑
print("\n练习8.7：专辑")
def make_album(artist, title, tracks=0):
    album = { 'artist': artist.title(), 'title': title.title()}
    if tracks:
        album['tracks'] = tracks
    return album

print(make_album('metallica', 'ride the lightning'))
print(make_album('iron maiden', 'piece of mind', tracks=20))
```

```python
# 练习8.9：消息
print("练习8.9：消息")
def show_messages(messages):
    print("Showing all messages:")
    for message in messages:
        print(message)

messages = ["hello there", "how are u?", ":)"]
show_messages(messages)


# 练习8.10：发送消息
print("\n练习8.10：发送消息")
def send_messages(out_messages, in_messages):
    print("Sending all messages:")
    while out_messages:
        message = out_messages.pop()
        print(message)
        in_messages.append(message)

out_messages = ["hello there", "how are u?", ":)"]
in_messages = []
send_messages(out_messages, in_messages)
show_messages(in_messages)


# 练习8.11：消息归档
print("\n练习8.11：消息归档")
out_messages2 = ["hello there", "how are u?", ":)"]
in_messages2 = []
send_messages(out_messages2[:], in_messages2)
show_messages(out_messages2)
show_messages(in_messages2)
```
