## 第2章　变量和简单的数据类型

### 2.1 变量

**变量是被赋值的标签，是存储数据的名称，在程序中可以随时修改变量值(始终指向最新值)**，命名规则如下：
- 变量名只能包含字母、数字和下划线，不能以数字开头
- 变量名不能使用Python关键字、函数名
- 变量名不能包含空格等特殊字符
- 变量名区分大小写，并且尽量使用小写的变量名

变量可以同时多个被赋值，**变量名和数值之间用逗号分隔**，如：

```python
a, b, c = 1, 2, 3
print(a, b, c)
```

### 2.2 字符串

**字符串是被单引号或双引号括起来的字符序列**，常见的操作如下：
- `title()`：将字符串中的每个单词首字母转为大写
- `upper()`：将字符串中的所有字母都转换为大写
- `lower()`：将字符串中的所有字母都转换为小写
- `rstrip()`：删除字符串末尾的空白字符
- `lstrip()`：删除字符串开头的空白字符
- `strip()`：删除字符串开头和结尾的空白字符
- `removeprefix()`：删除字符串开头的指定前缀
- `removesuffix()`：删除字符串结尾的指定后缀

**f符串**：在字符串中插入变量值的方法，方法如下：

```python
name = "Alice"
age = 18
print(f"我的名字叫{name}，今年{age}岁")
```

### 2.3 数

数包括整数和浮点数，常见的操作如下：
- 数之间可以进行+、-、*、/、%(求余)、//(整除)、**(乘方)运算
- 整数和浮点数之间运算结果为浮点数，**任意两个数相除，结果为浮点数**
- **大数可以用下划线来分隔**，如：123_456_789，输出结果时会去掉下划线

Python中没有内置的常量类型，通常用全大写的变量名来表示常量

<strong style="color: red;">🔔特别注意：浮点数的运算结果，包含的小数位数可能是不确定的，如：0.1 + 0.2 != 0.3</strong>


```python
# 0.30000000000000004
0.1 + 0.2
```

### 2.4 注释

在Python中，注释的方法有如下几种：
- 单行注释：使用 `#` 开头
- 多行注释：使用 `''''` 或 `"""` 开头和结尾
- 文档字符串：使用 `'''` 或 `"""` 开头和结尾

```python
# 练习2.1：简单消息
message = "一条消息赋给变量并打印出来"
print(message)

# 练习2.2：多条简单消息
message = "一条消息赋给变量并打印出来"
print(message)
message = "另一条消息赋给同个变量并打印出来"
print(message)

# 练习2.3：个性化消息
name = "Eric"
print(f"Hello {name}, would you like to learn some Python today?")

# 练习2.4：调整名字的大小写
name = "alice"
print(name.lower())  # 全小写
print(name.upper())  # 全大写
print(name.title())  # 首字母大写

# 练习2.5：名言1
print('Albert Einstein once said, "A person who never made a mistake never tried anything new."')

# 练习2.6：名言2
famous_person = "Albert Einstein"
message = f'{famous_person} once said, "A person who never made a mistake never tried anything new."'
print(message)

# 练习2.7：删除人名中的空白
name = "   Albert Einstein   "
print(name)
print(name.lstrip())  # 删除左端空格
print(name.rstrip())  # 删除右端空格
print(name.strip())   # 删除首尾空格

# 练习2.8：文件扩展名
filename = "python_note.txt"
print(filename)
print(filename.removesuffix(".txt"))

# 练习2.9：数字8
print(2+6)
print(10-2)
print(2*4)
print(16/2)

# 练习2.10：最喜欢的数
number = 8
message = f"最喜欢的数是{number}。"
print(message)

# 练习2.12: Python之禅
import this
```
