## 第6章 字典

### 6.1 什么是字典

- 字典：用花括号 `{}` 包含的一组用冒号 `:` 连接的键值对，键值对之间用逗号 `,` 分割，可以用其中的键来访问值。

```python
alien = {'color': 'green', 'points': 5}
print(alien['color'])

alien = {}
alien['color'] = 'green'
alien['points'] = 5
print(alien)
```

### 6.2 字典操作

- 添加键值对：指定字典名称、方括号包含起来的键名，再赋值给该键对应的值。

```python
alien = {'color': 'green', 'points': 5}
print(alien)
alien['x'] = 0
alien['y'] = 25
print(alien)
```

- 修改字典值：指定字典名称、方括号包含起来的键名，再赋值给该键对应的值。

```python
alien = {'color': 'green', 'points': 5}
print(alien)
alien['color'] = 'yellow'
print(alien)
```

- 删除键值对：通过 `del` 语句并指定键名来删除键值对。

```python
alien = {'color': 'green', 'points': 5}
del alien['points']
print(alien)
```

- 获取字典值：可以通过 `get(key, default)` 方法获取字典值，如果key不存在则返回default值，不指定default时返回None。

```python
alien = {'color': 'green', 'points': 5}
print(alien.get('color', 'No color'))
```

### 6.3 遍历字典

- 遍历键值对：`for key, value in dict.items()`

```python
users = {'username': 'efermi', 'first': 'enrico', 'last': 'fermi'}
for key, value in users.items():
    print(f'Key: {key}')
    print(f'Value: {value}\n')
```

- 遍历所有键：`for key in dict.keys()`
  - 按特定顺序遍历所有键：`for key in sorted(dict.keys())`

```python
languages = {'jen': 'python', 'sarah': 'c', 'edward': 'rust', 'phil': 'python'}
for name in languages.keys():
    print(name.title())
```

- 遍历所有值：`for value in dict.values()`
  - 剔除重复值： `for value in set(dict.values())`

```python
languages = {'jen': 'python', 'sarah': 'c', 'edward': 'rust', 'phil': 'python'}
for language in languages.values():
    print(language.title())
```

```python
# 练习6.7：人们
print("练习6.7：人们")
person1 = {'first_name': 'eric', 'last_name': 'mason', 'age': 25, 'city': 'hangzhou'}
person2 = {'first_name': 'josh', 'last_name': 'smith', 'age': 30, 'city': 'guangzhou'}
person3 = {'first_name': 'mandy', 'last_name': 'li', 'age': 28, 'city': 'beijing'}
peoples = [person1, person2, person3]
for person in peoples:
    name = f"{person['first_name'].title()} {person['last_name'].title()}"
    age = person['age']
    city = person['city'].title()
    print(f"{name} is {age} years old and lives in {city}.")

# 练习6.8：宠物
print("\n练习6.8：宠物")
pet1 = {'type': 'dog', 'name': 'willie', 'owner': 'eric', 'weight': 10}
pet2 = {'type': 'cat', 'name': 'kitty', 'owner': 'josh', 'weight': 5}
pet3 = {'type': 'rabbit', 'name': 'rabbit', 'owner': 'mandy', 'weight': 2}
pets = [pet1, pet2, pet3]
for pet in pets:
    print(f"Here's what I know about {pet['name'].title()}:")
    for key, value in pet.items():
        print(f"  {key.title()}: {value}")

# 练习6.9：喜欢的地方
print("\n练习6.9：喜欢的地方")
favorite_places = {
    'eric': ['hangzhou', 'shanghai', 'guangzhou'],
    'josh': ['wuhan', 'beijing', 'changsha'],
    'mandy': ['dalian', 'shenyang', 'shenzhen'],
}
for name, places in favorite_places.items():
    print(f"{name.title()} likes the following places:")
    for place in places:
        print(f" - {place.title()}")

# 练习6.11：城市
print("\n练习6.11：城市")
cities = {
    'beijing': {'country': 'china', 'population': 21_000_000, 'fact': 'Beijing is the capital of China'},
    'washington': {'country': 'usa', 'population': 8_400_000, 'fact': 'Washington is the capital of the USA'},
    'london': {'country': 'uk', 'population': 89_000_000, 'fact': 'London is the capital of the UK'},
}
for city, info in cities.items():
    print(f"{city.title()} is in {info['country'].title()}. It has a population of {info['population']}. {info['fact']}")
```
