### 15.1 matplotlib

- 绘制折线图：plt.subplots() → ax.plot() → fig.show()

```python
# 屏蔽警告
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

# 导入matplotlib库
%matplotlib inline
import matplotlib.pyplot as plt

inputs = list(range(1, 11))
squares = [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]

# 设置图表样式，可使用 plt.style.available 查看所有可用样式
plt.style.use("seaborn-v0_8")

# 创建一个图或一组子图，fig代表整个画布，ax代表单个子图
fig, ax = plt.subplots()

# 绘制线图，linewidth设置线宽
ax.plot(inputs,squares, linewidth=3)

# 设置图表标题
ax.set_title("Square Numbers", fontsize=24)

# 设置坐标轴标签
ax.set_xlabel("Value", fontsize=14)
ax.set_ylabel("Square of Value", fontsize=14)

# 设置刻度标记的字体大小
ax.tick_params(labelsize=14)

# 显示图表
fig.show()
```

- 绘制散点图：plt.subplots() → ax.scatter() → fig.show()

```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

%matplotlib inline
import matplotlib.pyplot as plt

x = range(1, 1001)
y = [i**2 for i in x]

plt.style.use("seaborn-v0_8")
fig, ax = plt.subplots()

# 绘制散点图，c设置值关联的颜色，cmap设置颜色映射，s设置点的大小
ax.scatter(x, y, c=y, cmap="Blues", edgecolor="none", s=15)

# 设置坐标轴的取值范围
ax.axis((0, 1100, 0, 1100000))

# 设置刻度标记样式，style="plain"表示不使用科学计数法
ax.ticklabel_format(style="plain")
ax.tick_params(labelsize=14)

# 设置图表标题、坐标轴标签
ax.set_title("Square Numbers", fontsize=24)
ax.set_xlabel("Value", fontsize=14)
ax.set_ylabel("Square of Value", fontsize=14)

fig.show()

# 保存图表，bbox_inches="tight"表示去除图表中多余空白区域
# fig.savefig("squares.png", bbox_inches="tight")
```

- 随机漫步：[mod/walk.py](mod/walk.py)

```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

%matplotlib inline
import matplotlib.pyplot as plt

from mod.walk import RandomWalk

# 生成随机漫步点
rw = RandomWalk(50000)
rw.fill_walk()
points = range(rw.points)

plt.style.use("classic")
# figsize设置图像大小，dpi设置图片像素密度
fig, ax = plt.subplots(figsize=(12, 6))

# 绘制随机漫步点，颜色渐变，edgecolor="none"表示点没有边框
ax.scatter(rw.x, rw.y, c=points, cmap="Reds", edgecolor="none", s=1)

# 绘制起点和终点
ax.scatter(0, 0, c="green", edgecolor="none", s=80)
ax.scatter(rw.x[-1], rw.y[-1], c="blue", edgecolor="none", s=80)

# 隐藏坐标轴
ax.get_xaxis().set_visible(False)
ax.get_yaxis().set_visible(False)

fig.show()
```

### 15.2 plotly

- 掷骰子：[mod/die.py](mod/die.py)

```python
import plotly.express as px # pyright: ignore[reportMissingTypeStubs]

from mod.die import Die

# 模拟掷2个骰子
d1 = Die()
d2 = Die(10)
results = [d1.roll() + d2.roll() for _ in range(50000)]

# 分析掷出数字的频率
nums = range(2, d1.sides+d2.sides+1)
freqs = [results.count(value) for value in nums]

# 利用plotly绘制柱状图
title = "Results of Rolling a D6 and a D10 50000 Times"
labels = {'x': 'Result', 'y': 'Frequency of Result'}
fig = px.bar(x=nums, y=freqs, title=title, labels=labels) # 绘制柱状图
fig.update_layout(xaxis_dtick=1) # 设置x轴刻度间隔为1

# 显示柱状图
fig.show()

# 保存柱状图
# fig.write_html('d6_d10.html')
```
